```python
import tkinter as tk
from tkinter import messagebox

# Create window
window = tk.Tk()
window.title("After-School Routine Checker")
window.geometry("400x350")

# Routine tasks
tasks = [
    "Finish homework",
    "Have a snack",
    "Play outside",
    "Read a book",
    "Get ready for bed"
]

# Heading
title = tk.Label(
    window,
    text="After-School Routine",
    font=("Arial", 18, "bold")
)
title.pack(pady=15)

# Entry box
entry = tk.Entry(window, width=30, font=("Arial", 12))
entry.pack(pady=10)

# Label to display last character
character_label = tk.Label(
    window,
    text="Last character typed: ",
    font=("Arial", 12)
)
character_label.pack(pady=10)

# Label to display next task
task_label = tk.Label(
    window,
    text="Next task: " + tasks[0],
    font=("Arial", 12)
)
task_label.pack(pady=10)


# Keyboard event
def key_pressed(event):
    character_label.config(
        text="Last character typed: " + event.char
    )


# Mouse click event
def mouse_clicked(event):
    task_label.config(
        text="Next task: " + tasks[0]
    )


# Button event
def check_routine():
    task = entry.get()

    if task == "":
        messagebox.showwarning(
            "Warning",
            "Please enter a task!"
        )
    else:
        task_label.config(
            text="Next task: " + tasks[0]
        )


# Bind keyboard event
window.bind("<Key>", key_pressed)

# Bind mouse click event
window.bind("<Button-1>", mouse_clicked)

# Button
button = tk.Button(
    window,
    text="Check Routine",
    command=check_routine,
    font=("Arial", 12)
)
button.pack(pady=15)

# Start program
window.mainloop()
```
