```python
from tkinter import *
from tkinter import messagebox

# --- Calculation Function ---
def calculate_schedule():
    try:
        # Get values from input boxes and convert to integers
        total_pages = int(total_pages_entry.get())
        pages_per_day = int(pages_per_day_entry.get())

        # Validation
        if total_pages <= 0 or pages_per_day <= 0:
            messagebox.showerror("Error", "Please enter numbers greater than 0!")
            return

        # Core Calculations
        complete_days = total_pages // pages_per_day  # Floor division for full days
        remaining_pages = total_pages % pages_per_day # Modulo for leftover pages

        # Prepare the result message
        result = f"Complete Days: {complete_days}\n"
        result += f"Leftover Pages for Last Day: {remaining_pages}\n"

        if remaining_pages > 0:
            result += f"Total Days to Finish: {complete_days + 1}"
        else:
            result += f"Total Days to Finish: {complete_days}"

        # Display result on the label
        result_label.config(text=result, fg="green")

    except ValueError:
        messagebox.showerror("Error", "Please enter valid whole numbers!")

# --- Function to Open the Toplevel Planner Window ---
def open_planner():
    global total_pages_entry, pages_per_day_entry, result_label

    # Create top level window
    top = Toplevel(root)
    top.title("Reading Planner")
    top.geometry("300x350")

    # Widgets inside Toplevel window
    Label(top, text="Total Pages in Book:", font=("Arial", 10, "bold")).pack(pady=10)
    total_pages_entry = Entry(top, font=("Arial", 10))
    total_pages_entry.pack()

    Label(top, text="Pages per Day:", font=("Arial", 10, "bold")).pack(pady=10)
    pages_per_day_entry = Entry(top, font=("Arial", 10))
    pages_per_day_entry.pack()

    # Calculate Button
    Button(top, text="Calculate", command=calculate_schedule, bg="#4CAF50", fg="white").pack(pady=15)

    # Result Display Label
    result_label = Label(top, text="Results will show here", font=("Arial", 10, "italic"), fg="gray")
    result_label.pack(pady=10)

# --- Main Window Setup ---
root = Tk()
root.title("Main Window")
root.geometry("350x200")

# Welcome Label
Label(root, text="Reading Schedule Planner", font=("Arial", 14, "bold")).pack(pady=20)

# Button to open the calculation window
Button(root, text="Open Planner", font=("Arial", 11), command=open_planner, bg="#2196F3", fg="white").pack(pady=10)

# Run the application (Will print an explanation if run on headless Google Colab)
try:
    root.mainloop()
except TclError:
    print("Tkinter display not detected. To see the GUI run, copy and execute this code in Python on your local computer.")
```

```python
#denominator calculator
from tkinter import *
from tkinter import messagebox

class DenominationCalculator:
    def __init__(self, root):
        self.root = root
        self.root.geometry("500x400")
        self.root.title("Denomination Calculator")
        self.root.config(bg="#f0f0f0")

        # Title Label
        title_label = Label(root, text="Denomination Calculator",
                           font=("Arial", 18, "bold"), bg="#f0f0f0")
        title_label.pack(pady=20)

        # Input Frame
        input_frame = Frame(root, bg="#f0f0f0")
        input_frame.pack(pady=10)

        Label(input_frame, text="Enter Amount:", font=("Arial", 12), bg="#f0f0f0").pack(side=LEFT, padx=10)

        self.amount_entry = Entry(input_frame, font=("Arial", 12), width=20)
        self.amount_entry.pack(side=LEFT, padx=5)

        # Calculate Button
        calc_button = Button(root, text="Calculate", font=("Arial", 12, "bold"),
                            bg="#4CAF50", fg="white", command=self.calculate, padx=20, pady=10)
        calc_button.pack(pady=15)

        # Result Frame
        result_frame = Frame(root, bg="white", relief=SUNKEN, bd=2)
        result_frame.pack(pady=15, padx=20, fill=BOTH, expand=True)

        Label(result_frame, text="Results:", font=("Arial", 12, "bold"),
              bg="white", justify=LEFT).pack(anchor=NW, padx=10, pady=10)#n,s,w,e

        # Result Display
        self.result_text = Text(result_frame, font=("Arial", 11), height=10, width=50)
        self.result_text.pack(padx=10, pady=10, fill=BOTH, expand=True)
        self.result_text.config(state=DISABLED)

        # Reset Button
        reset_button = Button(root, text="Reset", font=("Arial", 12, "bold"),
                             bg="#f44336", fg="white", command=self.reset, padx=20, pady=10)
        reset_button.pack(pady=10)

    def calculate(self):
        try:
            amount = int(self.amount_entry.get())

            if amount <= 0:
                messagebox.showerror("Error", "Please enter a positive amount!")
                return

            # Calculate denominations
            notes_2000 = amount // 2000
            remaining = amount % 2000
            #** power function 2 raise floor division // whole % reminder
            notes_500 = remaining // 500
            remaining = remaining % 500

            notes_100 = remaining // 100
            remaining = remaining % 100

            # Display results
            self.result_text.config(state=NORMAL)
            self.result_text.delete(1.0, END)

            result = f"Amount Entered: Rs. {amount}\n\n"
            result += f"Number of Rs. 2000 notes: {notes_2000}\n"
            result += f"Number of Rs. 500 notes: {notes_500}\n"
            result += f"Number of Rs. 100 notes: {notes_100}\n"
            result += f"\nTotal Notes Used: {notes_2000 + notes_500 + notes_100}\n"

            if remaining > 0:
                result += f"\nRemaining Amount: Rs. {remaining}\n"
                result += "(Cannot be denominated with 2000, 500, 100)"
            else:
                result += f"\nTotal Amount Covered: Rs. {amount}"

            self.result_text.insert(1.0, result)
            self.result_text.config(state=DISABLED)

        except ValueError:
            messagebox.showerror("Error", "Please enter a valid number!")

    def reset(self):
        self.amount_entry.delete(0, END)
        self.result_text.config(state=NORMAL)
        self.result_text.delete(1.0, END)
        self.result_text.config(state=DISABLED)
        self.amount_entry.focus()

# Create main window
root = Tk()
app = DenominationCalculator(root)
root.mainloop()
```

```python
# calculator
import tkinter as tk
from tkinter import messagebox

# Function to evaluate the expression
def press(key):
    entry_text.set(entry_text.get() + str(key))

def clear():
    entry_text.set("")

def backspace():
    current_text = entry_text.get()# name="nivedita"
    #name[:-1]
    entry_text.set(current_text[:-1])

def evaluate():
    try:
        result = str(eval(entry_text.get()))
        entry_text.set(result)
    except:
        messagebox.showerror("Error", "Invalid Expression")

# Create main window
root = tk.Tk()
root.title("Calculator")
root.geometry("550x650")
root.resizable(False, False)

# Entry widget
entry_text = tk.StringVar()
entry = tk.Entry(root, textvariable=entry_text, font=("Arial", 20), bd=10, relief="ridge", justify="right")
entry.pack(fill="both", ipadx=8, ipady=15, pady=10)

# Button Frame
frame = tk.Frame(root)
frame.pack()

buttons = [
    ("7", 1, 0), ("8", 1, 1), ("9", 1, 2), ("/", 1, 3),
    ("4", 2, 0), ("5", 2, 1), ("6", 2, 2), ("*", 2, 3),
    ("1", 3, 0), ("2", 3, 1), ("3", 3, 2), ("-", 3, 3),
    ("0", 4, 0), (".", 4, 1), ("=", 4, 2), ("+", 4, 3),
]

# Create buttons
for (text, row, col) in buttons:
    if text == "=":
        btn = tk.Button(frame, text=text, width=5, height=2, font=("Arial", 18),
                        command=evaluate, bg="lightgreen")
    else:
        btn = tk.Button(frame, text=text, width=5, height=2, font=("Arial", 18),
                        command=lambda key=text: press(key))
    btn.grid(row=row, column=col, padx=5, pady=5)

# Extra Buttons
tk.Button(frame, text="C", width=5, height=2, font=("Arial", 18), command=clear, bg="tomato").grid(row=5, column=0, padx=5, pady=5)
tk.Button(frame, text="⌫", width=5, height=2, font=("Arial", 18), command=backspace, bg="lightblue").grid(row=5, column=1, padx=5, pady=5)

root.mainloop()
```

```python
# password
from tkinter import *
from tkinter import messagebox
import re

class PasswordStrengthChecker:
    def __init__(self, root):
        self.root = root
        self.root.geometry("500x550")
        self.root.title("Password Strength Checker")
        self.root.config(bg="#f0f0f0")

        # Title Label
        title_label = Label(root, text="Password Strength Checker",
                           font=("Arial", 18, "bold"), bg="#f0f0f0", fg="#333333")
        title_label.pack(pady=20)

        # Password Input Frame
        input_frame = Frame(root, bg="#f0f0f0")
        input_frame.pack(pady=15)

        Label(input_frame, text="Enter Password:", font=("Arial", 12, "bold"),
              bg="#f0f0f0").pack(anchor=W, padx=20)

        self.password_entry = Entry(input_frame, font=("Arial", 12), width=35, show="●")
        self.password_entry.pack(padx=20, pady=5)
        self.password_entry.bind("<KeyRelease>", self.check_strength)

        # Strength Meter Frame
        meter_frame = Frame(root, bg="#f0f0f0")
        meter_frame.pack(pady=20)

        Label(meter_frame, text="Strength Level:", font=("Arial", 11, "bold"),
              bg="#f0f0f0").pack(anchor=W, padx=20)

        self.strength_bar = Frame(meter_frame, bg="#e0e0e0", height=20, width=300)
        self.strength_bar.pack(padx=20, pady=10, fill=X)

        # Strength Label
        self.strength_label = Label(root, text="Strength: Very Weak", font=("Arial", 12, "bold"),
                                   bg="#f0f0f0", fg="#d32f2f")
        self.strength_label.pack(pady=10)

        # Criteria Frame
        criteria_frame = Frame(root, bg="white", relief=SUNKEN, bd=2)
        criteria_frame.pack(pady=15, padx=20, fill=BOTH, expand=True)

        Label(criteria_frame, text="Password Criteria:", font=("Arial", 11, "bold"),
              bg="white").pack(anchor=NW, padx=10, pady=10)

        self.criteria_text = Text(criteria_frame, font=("Arial", 10), height=8, width=50)
        self.criteria_text.pack(padx=10, pady=10, fill=BOTH, expand=True)
        self.criteria_text.config(state=DISABLED)

        # Buttons Frame
        button_frame = Frame(root, bg="#f0f0f0")
        button_frame.pack(pady=15)

        check_button = Button(button_frame, text="Check Strength", font=("Arial", 11, "bold"),
                             bg="#4CAF50", fg="white", command=self.display_strength, padx=15, pady=8)
        check_button.pack(side=LEFT, padx=10)

        reset_button = Button(button_frame, text="Reset", font=("Arial", 11, "bold"),
                             bg="#f44336", fg="white", command=self.reset, padx=15, pady=8)
        reset_button.pack(side=LEFT, padx=10)

    def calculate_strength(self, password):
        """Calculate password strength score"""
        score = 0
        feedback = []

        # Length criteria
        if len(password) >= 8:
            score += 1
            feedback.append("✓ Length is at least 8 characters")
        else:
            feedback.append("✗ Length should be at least 8 characters")

        if len(password) >= 12:
            score += 1
            feedback.append("✓ Length is at least 12 characters")

        if len(password) >= 16:
            score += 1
            feedback.append("✓ Length is 16 or more characters")

        # Uppercase letters
        if re.search(r"[A-Z]", password):
            score += 1
            feedback.append("✓ Contains uppercase letters")
        else:
            feedback.append("✗ Add uppercase letters (A-Z)")

        # Lowercase letters
        if re.search(r"[a-z]", password):
            score += 1
            feedback.append("✓ Contains lowercase letters")
        else:
            feedback.append("✗ Add lowercase letters (a-z)")

        # Numbers
        if re.search(r"\d", password):
            score += 1
            feedback.append("✓ Contains numbers")
        else:
            feedback.append("✗ Add numbers (0-9)")

        # Special characters
        if re.search(r"[!@#$%^&*()_+\-=\[\]{};:'\",.<>?/\\|`~]", password):
            score += 1
            feedback.append("✓ Contains special characters")
        else:
            feedback.append("✗ Add special characters (!@#$%^&*)")

        return score, feedback

    def get_strength_level(self, score):
        """Determine strength level and color"""
        if score <= 1:
            return "Very Weak", "#d32f2f", 10
        elif score == 2:
            return "Weak", "#f57c00", 40
        elif score == 3:
            return "Fair", "#fbc02d", 60
        elif score == 4:
            return "Good", "#7cb342", 75
        elif score == 5:
            return "Strong", "#388e3c", 85
        else:
            return "Very Strong", "#1b5e20", 100

    def check_strength(self, event=None):
        """Real-time password strength checking"""
        password = self.password_entry.get()

        if password:
            score, feedback = self.calculate_strength(password)
            strength, color, percentage = self.get_strength_level(score)

            # Update strength label
            self.strength_label.config(text=f"Strength: {strength}", fg=color)

            # Update strength bar
            bar_width = int(300 * percentage / 100)
            self.strength_bar.config(bg=color)

            # Update criteria
            self.criteria_text.config(state=NORMAL)
            self.criteria_text.delete(1.0, END)
            criteria_display = "\n".join(feedback)
            self.criteria_text.insert(1.0, criteria_display)
            self.criteria_text.config(state=DISABLED)
        else:
            self.strength_label.config(text="Strength: Very Weak", fg="#d32f2f")
            self.strength_bar.config(bg="#e0e0e0")
            self.criteria_text.config(state=NORMAL)
            self.criteria_text.delete(1.0, END)
            self.criteria_text.config(state=DISABLED)

    def display_strength(self):
        """Display full strength report"""
        password = self.password_entry.get()

        if not password:
            messagebox.showwarning("Warning", "Please enter a password!")
            return

        score, feedback = self.calculate_strength(password)
        strength, color, percentage = self.get_strength_level(score)

        report = f"Password Strength Report\n"
        report += "=" * 40 + "\n\n"
        report += f"Overall Strength: {strength}\n"
        report += f"Score: {score}/7\n\n"
        report += "Details:\n"
        report += "\n".join(feedback)

        messagebox.showinfo("Password Strength Report", report)

    def reset(self):
        """Reset all fields"""
        self.password_entry.delete(0, END)
        self.strength_label.config(text="Strength: Very Weak", fg="#d32f2f")
        self.strength_bar.config(bg="#e0e0e0")
        self.criteria_text.config(state=NORMAL)
        self.criteria_text.delete(1.0, END)
        self.criteria_text.config(state=DISABLED)
        self.password_entry.focus()

# Create main window
root = Tk()
app = PasswordStrengthChecker(root)
root.mainloop()
```

```python
import tkinter as tk
import random

def play(choice):
    computer = random.choice(["Rock", "Paper", "Scissors"])

    if choice == computer:
        result = "Draw!"
    elif (choice == "Rock" and computer == "Scissors") or \
         (choice == "Paper" and computer == "Rock") or \
         (choice == "Scissors" and computer == "Paper"):
        result = "You Win!"
    else:
        result = "Computer Wins!"

    result_label.config(
        text=f"You: {choice}\nComputer: {computer}\n{result}"
    )

# GUI
window = tk.Tk()
window.title("Rock Paper Scissors")
window.geometry("350x250")

tk.Label(window, text="Rock Paper Scissors",
         font=("Arial", 18)).pack(pady=15)

tk.Button(window, text="Rock", width=12,
          command=lambda: play("Rock")).pack(pady=5)

tk.Button(window, text="Paper", width=12,
          command=lambda: play("Paper")).pack(pady=5)

tk.Button(window, text="Scissors", width=12,
          command=lambda: play("Scissors")).pack(pady=5)

result_label = tk.Label(window, text="Choose one!",
                        font=("Arial", 12))
result_label.pack(pady=15)

window.mainloop()
```

```python
# Import necessary modules from the tkinter library
from tkinter import *
from tkinter import messagebox

# Define a class for the Denomination Calculator application
class DenominationCalculator:
    # Constructor method for the class, initializes the GUI components
    def __init__(self, root):
        self.root = root  # Store the root Tkinter window
        self.root.geometry("500x400")  # Set the initial size of the window
        self.root.title("Denomination Calculator")  # Set the window title
        self.root.config(bg="#f0f0f0")  # Set the background color of the window

        # Create and pack a title label for the application
        title_label = Label(root, text="Denomination Calculator",
                           font=("Arial", 18, "bold"), bg="#f0f0f0")
        title_label.pack(pady=20)  # Pack the label with vertical padding

        # Create a frame to hold the input widgets
        input_frame = Frame(root, bg="#f0f0f0")
        input_frame.pack(pady=10)  # Pack the input frame with vertical padding

        # Create and pack a label for the amount input field
        Label(input_frame, text="Enter Amount:", font=("Arial", 12), bg="#f0f0f0").pack(side=LEFT, padx=10)

        # Create an entry widget for the user to input the amount
        self.amount_entry = Entry(input_frame, font=("Arial", 12), width=20)
        self.amount_entry.pack(side=LEFT, padx=5)  # Pack the entry field

        # Create a button to trigger the calculation
        calc_button = Button(root, text="Calculate", font=("Arial", 12, "bold"),
                            bg="#4CAF50", fg="white", command=self.calculate, padx=20, pady=10)
        calc_button.pack(pady=15)  # Pack the calculate button

        # Create a frame to display the results
        result_frame = Frame(root, bg="white", relief=SUNKEN, bd=2)
        result_frame.pack(pady=15, padx=20, fill=BOTH, expand=True)

        # Create a label for the results section
        Label(result_frame, text="Results:", font=("Arial", 12, "bold"),
              bg="white", justify=LEFT).pack(anchor=NW, padx=10, pady=10) # Anchor to North-West

        # Create a Text widget to display the calculation results
        self.result_text = Text(result_frame, font=("Arial", 11), height=10, width=50)
        self.result_text.pack(padx=10, pady=10, fill=BOTH, expand=True) # Pack the text widget
        self.result_text.config(state=DISABLED)  # Initially disable editing of the text widget

        # Create a reset button
        reset_button = Button(root, text="Reset", font=("Arial", 12, "bold"),
                             bg="#f44336", fg="white", command=self.reset, padx=20, pady=10)
        reset_button.pack(pady=10)  # Pack the reset button

    # Method to perform the denomination calculation
    def calculate(self):
        try:
            # Get the amount from the entry widget and convert it to an integer
            amount = int(self.amount_entry.get())

            # Check if the entered amount is positive
            if amount <= 0:
                messagebox.showerror("Error", "Please enter a positive amount!") # Show error if not positive
                return # Exit the function

            # Calculate the number of 2000 rupee notes
            notes_2000 = amount // 2000
            remaining = amount % 2000  # Calculate the remaining amount

            # Calculate the number of 500 rupee notes from the remaining amount
            notes_500 = remaining // 500
            remaining = remaining % 500  # Update the remaining amount

            # Calculate the number of 100 rupee notes from the remaining amount
            notes_100 = remaining // 100
            remaining = remaining % 100  # Update the remaining amount

            # Enable the result text widget for updating
            self.result_text.config(state=NORMAL)
            self.result_text.delete(1.0, END)  # Clear any previous content

            # Format the results into a string
            result = f"Amount Entered: Rs. {amount}\n\n"
            result += f"Number of Rs. 2000 notes: {notes_2000}\n"
            result += f"Number of Rs. 500 notes: {notes_500}\n"
            result += f"Number of Rs. 100 notes: {notes_100}\n"
            result += f"\nTotal Notes Used: {notes_2000 + notes_500 + notes_100}\n"

            # Add information about any remaining amount that couldn't be denominated
            if remaining > 0:
                result += f"\nRemaining Amount: Rs. {remaining}\n"
                result += "(Cannot be denominated with 2000, 500, 100)"
            else:
                result += f"\nTotal Amount Covered: Rs. {amount}"

            self.result_text.insert(1.0, result)  # Insert the formatted results into the text widget
            self.result_text.config(state=DISABLED)  # Disable editing again

        except ValueError:
            messagebox.showerror("Error", "Please enter a valid number!")  # Show error for invalid input

    # Method to reset the input and results fields
    def reset(self):
        self.amount_entry.delete(0, END)  # Clear the amount input field
        self.result_text.config(state=NORMAL)  # Enable text widget to clear
        self.result_text.delete(1.0, END)  # Clear the results display
        self.result_text.config(state=DISABLED)  # Disable text widget
        self.amount_entry.focus()  # Set focus back to the amount entry field

# Create the main Tkinter window
root = Tk()
# Create an instance of the DenominationCalculator class, passing the root window
app = DenominationCalculator(root)
# Start the Tkinter event loop, which keeps the window open and responsive
root.mainloop()
```
