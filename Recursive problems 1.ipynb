```python
# 01-digit-flip.py
# Topic: Digit Extraction with % and //, Reversing a Number with Recursion

def flip_number(num):
    # Base case: if num is a single digit (e.g., 0-9), return it as is.
    if num // 10 == 0:
        return num
    # Extract the last digit of the number using the modulo operator.
    last = num % 10
    # Recursively call flip_number with the rest of the digits (num without the last digit).
    # Integer division (//) removes the last digit.
    rest = flip_number(num // 10)
    # Construct the new number: place the 'last' digit at the front.
    # Calculate the power of 10 needed to shift 'last' digit to the leftmost position.
    # len(str(rest)) gives the number of digits in the 'rest' part.
    return last * pow(10, len(str(rest))) + rest

# Inform the user about the function's logic.
input("flipNumber peels last digit with % 10 then recurses on // 10.  Press Enter ")
# Demonstrate the function with example numbers.
print("  flip_number(123) =", flip_number(123))
print("  flip_number(456) =", flip_number(456))

# Get user input for a number to test.
n = int(input("Enter a number (try 789 or 1234): "))
# Ask the user for their guess.
guess = input("What is flip_number(" + str(n) + ")? ")
# Explain the process again before revealing the result.
input("flip_number peels last digit and places it at the front each step.  Press Enter ")
# Display the actual result and compare it with the user's guess.
print("  flip_number(" + str(n) + ") =", flip_number(n), "  your guess:", guess)
```

```python

# -------------------------------------------------
# 1. Extract digits from a number recursively
# -------------------------------------------------

def extract_digits(number):
    # Base case
    if number < 10:
        return [number]

    # Recursive case
    return extract_digits(number // 10) + [number % 10]


# -------------------------------------------------
# 2. Reverse a number recursively
# -------------------------------------------------

def reverse_number(number):
    # Convert the number to a string and use
    # recursive string reversal
    reversed_value = reverse_string(str(number))
    return int(reversed_value)


# -------------------------------------------------
# 3. Reverse a string recursively
# -------------------------------------------------

def reverse_string(text):
    # Base case
    if len(text) <= 1:
        return text

    # Recursive case
    return reverse_string(text[1:]) + text[0]


# -------------------------------------------------
# 4. Check whether a number is a power of 4
# -------------------------------------------------

def is_power_of_four(number):
    # Base cases
    if number == 1:
        return True

    if number <= 0 or number % 4 != 0:
        return False

    # Recursive case
    return is_power_of_four(number // 4)


# -------------------------------------------------
# Main Program
# -------------------------------------------------

print("================================")
print("      FLIP IT WITH RECURSION")
print("================================")

while True:
    print("\nChoose an option:")
    print("1. Extract digits from a number")
    print("2. Reverse a number")
    print("3. Reverse a string")
    print("4. Check if a number is a power of 4")
    print("5. Exit")

    choice = input("\nEnter your choice (1-5): ")

    if choice == "1":
        try:
            number = int(input("Enter a positive number: "))

            if number < 0:
                print("Please enter a positive number.")
            else:
                digits = extract_digits(number)
                print("Digits:", digits)

        except ValueError:
            print("Invalid input. Please enter an integer.")

    elif choice == "2":
        try:
            number = int(input("Enter a positive number: "))

            if number < 0:
                print("Please enter a positive number.")
            else:
                print("Original number:", number)
                print("Reversed number:", reverse_number(number))

        except ValueError:
            print("Invalid input. Please enter an integer.")

    elif choice == "3":
        text = input("Enter a string: ")

        print("Original string:", text)
        print("Reversed string:", reverse_string(text))

    elif choice == "4":
        try:
            number = int(input("Enter a number: "))

            if is_power_of_four(number):
                print(number, "is a power of 4.")
            else:
                print(number, "is NOT a power of 4.")

        except ValueError:
            print("Invalid input. Please enter an integer.")

    elif choice == "5":
        print("\nThank you for using Flip It with Recursion!")
        break

    else:
        print("Invalid choice. Please select a number from 1 to 5.")
```

```python
# 02-flip-name.py
# Topic: Reversing a String with Recursion

def flip_name(s):
    # Base case: if the string has only one character, return it as is.
    if len(s) == 1:
        return s
    # Recursive step: call flip_name on the rest of the string (from the second character onwards),
    # then concatenate the first character of the original string to the end of the result.
    return flip_name(s[1:]) + s[0]

# Inform the user about the function's logic.
input("flip_name recurses on s[1:] then attaches s[0] at the end.  Press Enter ")
# Demonstrate the function with example strings.
print("  flip_name('Maya') =", flip_name('Maya'))
print("  flip_name('Code') =", flip_name('Code'))

# Get user input for a name to test.
name = input("Enter a name (try 'Riya' or 'Dev'): ")
# Ask the user for their guess.
guess = input("What is flip_name('" + name + "')? ")
# Explain the process again before revealing the result.
input("flip_name(s) = flip_name(s[1:]) + s[0]  first character lands last.  Press Enter ")
# Display the actual result and compare it with the user's guess.
print("  flip_name('" + name + "') =", flip_name(name), "  your guess:", guess)
```

```python
# 03-power-of-4.py
# Topic: Checking Powers of 4 with Recursion, Two Stopping Conditions

def is_power4(n):
    # Base case 1: If n is less than or equal to 0, it cannot be a power of 4.
    if n <= 0:
        return False
    # Base case 2: If n is 1, it is a power of 4 (4^0).
    if n == 1:
        return True
    # Recursive step: If n is divisible by 4, divide it by 4 and recurse.
    # This checks if n can be repeatedly divided by 4 until it reaches 1.
    if n % 4 == 0:
        return is_power4(n // 4)
    # If n is not divisible by 4 (and not 1 or <= 0), it is not a power of 4.
    return False

# Inform the user about the function's logic.
input("is_power4 divides by 4 - n==1 returns True  remainder returns False.  Press Enter ")
# Demonstrate the function with example numbers.
print("  is_power4(16) =", is_power4(16))
print("  is_power4(12) =", is_power4(12))

# Get user input for a number to test.
n = int(input("Enter a number (try 64 or 48): "))
# Ask the user for their guess.
guess = input("What is is_power4(" + str(n) + ")? ")
# Explain the stopping conditions again before revealing the result.
input("is_power4 has two stops - n==1 is True  remainder or n<=0 is False.  Press Enter ")
# Display the actual result and compare it with the user's guess.
print("  is_power4(" + str(n) + ") =", is_power4(n), "  your guess:", guess)
```
