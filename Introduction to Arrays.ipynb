```python
# Mean of an array = sum of elements / number of elements
# Define a function to calculate the mean of an array
def arrayMean(arr, arr_size):

    total_sum = 0 # Initialize a variable to store the sum of array elements
    # Iterate through the array to sum all elements
    for i in range(0, arr_size):
        total_sum += arr[i] # Add each element to total_sum

    return float(total_sum/arr_size) # Calculate and return the mean as a float

# Median for a sorted array depends if size is even or odd
# Define a function to calculate the median of an array
def arrayMedian(arr, arr_size):

    # Sort array (Note: sorted(arr) returns a new sorted list, it doesn't sort in-place)
    # To sort in-place, use arr.sort() or assign the result: arr = sorted(arr)
    sorted(arr)

    # Return element for even and odd sized array
    # Check if the array size is odd
    if arr_size % 2 != 0:
        # If odd, return the middle element
        return float(arr[int(arr_size/2)])

    # If the array size is even, return the average of the two middle elements
    return float((arr[int((arr_size-1)/2)] +
                arr[int(arr_size/2)])/2.0)


arr = [1,4,5,2,5,8,5,2,6,8] # Define an example array
arr_size = len(arr) # Get the size (number of elements) of the array

print("Mean =", arrayMean(arr, arr_size)) # Call arrayMean and print the result
print("Median =", arrayMedian(arr, arr_size)) # Call arrayMedian and print the result
```

```python
# Minimum Function
# Define a function to find the minimum element in an array
def minElement(a, size):
    temp = a[0] # Initialize 'temp' with the first element of the array
    # Iterate through the array starting from the second element
    for i in range(1,size):
        temp = min(temp, a[i]) # Compare 'temp' with the current element and update 'temp' if a smaller element is found
    return temp # Return the smallest element found

# Maximum Function
# Define a function to find the maximum element in an array
def maxElement(a, size):
    temp = a[0] # Initialize 'temp' with the first element of the array
    # Iterate through the array starting from the second element
    for i in range(1,size):
        temp = max(temp, a[i]) # Compare 'temp' with the current element and update 'temp' if a larger element is found
    return temp # Return the largest element found


a = [12, 1234, 45, 67, 1] # Define an example array
size = len(a) # Get the size (number of elements) of the array
print ("Minimum element of array:", minElement(a, size)) # Call minElement and print the result
print ("Maximum element of array:", maxElement(a, size)) # Call maxElement and print the result
```

```python
# Function to find the second largest number and print it.
def print2largest(a, a_size):

    # Initialize largest and secondLargest with the smallest possible integer value
    largest = secondLargest = -2147483648
    # Iterate through each element of the array
    for i in range(a_size):

        # If the current element of the array is greater than our current largest number
        if (a[i] > largest):

            # The current largest becomes the second largest
            secondLargest = largest
            # The current element becomes the new largest
            largest = a[i]


        # Else, if current element is less than current largest but greater than second largest
        # and is not equal to the largest (to handle duplicates)
        elif (a[i] > secondLargest and a[i] != largest):
          #a[i] = 40 secondLargest = 30 40 > 30   # True largest = 50 a[i] = 50 50 != 50   # False
          #Array: 10  50  30  40 largest = 50 secondLargest = 10 40 > 10       # True 40 != 50      # True secondLargest = 40
            secondLargest = a[i] # Update secondLargest to the current element


    print(secondLargest) # Print the found second largest number


a = [1,2,3,4,5,6,7,8,9] # Define an example array
a_size = len(a) # Get the size (number of elements) of the array
print2largest(a, a_size) # Call the function to find and print the second largest element
```
