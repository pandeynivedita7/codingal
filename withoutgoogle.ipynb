```python
import requests

def translate_text(text, target_language="hi"):
    url = "https://libretranslate.com/translate"

    payload = {
        "q": text,
        "source": "en",
        "target": target_language,
        "format": "text"
    }

    response = requests.post(url, data=payload)

    if response.status_code == 200:
        translated = response.json()["translatedText"]
        print("Translated:", translated)
        return translated
    else:
        print("Translation failed.")
        return text
```

```python
import requests

def translate_text(text, target_language):
    url = "https://libretranslate.com/translate"

    payload = {
        "q": text,
        "source": "en",
        "target": target_language,
        "format": "text"
    }

    response = requests.post(url, data=payload)

    if response.status_code == 200:
        translated = response.json()["translatedText"]
        print("Translated Text:", translated)
        return translated
    else:
        print("Translation failed.")
        return text

# -------- Take input from the user --------

text = input("Enter text in English: ")

print("\nChoose Target Language:")
print("1. Hindi")
print("2. Tamil")
print("3. Telugu")
print("4. Bengali")
print("5. Marathi")
print("6. Gujarati")
print("7. Malayalam")
print("8. Punjabi")

choice = input("Enter your choice (1-8): ")

language_dict = {
    "1": "hi",
    "2": "ta",
    "3": "te",
    "4": "bn",
    "5": "mr",
    "6": "gu",
    "7": "ml",
    "8": "pa"
}

target_language = language_dict.get(choice, "hi")

# Call the function
translate_text(text, target_language)
```

```python
import argostranslate.translate

def translate_text(text):
    installed_languages = argostranslate.translate.get_installed_languages()

    from_lang = next(filter(lambda x: x.code == "en", installed_languages))
    to_lang = next(filter(lambda x: x.code == "hi", installed_languages))

    translation = from_lang.get_translation(to_lang)
    return translation.translate(text)

# Take input from the user
text = input("Enter text in English: ")

# Translate and display the result
translated = translate_text(text)

print("Translated Text:", translated)
```
