```python
import matplotlib.pyplot as plt
import numpy as np

# Step 1: Create the Month Values
months = np.arange(1, 13)

# Step 2: Create the Savings Values
regular_savings = months * 500
goal_savings = months * 700

# Step 3: Plot Regular Savings Line
plt.plot(months, regular_savings,
         marker='o',
         linestyle='--',
         label='Regular Plan: ₹500 per month')

# Step 4: Plot Goal Savings Line
plt.plot(months, goal_savings,
         marker='D',
         linestyle='-',
         label='Goal Plan: ₹700 per month')

# Step 5: Fill the area between the two lines
plt.fill_between(months, regular_savings, goal_savings,
                 alpha=0.3,
                 label='Savings Difference')

# Step 6: Add chart title and axis labels
plt.title("Monthly Savings Progress")
plt.xlabel("Month")
plt.ylabel("Total Savings (₹)")

# Step 7: Set axis ranges
plt.xlim(1, 12)
plt.ylim(0, 9000)

# Step 8: Add legend
plt.legend()

# Step 9: Display the chart
plt.show()
```
