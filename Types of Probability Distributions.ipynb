## PMF vs. CDF: What Is the Difference?
Before diving into specific distributions, it helps to understand the two ways we calculate their probabilities:

* Probability Mass Function (PMF): Calculates the probability of getting an exact value, written mathematically as $P(X = k)$.
* Cumulative Distribution Function (CDF): Calculates the total probability of getting a value less than or equal to a specific amount, written as $P(X \leq k)$.

------------------------------
## 1. Binomial Distribution (Focus on CDF)
The Binomial Distribution tracks the number of successes across a fixed number of independent trials, where each trial has only two outcomes: success or failure.

* Core Example: Tossing a coin 10 times to find the chance of getting heads.
* Key Condition: The total number of trials ($n$) is fixed and known beforehand.
* Cumulative Distribution Function (CDF): Finds the likelihood of obtaining up to a certain number of successes. For instance, finding the probability of getting 3 or fewer heads out of 10 coin flips.

------------------------------
## 2. Poisson Distribution (Overview, PMF, and CDF)
The Poisson Distribution models the number of times an event occurs within a specified, continuous interval like time, distance, or area.

* Core Example: Counting the number of raindrops hitting a square meter of ground in one minute.
* Key Condition: The events must occur independently and at a constant average rate ($\lambda$).
* Probability Mass Function (PMF): Computes the chance of an exact number of occurrences. For example, calculating the exact probability that exactly 4 customer calls arrive at a support desk in the next hour.
* Cumulative Distribution Function (CDF): Accumulates the probabilities up to a threshold. For example, determining the probability that at most 4 customer calls arrive during that hour (meaning 0, 1, 2, 3, or 4 calls).

------------------------------
## Summary Checklist

```python
import scipy.stats as stats # Import the scipy.stats module, which contains functions for various probability distributions.

# calculate the probability
# stats.binom.cdf(k, n, p) calculates the cumulative distribution function (CDF) for a binomial distribution.
# k: The number of successes (in this case, 6 heads).
# n: The total number of trials (in this case, 10 coin flips).
# p: The probability of success on a single trial (in this case, 0.5 for getting a head).
# 1 - stats.binom.cdf(6, 10, 0.5) calculates the probability of getting *more than* 6 heads.
# This is because the CDF gives P(X <= k), so 1 - P(X <= k) gives P(X > k).
prob = 1 - stats.binom.cdf(6, 10, 0.5)

print("the probability of getting more than 6 heads in 10 coin flips is :", prob) # Print the calculated probability.
```

```python
import scipy.stats as stats # Import the scipy.stats module, which contains functions for various probability distributions.

# Calculate the probability using Poisson Probability Mass Function (PMF)
# stats.poisson.pmf(k, mu) calculates the probability of exactly 'k' events occurring,
# given an average rate of 'mu' events in the interval.
# Here, k=6 (6 days of rain) and mu=10 (expected value).
prob1 = stats.poisson.pmf(6, 10)
print("probability of raining for exactly 6 days :", prob1)

# Calculate the probability of observing 12, 13, or 14 days of rain.
# This is done by summing the individual PMF values for each of these outcomes.
prob2 = stats.poisson.pmf(12, 10) + stats.poisson.pmf(13, 10) + stats.poisson.pmf(14, 10)
print("probability of raining for 12-14 days :", prob2)
```

```python
import scipy.stats as stats # Import the scipy.stats module, which contains functions for various probability distributions.

## Checkpoint 1
# Calculate the probability of more than 20 events using Poisson Cumulative Distribution Function (CDF).
# stats.poisson.cdf(k, mu) calculates the probability of at most 'k' events.
# Here, k=20 (at most 20 events) and mu=15 (expected value).
# 1 - stats.poisson.cdf(20, 15) gives the probability of *more than* 20 events (P(X > 20)).
prob1 = 1 - stats.poisson.cdf(20, 15)

# Print the calculated probability.
print(prob1)

## Checkpoint 2
# Calculate the probability of events occurring between 17 and 21 (inclusive).
# This is calculated as P(X <= 21) - P(X <= 16).
# P(X <= 21) is stats.poisson.cdf(21, 15).
# P(X <= 16) is stats.poisson.cdf(16, 15).
prob2 = stats.poisson.cdf(21, 15) - stats.poisson.cdf(16, 15)

# Print the calculated probability.
print(prob2)
```
