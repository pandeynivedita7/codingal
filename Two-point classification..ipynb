```python
# Commented out IPython magic to ensure Python compatibility.
# Importing required libraries

import pandas as pd # Import pandas for data manipulation
from matplotlib import pyplot as plt # Import matplotlib for plotting
from google.colab import files # Import files to handle file uploads in Colab
# %matplotlib inline # This magic command makes matplotlib plots appear inline in the notebook

# Importing Dataset required

# This line allows you to upload files from your local system to the Colab environment.
uploaded = files.upload()
# Read the uploaded CSV file into a pandas DataFrame.
df = pd.read_csv("insurance_data.csv")
# Display the first few rows of the DataFrame to inspect the data.
df.head()

# Plotting the dataset

# Create a scatter plot to visualize the relationship between 'age' and 'bought_insurance'.
# 'age' is on the x-axis, 'bought_insurance' is on the y-axis.
# 'marker='+' specifies plus signs as markers, and 'color='red' sets the color to red.
plt.scatter(df.age,df.bought_insurance,marker='+',color='red')

# Splitting the training Values using sklearn

from sklearn.model_selection import train_test_split # Import train_test_split for splitting data

# Defining the Training and Testing Variables

# Split the DataFrame into training and testing sets.
# X will contain the 'age' feature (input), and y will contain 'bought_insurance' (target).
# train_size=0.8 means 80% of the data will be used for training, 20% for testing.
X_train, X_test, y_train, y_test = train_test_split(df[['age']],df.bought_insurance,train_size=0.8)

# Display the test features (ages) to verify the split.
X_test

# Using Logistic Regression Algorithm

from sklearn.linear_model import LogisticRegression # Import LogisticRegression model
model = LogisticRegression() # Initialize the Logistic Regression model

# Fitting the model

# Train the logistic regression model using the training data.
model.fit(X_train, y_train)

# Display the test features again (this line is redundant but was in original code).
X_test

# Predicting the Y

# Use the trained model to make predictions on the test features.
y_predicted = model.predict(X_test)

# Predict probabilities for each class (0: not bought, 1: bought) for the test set.
model.predict_proba(X_test)

# Testing model

# Evaluate the model's accuracy on the test set.
model.score(X_test,y_test)

# Display the predicted values for the test set.
y_predicted

# Display the test features again (this line is redundant but was in original code).
X_test

# Display the coefficients of the logistic regression model (slope).
model.coef_

# Display the intercept of the logistic regression model.
model.intercept_

import math # Import the math module for mathematical functions
def sigmoid(x):
  # Define the sigmoid activation function, which squashes values between 0 and 1.
  return 1 / (1 + math.exp(-x))

def prediction_function(age):
    # This custom prediction function uses the learned coefficients and intercept
    # from the logistic regression model (approximated from model.coef_ and model.intercept_).
    # It calculates 'z' (the linear combination) and then applies the sigmoid function.
    z = 0.042 * age - 1.53 # 0.04150133 ~ 0.042 and -1.52726963 ~ -1.53
    y = sigmoid(z) # Apply sigmoid to get a probability
    return y

# Test the custom prediction function with an age of 35.
age = 35
prediction_function(age)

# Test the custom prediction function with an age of 43.
age = 43
prediction_function(age)

# This comment explains the interpretation of the prediction for age 43.
# 0.485 is more than 0.5 which means person with 43 will buy the insurance
```

```python
import numpy as np

# Generate a range of ages for plotting the sigmoid curve
age_range = np.linspace(df['age'].min(), df['age'].max(), 100)

# Calculate probabilities for the age_range using the model's coefficients
# The model.coef_ gives the coefficient for 'age' and model.intercept_ is the bias
z_plot = age_range * model.coef_[0][0] + model.intercept_[0]
probabilities = 1 / (1 + np.exp(-z_plot))

# Plot the original scatter data
plt.figure(figsize=(10, 6))
plt.scatter(df.age, df.bought_insurance, marker='+', color='red', label='Actual Data')

# Plot the sigmoid curve
plt.plot(age_range, probabilities, color='blue', label='Logistic Regression Sigmoid Curve')

plt.xlabel('Age')
plt.ylabel('Bought Insurance (1=Yes, 0=No) / Probability')
plt.title('Age vs. Insurance Purchase Probability with Logistic Regression')
plt.legend()
plt.grid(True)
plt.show()
```
