```python
import textblob # Import the TextBlob library for sentiment analysis
from textblob import TextBlob # Specifically import the TextBlob class
import colorama # Import the Colorama library for colored console output
from colorama import Fore, Style # Specifically import Fore and Style classes for text coloring

# Install TextBlob and Colorama if not already installed (this is a shell command)
!pip install textblob colorama

# Print a welcome message in cyan color, then reset to default style
print(f"{Fore.CYAN}🔍 Welcome to Sentiment Spy! {Style.RESET_ALL}")

# Prompt the user to enter their name in magenta color
user_name = input(f"{Fore.MAGENTA}Please enter your name: {Style.RESET_ALL} ").strip() # Read user input and remove leading/trailing whitespace

# If the user doesn't provide a name, set a default name
if not user_name:
    user_name = "Mystery Agent"  # Fallback if user doesn't provide a name

# Initialize an empty list to store conversation history.
# Each entry will be a tuple: (text, polarity, sentiment_type)
conversation_history = [] # []list tuple ()

# Greet the user with their name and provide instructions
print(f"\n{Fore.CYAN}Hello, Agent {user_name}!{Style.RESET_ALL}")
print(f"Type a sentence and I will analyze your sentences with TextBlob and show you the sentiment. 🕵️‍♂️")
print(f"Type '{Fore.YELLOW}reset{Fore.CYAN}', '{Fore.YELLOW}history{Fore.CYAN}', " # Instruct on available commands: reset, history, exit
      f"or '{Fore.YELLOW}exit{Fore.CYAN}' to quit.{Style.RESET_ALL}\n")

# Start an infinite loop to continuously get user input and analyze sentiment
while True: #while loop
    # Prompt the user for input in green color
    user_input = input(f"{Fore.GREEN}> {Style.RESET_ALL}").strip()

    # If the user input is empty, ask for valid input and continue the loop
    if not user_input:
        print(f"{Fore.RED}Please enter some text or a valid command.{Style.RESET_ALL}")
        continue

    # Check for commands (exit, reset, history) in a case-insensitive manner
    if user_input.lower() == "exit":
        # Print an exit message and break out of the loop
        print(f"{Fore.BLUE}👋 Exiting Sentiment Spy. Farewell, Agent {user_name}! {Style.RESET_ALL}")
        break

    elif user_input.lower() == "reset":
        # Clear the conversation history list
        conversation_history.clear()
        print(f"{Fore.CYAN}🔄 All conversation history cleared!{Style.RESET_ALL}")
        continue # Continue to the next iteration of the loop

    elif user_input.lower() == "history":
        # If history is empty, inform the user
        if not conversation_history:
            print(f"{Fore.YELLOW}No conversation history yet.{Style.RESET_ALL}")
        else:
            # Print the conversation history with sentiment details
            print(f"{Fore.CYAN}📜 Conversation History:{Style.RESET_ALL}")
            # Iterate through history, assigning a number to each entry
            for idx, (text, polarity, sentiment_type) in enumerate(conversation_history, start=1):
                # Choose color and emoji based on the sentiment type
                if sentiment_type == "Positive":
                    color = Fore.GREEN
                    emoji = "😊"
                elif sentiment_type == "Negative":
                    color = Fore.RED
                    emoji = "😞"
                else:
                    color = Fore.YELLOW
                    emoji = "😐"

                # Print each history entry with its sentiment, polarity, color, and emoji
                print(f"{idx}. {color}{emoji} {text} "
                      f"(Polarity: {polarity:.2f}, {sentiment_type}){Style.RESET_ALL}")
        continue # Continue to the next iteration of the loop

    # If the input is not a command, perform sentiment analysis
    # Get the sentiment polarity from TextBlob (a float between -1.0 and 1.0)
    polarity = TextBlob(user_input).sentiment.polarity

    # Determine the sentiment type (Positive, Negative, Neutral) based on polarity thresholds
    if polarity > 0.25:
        sentiment_type = "Positive"
        color = Fore.GREEN
        emoji = "😊"
    elif polarity < -0.25:
        sentiment_type = "Negative"
        color = Fore.RED
        emoji = "😞"
    else:
        sentiment_type = "Neutral"
        color = Fore.YELLOW
        emoji = "😐"

    # Save the current user input, polarity, and sentiment type to conversation history
    conversation_history.append((user_input, polarity, sentiment_type)) #[] add things list

    # Print the sentiment analysis result with appropriate color and emoji
    print(f"{color}{emoji} Sentiment: {sentiment_type} "
          f"(Polarity: {polarity:.2f}){Style.RESET_ALL}")
```
