```python
# ✅ Install required packages in Colab (run this once)
!pip install pandas scikit-learn textblob

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from textblob import TextBlob
import time

# Load and preprocess the dataset
def load_data(file_path='imdb_top_1000.csv'):
    try:
        df = pd.read_csv(file_path)
        df['combined_features'] = df['Genre'].fillna('') + ' ' + df['Overview'].fillna('')
        return df
    except FileNotFoundError:
        print(f"❌ Error: The file '{file_path}' was not found. Please upload it to Colab.")
        return None

movies_df = load_data()

# Vectorize the combined features and compute cosine similarity
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(movies_df['combined_features'])
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

# List all unique genres
def list_genres(df):
    return sorted(set(
        genre.strip()
        for sublist in df['Genre'].dropna().str.split(', ')
        for genre in sublist
    ))

genres = list_genres(movies_df)

# Recommend movies
def recommend_movies(genre=None, mood=None, rating=None, top_n=5):
    filtered_df = movies_df
    if genre:
        filtered_df = filtered_df[filtered_df['Genre'].str.contains(genre, case=False, na=False)]
    if rating:
        filtered_df = filtered_df[filtered_df['IMDB_Rating'] >= rating]

    filtered_df = filtered_df.sample(frac=1).reset_index(drop=True)

    recommendations = []
    for _, row in filtered_df.iterrows():
        overview = row['Overview']
        if pd.isna(overview):
            continue
        polarity = TextBlob(overview).sentiment.polarity
        if (mood and ((TextBlob(mood).sentiment.polarity < 0 and polarity > 0) or polarity >= 0)) or not mood:
            recommendations.append((row['Series_Title'], polarity))
        if len(recommendations) == top_n:
            break

    return recommendations if recommendations else "⚠️ No suitable movie recommendations found."

# Display recommendations
def display_recommendations(recs, name):
    print(f"\n🍿 AI-Analyzed Movie Recommendations for {name}:")
    for idx, (title, polarity) in enumerate(recs, 1):
        sentiment = "Positive 😊" if polarity > 0 else "Negative 😞" if polarity < 0 else "Neutral 😐"
        print(f"{idx}. 🎥 {title} (Polarity: {polarity:.2f}, {sentiment})")

# Main flow (with preset values instead of input)
def handle_ai(name="Colab User", genre_choice="Drama", mood="happy", rating=8.0):
    print("🎥 Welcome to your Personal Movie Recommendation Assistant! 🎥\n")

    print("🔍 Let's find the perfect movie for you!\n")
    print("Available Genres (first 15 shown):", ", ".join(genres[:15]), "...\n")

    print(f"👉 Selected Genre: {genre_choice}")
    print(f"👉 Mood: {mood}")
    print(f"👉 Min Rating: {rating}\n")

    print("⏳ Analyzing mood...")
    time.sleep(1)
    polarity = TextBlob(mood).sentiment.polarity
    mood_desc = "positive 😊" if polarity > 0 else "negative 😞" if polarity < 0 else "neutral 😐"
    print(f"Your mood is {mood_desc} (Polarity: {polarity:.2f}).\n")

    print("⏳ Finding movies...")
    time.sleep(1)

    recs = recommend_movies(genre=genre_choice, mood=mood, rating=rating, top_n=5)
    if isinstance(recs, str):
        print(recs)
    else:
        display_recommendations(recs, name)

# ✅ Run with your chosen parameters
handle_ai(name="Swapnil", genre_choice="Action", mood="excited", rating=8.0)
```

```python
"""
movie_recommender.py

Requirements:
    pip install pandas scikit-learn nltk textblob

Notes:
- Dataset CSV should contain columns: 'title', 'genres', 'overview', 'imdb_rating'
  - 'genres' can be a comma- or pipe-separated string (e.g., "Drama,Romance" or "Drama|Romance")
- If you don't provide a dataset, a small demo dataset will be used.

Author: Generated by assistant
"""

import os
import sys
import random
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sentiment: try TextBlob, fallback to NLTK VADER
def try_import_sentiment():
    try:
        from textblob import TextBlob
        def tb_sentiment(text):
            try:
                s = TextBlob(str(text)).sentiment.polarity
            except Exception:
                s = 0.0
            if s > 0.05:
                return "Positive"
            elif s < -0.05:
                return "Negative"
            else:
                return "Neutral"
        return tb_sentiment
    except Exception:
        try:
            import nltk
            from nltk.sentiment import SentimentIntensityAnalyzer
            # Ensure lexicon is downloaded
            try:
                _ = SentimentIntensityAnalyzer()
            except Exception:
                nltk.download('vader_lexicon')
            sia = SentimentIntensityAnalyzer()
            def vader_sentiment(text):
                text = str(text)
                score = sia.polarity_scores(text)['compound']
                if score >= 0.05:
                    return "Positive"
                elif score <= -0.05:
                    return "Negative"
                else:
                    return "Neutral"
            return vader_sentiment
        except Exception:
            # Last fallback: crude rule-based neutral
            def neutral_sentiment(text):
                return "Neutral"
            return neutral_sentiment

get_sentiment = try_import_sentiment()

def load_dataset(path=None):
    if path and os.path.exists(path):
        df = pd.read_csv(path)
        # normalize expected columns
        expected = ['title', 'genres', 'overview', 'imdb_rating']
        # Try common alternative column names
        col_map = {}
        cols = [c.lower() for c in df.columns]
        for e in expected:
            if e in df.columns:
                col_map[e] = e
            else:
                # search case-insensitive
                for c in df.columns:
                    if c.lower() == e:
                        col_map[e] = c
                        break
        # If any missing, attempt to detect alternatives
        if 'title' not in col_map:
            for c in df.columns:
                if 'title' in c.lower():
                    col_map['title'] = c; break
        if 'genres' not in col_map:
            for c in df.columns:
                if 'genre' in c.lower():
                    col_map['genres'] = c; break
        if 'overview' not in col_map:
            for c in df.columns:
                if 'overview' in c.lower() or 'description' in c.lower() or 'plot' in c.lower():
                    col_map['overview'] = c; break
        if 'imdb_rating' not in col_map:
            for c in df.columns:
                if 'rating' in c.lower() or 'vote_average' in c.lower():
                    col_map['imdb_rating'] = c; break

        # Rename discovered columns to expected
        rename = {col_map[k]: k for k in col_map if col_map.get(k) and col_map[k] != k}
        if rename:
            df = df.rename(columns=rename)

        # Ensure expected columns present; fill missing
        for e in expected:
            if e not in df.columns:
                df[e] = np.nan

        # Clean and coerce rating
        def safe_float(x):
            try:
                return float(x)
            except Exception:
                return np.nan
        df['imdb_rating'] = df['imdb_rating'].apply(safe_float)

        return df[expected].copy()
    else:
        # Demo small dataset
        data = [
            {"title":"The Shawshank Redemption", "genres":"Drama|Crime", "overview":"Two imprisoned men bond over a number of years, finding solace and eventual redemption through acts of common decency.", "imdb_rating":9.3},
            {"title":"The Dark Knight", "genres":"Action|Crime|Drama", "overview":"When the menace known as the Joker emerges, Batman must accept one of the greatest psychological and physical tests of his ability to fight injustice.", "imdb_rating":9.0},
            {"title":"Forrest Gump", "genres":"Comedy|Drama|Romance", "overview":"The presidencies of Kennedy and Johnson, the Vietnam War, the Watergate scandal and other historical events unfold from the perspective of an Alabama man with an IQ of 75.", "imdb_rating":8.8},
            {"title":"Toy Story", "genres":"Animation|Adventure|Comedy", "overview":"A cowboy doll is profoundly threatened and jealous when a new spaceman figure supplants him as top toy in a boy's room.", "imdb_rating":8.3},
            {"title":"Parasite", "genres":"Comedy|Drama|Thriller", "overview":"Greed and class discrimination threaten the newly formed symbiotic relationship between the wealthy Park family and the destitute Kim clan.", "imdb_rating":8.6},
        ]
        return pd.DataFrame(data)

def preprocess(df):
    # Ensure string types
    df['title'] = df['title'].fillna('').astype(str)
    df['overview'] = df['overview'].fillna('').astype(str)
    df['genres'] = df['genres'].fillna('').astype(str)

    # Normalize genres into comma-separated
    def normalize_genres(s):
        if not s: return ''
        if '|' in s:
            parts = [p.strip() for p in s.split('|') if p.strip()]
        elif ',' in s:
            parts = [p.strip() for p in s.split(',') if p.strip()]
        else:
            parts = [p.strip() for p in s.split() if p.strip()]
        return ', '.join(parts)
    df['genres_norm'] = df['genres'].apply(normalize_genres)

    # Combined field for TF-IDF
    df['combined'] = (df['overview'] + ' ' + df['genres_norm']).str.lower()

    # Sentiment column
    df['sentiment'] = df['overview'].apply(get_sentiment)

    # fill missing imdb_rating with median
    if 'imdb_rating' in df.columns:
        try:
            df['imdb_rating'] = pd.to_numeric(df['imdb_rating'], errors='coerce')
            median = df['imdb_rating'].median()
            df['imdb_rating'] = df['imdb_rating'].fillna(median if not np.isnan(median) else 0.0)
        except Exception:
            df['imdb_rating'] = df['imdb_rating'].fillna(0.0)
    else:
        df['imdb_rating'] = 0.0

    df = df.reset_index(drop=True)
    return df

class Recommender:
    def __init__(self, df):
        self.df = df.copy()
        self.tfidf = TfidfVectorizer(stop_words='english', max_features=20000)
        self.matrix = self.tfidf.fit_transform(self.df['combined'].values)
        self.cosine_sim = cosine_similarity(self.matrix, self.matrix, dense_output=False)

    def recommend_by_title(self, title, topn=5):
        # find best match ignoring case
        matches = self.df[self.df['title'].str.lower() == title.strip().lower()]
        if matches.empty:
            # try partial match
            matches = self.df[self.df['title'].str.lower().str.contains(title.strip().lower())]
            if matches.empty:
                return pd.DataFrame()  # no match
        idx = matches.index[0]
        sim_scores = list(enumerate(self.cosine_sim[idx].toarray()[0]))
        sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
        top = [i for i,score in sim_scores[1:topn+1]]
        return self.df.loc[top, ['title','genres_norm','imdb_rating','sentiment']]

    def recommend_by_genre(self, genre, limit=10):
        if not genre: return pd.DataFrame()
        g = genre.strip().lower()
        mask = self.df['genres_norm'].str.lower().str.contains(g)
        return self.df[mask].sort_values('imdb_rating', ascending=False).head(limit)[['title','genres_norm','imdb_rating','sentiment']]

    def recommend_by_mood(self, mood, limit=10):
        mood_map = {
            "happy": ["comedy","family","animation"],
            "sad": ["drama"],
            "thriller": ["thriller","mystery","crime"],
            "romantic": ["romance","drama"],
            "action": ["action","adventure"],
            "funny": ["comedy"],
            "scary": ["horror","thriller"],
            "feelgood": ["family","comedy","drama"]
        }
        genres = mood_map.get(mood.lower(), [])
        if not genres:
            # try using mood as genre directly
            genres = [mood.lower()]
        pattern = '|'.join(genres)
        mask = self.df['genres_norm'].str.lower().str.contains(pattern)
        return self.df[mask].sort_values('imdb_rating', ascending=False).head(limit)[['title','genres_norm','imdb_rating','sentiment']]

    def recommend_by_rating(self, min_rating=7.0, limit=10):
        try:
            min_r = float(min_rating)
        except Exception:
            min_r = 0.0
        mask = self.df['imdb_rating'] >= min_r
        return self.df[mask].sort_values('imdb_rating', ascending=False).head(limit)[['title','genres_norm','imdb_rating','sentiment']]

    def random_movie(self):
        return self.df.sample(1)[['title','genres_norm','imdb_rating','sentiment']]

    def random_filtered(self, genre=None, min_rating=None):
        temp = self.df
        if genre:
            temp = temp[temp['genres_norm'].str.lower().str.contains(genre.strip().lower())]
        if min_rating is not None:
            try:
                mr = float(min_rating)
            except Exception:
                mr = None
            if mr is not None:
                temp = temp[temp['imdb_rating'] >= mr]
        if temp.empty:
            return pd.DataFrame()
        return temp.sample(1)[['title','genres_norm','imdb_rating','sentiment']]

def print_results(df):
    if df is None or df.empty:
        print("No results found.")
        return
    for _, row in df.iterrows():
        print(f"Title: {row['title']}")
        print(f"Genre: {row.get('genres_norm','')}")
        print(f"IMDb Rating: {row.get('imdb_rating','')}")
        print(f"Sentiment of Overview: {row.get('sentiment','')}")
        print("-" * 40)

def main():
    print("=== Movie Recommendation System ===")
    path = input("Enter CSV dataset path (press Enter to use demo dataset): ").strip()
    df = load_dataset(path if path else None)
    df = preprocess(df)
    rec = Recommender(df)

    menu = """
Choose recommendation type:
 1. AI-based: recommend by movie title (content-similarity)
 2. By Genre
 3. By Mood
 4. By IMDb minimum rating
 5. Pure Random
 6. Random within filters (genre and/or min rating)
 7. Exit
Enter choice [1-7]: """
    while True:
        choice = input(menu).strip()
        if choice == '1':
            title = input("Enter movie title (or part of it): ").strip()
            result = rec.recommend_by_title(title, topn=5)
            print_results(result)
        elif choice == '2':
            genre = input("Enter genre (e.g., Drama, Comedy): ").strip()
            result = rec.recommend_by_genre(genre, limit=10)
            print_results(result)
        elif choice == '3':
            mood = input("Enter mood (happy, sad, thriller, romantic, action, funny, scary, feelgood): ").strip()
            result = rec.recommend_by_mood(mood, limit=10)
            print_results(result)
        elif choice == '4':
            r = input("Enter minimum IMDb rating (e.g., 7.5): ").strip()
            result = rec.recommend_by_rating(min_rating=r, limit=10)
            print_results(result)
        elif choice == '5':
            result = rec.random_movie()
            print_results(result)
        elif choice == '6':
            g = input("Genre filter (press Enter to skip): ").strip()
            mr = input("Minimum rating filter (press Enter to skip): ").strip()
            mr_val = float(mr) if mr else None
            result = rec.random_filtered(genre=g if g else None, min_rating=mr_val)
            print_results(result)
        elif choice == '7':
            print("Exiting.")
            break
        else:
            print("Invalid choice. Enter 1-7.")

if __name__ == "__main__":
    main()
```
