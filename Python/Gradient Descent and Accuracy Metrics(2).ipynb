```python
from numpy import asarray
from numpy import arange
from numpy.random import rand
from matplotlib import pyplot

# Define the objective function to be minimized. This is a simple quadratic function (x^2).
def objective(x):
    return x**2.0

# Define the derivative of the objective function. This is required for gradient descent.
# The derivative of x^2 is 2x.
def derivative(x):
    return x * 2.0

# Implement the gradient descent algorithm.
# objective: The function to minimize.
# derivative: The derivative of the objective function.
# bounds: The search space (e.g., [-1.0, 1.0]) where the initial solution is chosen.
# n_iter: The number of iterations (steps) to perform.
# step_size: The learning rate, controlling how large each step is.
def gradient_descent(objective, derivative, bounds, n_iter, step_size):
    # Initialize lists to store the solution (x-values) and their corresponding scores (y-values) at each iteration.
    solutions, scores = list(), list()
    # Generate a random initial point (solution) within the defined bounds.
    solution = bounds[:, 0] + rand(len(bounds)) * (bounds[:, 1] - bounds[:, 0])

    # Loop through the specified number of iterations.
    for i in range(n_iter):
        # Calculate the gradient (slope) of the objective function at the current solution.
        gradient = derivative(solution)
        # Update the solution by moving in the opposite direction of the gradient.
        # The step size determines how much we move in that direction.
        solution = solution - step_size * gradient
        # Evaluate the objective function at the new candidate solution.
        solution_eval = objective(solution)
        # Store the new solution and its evaluated score.
        solutions.append(solution)
        scores.append(solution_eval)
        # Print the progress: iteration number, current solution, and its objective function value.
        print('>%d f(%s) = %.5f' % (i, solution, solution_eval))
    # Return all the solutions and their scores found during the descent.
    return [solutions, scores]

# Define the search space for the input variable (x).
# Here, x will be searched between -1.0 and 1.0.
bounds = asarray([[-1.0, 1.0]])
# Set the total number of iterations for the gradient descent algorithm.
n_iter = 30
# Set the learning rate (step size) for updating the solution.
step_size = 0.1

# Perform the gradient descent search using the defined objective, derivative, bounds, iterations, and step size.
solutions, scores = gradient_descent(objective, derivative, bounds, n_iter, step_size)

# Generate a range of input values (x-axis) to plot the objective function.
# This creates points from the lower bound to the upper bound with a 0.1 increment.
inputs = arange(bounds[0,0], bounds[0,1]+0.1, 0.1)
# Compute the corresponding objective function values (y-axis) for the generated input values.
results = objective(inputs)

# Create a line plot of the objective function (inputs vs results).
pyplot.plot(inputs, results, label='Objective Function')
# Plot the path taken by the gradient descent algorithm.
# '.-' creates a line with markers, and 'red' sets the color.
pyplot.plot(solutions, scores, '.-', color='red', label='Gradient Descent Path')

# Add labels and title for clarity.
pyplot.xlabel('x')
pyplot.ylabel('f(x)')
pyplot.title('Gradient Descent on x^2')
pyplot.legend()

# Display the plot.
pyplot.show()
```
