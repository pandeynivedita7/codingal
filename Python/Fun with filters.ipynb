```python
import cv2 # Import the OpenCV library for image processing functions.
import numpy as np # Import the NumPy library for numerical operations, especially with arrays.

def apply_color_filter(image, filter_type):
    """Apply the specified color filter to the image."""
    # Create a copy of the image to avoid modifying the original image directly.
    filtered_image = image.copy()

    # Check the type of filter requested by the user.
    if filter_type == "red_tint":
        # For a red tint, set the green (index 1) and blue (index 0) channels to 0.
        filtered_image[:, :, 1] = 0  # Green channel to 0
        filtered_image[:, :, 0] = 0  # Blue channel to 0
    elif filter_type == "blue_tint":
        # For a blue tint, set the green (index 1) and red (index 2) channels to 0.
        filtered_image[:, :, 1] = 0  # Green channel to 0
        filtered_image[:, :, 2] = 0  # Red channel to 0
    elif filter_type == "green_tint":
        # For a green tint, set the blue (index 0) and red (index 2) channels to 0.
        filtered_image[:, :, 0] = 0  # Blue channel to 0
        filtered_image[:, :, 2] = 0  # Red channel to 0
    elif filter_type == "increase_red":
        # To increase red intensity, add 50 to the red channel (index 2).
        filtered_image[:, :, 2] = cv2.add(filtered_image[:, :, 2], 50)  # Increase red channel
    elif filter_type == "decrease_blue":
        # To decrease blue intensity, subtract 50 from the blue channel (index 0).
        filtered_image[:, :, 0] = cv2.subtract(filtered_image[:, :, 0], 50)  # Decrease blue channel

    # Return the image with the applied filter.
    return filtered_image

# Define the path to the image file.
image_path = 'example.jpg'  # Provide your image path
# Load the image using OpenCV. `cv2.imread` returns None if the image is not found.
image = cv2.imread(image_path)

# Check if the image was loaded successfully.
if image is None:
    print("Error: Image not found!") # Print an error message if the image is not found.
else:
    filter_type = "original"  # Initialize the filter type to "original" (no filter applied initially).

    # Display instructions to the user on how to interact with the program.
    print("Press the following keys to apply filters:")
    print("r - Red Tint")
    print("b - Blue Tint")
    print("g - Green Tint")
    print("i - Increase Red Intensity")
    print("d - Decrease Blue Intensity")
    print("q - Quit")

    # Start an infinite loop to continuously display images and apply filters based on user input.
    while True:
        # Apply the currently selected filter to the original image.
        filtered_image = apply_color_filter(image, filter_type)
        # Display the filtered image in a window named "Filtered Image".
        cv2.imshow("Filtered Image", filtered_image)
        # Wait for a key press indefinitely (0 milliseconds wait time).
        # `& 0xFF` is used to get the ASCII value of the key pressed.
        key = cv2.waitKey(0) & 0xFF

        # Check which key was pressed and update the `filter_type` accordingly.
        if key == ord('r'): # 'r' key for red tint.
            filter_type = "red_tint"
        elif key == ord('b'): # 'b' key for blue tint.
            filter_type = "blue_tint"
        elif key == ord('g'): # 'g' key for green tint.
            filter_type = "green_tint"
        elif key == ord('i'): # 'i' key for increasing red intensity.
            filter_type = "increase_red"
        elif key == ord('d'): # 'd' key for decreasing blue intensity.
            filter_type = "decrease_blue"
        elif key == ord('q'): # 'q' key to quit the application.
            print("Exiting...") # Inform the user that the program is exiting.
            break # Exit the while loop.
        else:
            # If an invalid key is pressed, inform the user.
            print("Invalid key! Please use 'r', 'b', 'g', 'i', 'd', or 'q'.")

# Destroy all OpenCV windows when the loop is exited.
cv2.destroyAllWindows()
```
