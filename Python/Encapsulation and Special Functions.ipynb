```python
# Class creation
class myClass:

	# private variable
	# This declares a private class variable named '__privateVar' and initializes it with the value 27.
	# In Python, a variable prefixed with double underscores is name-mangled, making it private to the class.
	__privateVar = 27;

	# private method
	# This defines a private method named '__privMeth'.
	# Similar to variables, methods prefixed with double underscores are name-mangled and private.
	def __privMeth(self):
		# This line prints a string indicating that the method has been called.
		print("I'm inside class myClass")

	# Function to print value of private variable
	# This defines a public method named 'hello'.
	def hello(self):
		# This line prints the value of the private variable '__privateVar' by accessing it through the class name 'myClass'.
		print("Private Variable value: ",myClass.__privateVar)

# Object creation and method call
# This line creates an instance (object) of the 'myClass' and assigns it to the variable 'foo'.
foo = myClass()
# This line calls the 'hello' method on the 'foo' object, which in turn prints the value of the private variable.
foo.hello()
# This line attempts to call the private method '__privMeth' directly from the object 'foo'.
# This will raise an AttributeError because private methods (due to name mangling) cannot be directly accessed from outside the class instance.
foo.__privMeth
```

```python
class Computer:

    # The constructor method, automatically called when a new object is created.
    def __init__(self):
        # Declares a private instance variable '__maxprice' and initializes it to 900.
        # In Python, variables prefixed with double underscores are name-mangled to make them private.
        self.__maxprice = 900

    # A public method to display the selling price.
    def sell(self):
        # Prints the value of the private '__maxprice' variable.
        print("Selling Price: {}".format(self.__maxprice))

    # A public setter method to modify the private '__maxprice' variable.
    def setMaxPrice(self, price):
        # Assigns the new 'price' value to the private '__maxprice' variable.
        self.__maxprice = price

# Creates an instance (object) of the 'Computer' class and assigns it to 'c'.
c = Computer()
# Calls the 'sell' method on the 'c' object, displaying its initial private price.
c.sell()

# Attempts to change the private variable '__maxprice' directly from outside the class.
# This will actually create a NEW public attribute '__maxprice' for the 'c' object,
# leaving the original private '__maxprice' within the object unchanged.
c.__maxprice = 1000
# Calls 'sell' again. It will still print the original private price (900) because the direct
# assignment above did not modify the actual private variable.
c.sell()

# using setter function
# Calls the public 'setMaxPrice' method to correctly modify the private '__maxprice' variable.
c.setMaxPrice(1000)
# Calls 'sell' again. Now it will print the updated private price (1000) because the setter was used.
c.sell()
```

```python
class Account:
    def __init__(self, username, pin):
        self.username = username
        self.__pin = pin  # 1. Private data

    def set_pin(self, old_pin, new_pin): # 3. Safe setter
        if old_pin == self.__pin and len(new_pin) == 4:
            self.__pin = new_pin
            print("PIN updated!")

    def __str__(self): # 4. Print control
        return f"User: {self.username} | PIN: ****"

# --- Testing ---
acc = Account("Alice", "1234")
print(acc)  # Uses __str__

# 2. Testing private access outside class (Will crash if uncommented)
# print(acc.__pin)

acc.set_pin("1234", "5678")  # Updates safely
```

```python
# Create class
class Point:
    # The constructor method for the Point class.
    # It initializes a new Point object with x and y coordinates, defaulting to 0 if not provided.
    def __init__(self, x=0, y=0):
        # Assigns the value of the 'x' parameter to the instance variable 'self.x'.
        self.x = x
        # Assigns the value of the 'y' parameter to the instance variable 'self.y'.
        self.y = y
	# Method to print points in coordinate format
    # This special method defines how the object should be represented as a string.
    # It's called when you use `print()` on an object or `str()` on an object.
    def __str__(self):
        # Returns a formatted string representing the point as coordinates (x, y).
        return "({0}, {1})".format(self.x, self.y)

# Create Object
# Creates an instance (object) of the 'Point' class with x=2 and y=3, and assigns it to 'p1'.
p1 = Point(2, 3)
# Calls the `__str__` method implicitly to get the string representation of 'p1' and prints it.
print(p1)
```
