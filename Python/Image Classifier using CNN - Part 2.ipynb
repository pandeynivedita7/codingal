```python
# 1. Install required library
# This command installs the `np_utils` library, which might be a typo for `keras.utils.np_utils` or `tensorflow.keras.utils.to_categorical` which is part of Keras/TensorFlow for one-hot encoding.
!pip install np_utils

# 2. Import required libraries
# Imports the `files` module from `google.colab` to handle file uploads in Colab.
from google.colab import files
# Imports the `numpy` library, commonly used for numerical operations, and aliases it as `np`.
import numpy as np
# Imports `load_img` from `keras.preprocessing.image` to load an image file.
from keras.preprocessing.image import load_img
# Imports `img_to_array` from `keras.preprocessing.image` to convert an image to a NumPy array.
from keras.preprocessing.image import img_to_array
# Imports `argmax` from `numpy` to find the index of the maximum value in an array.
from numpy import argmax
# Imports `load_model` from `keras.models` to load a pre-trained Keras model.
from keras.models import load_model

# 3. Upload image and model file
# Prints a message to the user prompting them to upload an image and the model file.
print("Step 1: Please upload the image to be classified AND the 'classifier.h5' model file.")

# Uses `files.upload()` to open a file upload dialog in the Colab environment.
# The user should select both their image file (e.g., 'my_cat.jpg') and 'classifier.h5' model file.
uploaded = files.upload()

image_filename = None
model_file_found = False

# Iterate through the uploaded files to identify the image and the model file.
for f_name in uploaded.keys():
    if f_name == 'classifier.h5':
        model_file_found = True
        print(f"Model file '{f_name}' uploaded successfully.")
    else:
        # Assume any other uploaded file is the image to classify.
        image_filename = f_name
        print(f"Image file '{f_name}' uploaded successfully.")

# Check if the model file was uploaded. If not, raise an error.
if not model_file_found:
    raise FileNotFoundError("Error: 'classifier.h5' model file was not uploaded. Please upload it along with your image.")

# Check if an image file was uploaded. If not, raise an error.
if image_filename is None:
    raise ValueError("Error: No image file detected for classification. Please upload an image.")

# Assign the identified image filename to the global 'filename' variable for later use.
filename = image_filename

print("All necessary files uploaded successfully!")
print("Image to classify:", filename)

# 4. Create image preparation function
# Informs the user that the image preparation function is ready.
print("Step 2: Image preparation function is ready.")

# Defines a function `load_image` that takes a `filename` as input.
def load_image(filename):
    # Prints a message indicating that the image loading process has started.
    print("Loading image...")

    # 5. Load the image and resize it to 32x32
    # Loads the image from the given filename and resizes it to 32x32 pixels, a common input size for CNNs.
    img = load_img(filename, target_size=(32, 32))

    # Confirms that the image was loaded successfully.
    print("Image loaded successfully.")
    # Informs the user about the resizing process.
    print("Resizing image to 32x32 pixels...")

    # 6. Convert image to array
    # Converts the loaded image into a NumPy array, which is the format Keras models expect.
    img = img = img_to_array(img)
    # Confirms the conversion of the image to a numerical format.
    print("Image converted into number format.")

    # 7. Reshape image for the model
    # Reshapes the image array. The model expects a batch dimension (1 image), height (32), width (32), and channels (3 for RGB).
    img = img.reshape(1, 32, 32, 3)
    # Confirms that the image has been reshaped appropriately for the model's input.
    print("Image reshaped for the model.")

    # 8. Normalize image, same as training data
    # Changes the data type of the image array to float32, which is standard for neural network inputs.
    img = img.astype('float32')
    # Normalizes the pixel values from the 0-255 range to 0.0-1.0 by dividing by 255.0. This is crucial for model performance.
    img = img / 255.0
    # Confirms successful normalization of the image.
    print("Image normalized successfully.")

    # Returns the prepared image array.
    return img

# 9. Prepare prediction code
# Informs the user that the prediction code is ready.
print("Step 3: Prediction code is ready.")

# Defines the main function to run the prediction example.
def run_example():
    # Prints a message indicating the start of image preparation for prediction.
    print("Preparing the uploaded image for prediction...")

    # 10. Load the uploaded image automatically
    # Calls the `load_image` function to process the previously uploaded image.
    img = load_image(filename)

    # Prints a message indicating that the trained model is being loaded.
    print("Loading trained model...")

    # 11. Load model
    # Loads the pre-trained Keras model saved in 'classifier.h5'.
    model = load_model('classifier.h5')

    # Confirms successful loading of the model.
    print("Model loaded successfully.")
    # Prints a message indicating the start of the prediction process.
    print("Predicting the image class...")

    # 12. Predict the class
    # Uses the loaded model to predict the class of the prepared image.
    result = model.predict(img)

    # Confirms successful completion of the prediction.
    print("Prediction completed successfully!")
    # Displays the raw prediction probabilities output by the model.
    print("Prediction probabilities:")
    print(result)

    # 13. Get class with highest probability
    # Uses `argmax` to find the index of the class with the highest probability from the prediction results.
    predicted_class = argmax(result)

    # Prints a message indicating the final predicted class.
    print("Final predicted class:")

    # 14. Print the predicted class name
    # A series of conditional statements to map the numerical `predicted_class` to its corresponding string label.
    if predicted_class == 0:
        print('Airplane')
    elif predicted_class == 1:
        print('Automobile')
    elif predicted_class == 2:
        print('Bird')
    elif predicted_class == 3:
        print('Cat')
    elif predicted_class == 4:
        print('Deer')
    elif predicted_class == 5:
        print('Dog')
    elif predicted_class == 6:
        print('Frog')
    elif predicted_class == 7:
        print('Horse')
    elif predicted_class == 8:
        print('Ship')
    elif predicted_class == 9:
        print('Truck')

# 15. Entry point, run the example
# Calls the `run_example` function to start the entire process of image upload, preparation, and prediction.
run_example()
```

```python
# ✅ 1. Install required library
# np_utils is not needed. Use TensorFlow's Keras utilities instead.
!pip install tensorflow

# ✅ 2. Import required libraries
from google.colab import files
import numpy as np
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from numpy import argmax
from tensorflow.keras.models import load_model

# ✅ 3. Upload image and model file
print("Step 1: Please upload the image to be classified AND the 'classifier.h5' model file.")
uploaded = files.upload()

image_filename = None
model_file_found = False

for f_name in uploaded.keys():
    if f_name == 'classifier.h5':
        model_file_found = True
        print(f"Model file '{f_name}' uploaded successfully.")
    else:
        image_filename = f_name
        print(f"Image file '{f_name}' uploaded successfully.")

if not model_file_found:
    raise FileNotFoundError("Error: 'classifier.h5' model file was not uploaded. Please upload it along with your image.")

if image_filename is None:
    raise ValueError("Error: No image file detected for classification. Please upload an image.")

filename = image_filename
print("All necessary files uploaded successfully!")
print("Image to classify:", filename)

# ✅ 4. Image preparation function
def load_image(filename):
    print("Loading image...")
    img = load_img(filename, target_size=(32, 32))
    print("Image loaded and resized to 32x32 pixels.")

    img = img_to_array(img)
    print("Image converted into number format.")

    img = img.reshape(1, 32, 32, 3)
    print("Image reshaped for the model.")

    img = img.astype('float32') / 255.0
    print("Image normalized successfully.")

    return img

# ✅ 5. Prediction function
def run_example():
    print("Preparing the uploaded image for prediction...")
    img = load_image(filename)

    print("Loading trained model...")
    model = load_model("/content/classifier.h5")  # ✅ Correct path
    print("Model loaded successfully.")

    print("Predicting the image class...")
    result = model.predict(img)
    print("Prediction completed successfully!")
    print("Prediction probabilities:", result)

    predicted_class = argmax(result)
    print("Final predicted class:")

    # ✅ CIFAR-10 class mapping
    classes = ['Airplane', 'Automobile', 'Bird', 'Cat', 'Deer',
               'Dog', 'Frog', 'Horse', 'Ship', 'Truck']
    print(classes[predicted_class])

# ✅ 6. Run the example
run_example()
```
