```python
from IPython import get_ipython
from IPython.display import display
# %%
from IPython import get_ipython
from IPython.display import display
# %%

import torch  # Used for building and training neural networks. Provides tensor computation with GPU acceleration.
import torchvision # Provides access to datasets, models, and image transformations, specifically for computer vision.
import torchvision.transforms as transforms # Contains common image transformations used for data augmentation and preprocessing.
import torch.nn as nn # Provides modules and classes for building neural networks (e.g., layers, loss functions).
import torch.optim as optim # Provides optimization algorithms (e.g., SGD, Adam) for updating model weights during training.
from torchvision import models # Provides pre-trained models for computer vision tasks (e.g., ResNet).

# Libraries from scikit-learn for evaluation metrics and machine learning algorithms
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, silhouette_score
# accuracy_score: To calculate the proportion of correctly classified instances.
# precision_score: To calculate the ratio of true positives to the total predicted positives.
# recall_score: To calculate the ratio of true positives to the total actual positives.
# f1_score: To calculate the harmonic mean of precision and recall.
# confusion_matrix: To evaluate the performance of a classification model, showing true positives, true negatives, false positives, and false negatives.
# silhouette_score: To evaluate the quality of clusters produced by a clustering algorithm.

from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
# KMeans: A centroid-based clustering algorithm.
# DBSCAN: A density-based spatial clustering of applications and noise algorithm.
# AgglomerativeClustering: A hierarchical clustering algorithm.

from sklearn.decomposition import PCA # Principal Component Analysis, used for dimensionality reduction.
from sklearn.manifold import TSNE # t-Distributed Stochastic Neighbor Embedding, used for visualizing high-dimensional data.

from sklearn.ensemble import RandomForestClassifier # An ensemble learning method for classification, used here as a classifier.
from sklearn.svm import SVC # Support Vector Classifier, a powerful algorithm for classification.
from sklearn.neighbors import KNeighborsClassifier # K-Nearest Neighbors classifier.
from xgboost import XGBClassifier # Extreme Gradient Boosting classifier.

from sklearn.model_selection import cross_val_score, StratifiedKFold, train_test_split
# cross_val_score: To evaluate a score by cross-validation. (Not directly used in this code, but imported)
# StratifiedKFold: Provides train/test indices to split data in train/test sets. (Not directly used in this code, but imported)
# train_test_split: To split arrays or matrices into random train and test subsets.

from sklearn.linear_model import Lasso # Linear Model trained with L1 prior as regularizer. (Imported but not used in this code)

import matplotlib.pyplot as plt # Used for creating visualizations, such as plots and charts (e.g., confusion matrices).
import seaborn as sns # Provides a high-level interface for drawing attractive statistical graphics (used for heatmaps).
import numpy as np # Provides support for large, multi-dimensional arrays and matrices, along with a large collection of high-level mathematical functions.
import pandas as pd # Provides data structures (like DataFrames) and data analysis tools. Used here for creating summary tables.

import warnings # Provides functionality to control the behavior of warnings.
warnings.filterwarnings("ignore") # Ignore warning messages to keep the output clean.

# 1. Data Preparation
# Define transformations for data augmentation and normalization
transform = transforms.Compose([
    transforms.RandomHorizontalFlip(), # Randomly flip images horizontally - Data augmentation to increase the size and diversity of the training dataset.
    transforms.RandomCrop(32, padding=4), # Randomly crop images with padding - Data augmentation.
    transforms.ToTensor(), # Convert images to PyTorch tensors - Required format for PyTorch models.
    transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5)) # Normalize with mean and std - Standardizes the pixel values, which can help model convergence.
])

# Download and load the CIFAR-100 training dataset
trainset = torchvision.datasets.CIFAR100(root='./data', train=True, download=True, transform=transform) # Downloads the dataset and applies the defined transformations.
# Download and load the CIFAR-100 test dataset
testset = torchvision.datasets.CIFAR100(root='./data', train=False, download=True, transform=transform) # Downloads the dataset and applies the defined transformations.
# Create data loader for the training set
trainloader = torch.utils.data.DataLoader(trainset, batch_size=128, shuffle=True, num_workers=2) # Creates an iterable over the dataset, allowing batch processing and shuffling.
# Create data loader for the test set
testloader = torch.utils.data.DataLoader(testset, batch_size=128, shuffle=False, num_workers=2) # Creates an iterable for the test set, without shuffling.

# 2. Model Definition and Training (Comparative Optimization)
# Define the neural network model using a pre-trained ResNet18 (without pre-trained weights initially)
class FeatureNet(nn.Module): # Defines a custom PyTorch model by inheriting from nn.Module.
    def __init__(self):
        super(FeatureNet, self).__init__()
        self.model = models.resnet18(pretrained=False) # Initializes a ResNet18 model from torchvision.models. 'pretrained=False' means random initialization, not using ImageNet weights.
        self.model.fc = nn.Linear(self.model.fc.in_features, 100) # Replaces the final fully connected layer to output 100 classes (for CIFAR-100).

    def forward(self, x):
        return self.model(x) # Defines the forward pass of the model, passing the input through the ResNet.

# Function to train the FeatureNet model with different optimizers and compare
def train_model_comparative():
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu") # Checks if a CUDA-enabled GPU is available and selects the device accordingly.

    # Define the optimizers to compare
    optimizers_to_compare = {
        'SGD': optim.SGD,
        'Adam': optim.Adam,
        'Adagrad': optim.Adagrad,
        # Add more optimizers here as needed, e.g., 'RMSprop': optim.RMSprop
    }

    results = {} # Dictionary to store the trained model and training losses for each optimizer.
    l1_lambda = 1e-5 # Sets the strength of the L1 regularization penalty.

    for name, OptimizerClass in optimizers_to_compare.items(): # Iterate through the defined optimizers.
        print(f"\nTraining with {name} optimizer...")

        model = FeatureNet() # Instantiate a new model for each optimizer to ensure a fair comparison from the same starting point.
        model = model.to(device) # Move the model to the selected device.

        # Define the optimizer for the current iteration.
        # Note: You might need to experiment with learning rates for each optimizer as optimal learning rates can differ.
        if name == 'SGD':
             optimizer = OptimizerClass(model.parameters(), lr=0.01, momentum=0.9) # SGD often benefits from momentum.
        elif name == 'Adam':
            optimizer = OptimizerClass(model.parameters(), lr=0.001) # Adam typically uses a lower learning rate.
        elif name == 'Adagrad':
             optimizer = OptimizerClass(model.parameters(), lr=0.01) # Adagrad can also start with a similar learning rate to SGD.
        else:
             optimizer = OptimizerClass(model.parameters(), lr=0.01) # Default learning rate for other optimizers.


        criterion = nn.CrossEntropyLoss() # Defines the Cross-Entropy loss function.

        model.train() # Set the model to training mode.
        epoch_losses = [] # List to store the average loss for each epoch during training with the current optimizer.

        for epoch in range(5): # Loop for the specified number of training epochs.
            running_loss = 0.0
            for inputs, labels in trainloader: # Iterate through the training data in batches.
                inputs, labels = inputs.to(device), labels.to(device) # Move data to the selected device.
                optimizer.zero_grad() # Clear the gradients.
                outputs = model(inputs) # Perform a forward pass.
                loss = criterion(outputs, labels) # Calculate the base loss.
                l1_penalty = sum(p.abs().sum() for p in model.parameters()) # Calculate the L1 penalty.
                loss += l1_lambda * l1_penalty # Add the L1 penalty to the loss.
                loss.backward() # Compute gradients.
                optimizer.step() # Update model parameters.
                running_loss += loss.item() # Accumulate loss.

            avg_loss = running_loss / len(trainloader) # Calculate average loss for the epoch.
            epoch_losses.append(avg_loss) # Store the average loss.
            print(f"Epoch {epoch+1}: Loss = {avg_loss:.4f}") # Print the loss for the epoch.

        results[name] = {"model": model, "losses": epoch_losses} # Store the trained model and its training losses in the results dictionary.

    return results # Return the dictionary containing results for each optimizer.

# 3. Feature Extraction and Dimensionality Reduction

# Function to extract features from a given model and reduce dimensionality
def extract_features(model):
    model.eval() # Sets the model to evaluation mode.
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu") # Checks and selects the device.
    model.to(device) # Moves the model to the selected device.
    features, labels = [], []
    with torch.no_grad(): # Disables gradient calculation.
        for inputs, targets in testloader: # Iterates through the test data.
            inputs = inputs.to(device) # Moves input data to the device.
            # Extract features from the convolutional layers.
            x = model.model.conv1(inputs)
            x = model.model.bn1(x)
            x = model.model.relu(x)
            x = model.model.maxpool(x)
            x = model.model.layer1(x)
            x = model.model.layer2(x)
            x = model.model.layer3(x)
            x = model.model.layer4(x)
            x = model.model.avgpool(x)
            x = torch.flatten(x, 1) # Flattens the output.
            features.append(x.cpu().numpy()) # Append features as NumPy array.
            labels.extend(targets.numpy()) # Append true labels.
    features = np.vstack(features) # Stack features vertically.
    pca = PCA(n_components=50) # Initialize PCA.
    reduced = pca.fit_transform(features) # Fit and transform with PCA.
    tsne = TSNE(n_components=2, perplexity=30) # Initialize t-SNE.
    tsne_feat = tsne.fit_transform(reduced) # Fit and transform with t-SNE.
    return reduced, np.array(labels), tsne_feat # Return reduced features, labels, and t-SNE features.

# 4. Classification with ML Algorithms

# Function to run various machine learning classifiers
def run_classifiers(X, y, optimizer_name): # Added optimizer_name to track which model's features are used.
    print(f"\n--- Running Classifiers on features trained with {optimizer_name} ---")
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42) # Split data.
    classifiers = {
        'SVM': SVC(),
        'KNN': KNeighborsClassifier(),
        'RandomForest': RandomForestClassifier(),
        'XGBoost': XGBClassifier(use_label_encoder=False, eval_metric='mlogloss')
    }
    summary = []
    for name, clf in classifiers.items(): # Iterate through classifiers.
        clf.fit(X_train, y_train) # Train classifier.
        y_pred = clf.predict(X_test) # Make predictions.
        acc = accuracy_score(y_test, y_pred) # Calculate metrics.
        prec = precision_score(y_test, y_pred, average='macro')
        rec = recall_score(y_test, y_pred, average='macro')
        f1 = f1_score(y_test, y_pred, average='macro')
        cm = confusion_matrix(y_test, y_pred)
        summary.append({"Model": name, "Accuracy": acc, "Precision": prec, "Recall": rec, "F1-Score": f1})
        plt.figure(figsize=(6, 4))
        sns.heatmap(cm, annot=False, fmt='d', cmap='Blues')
        plt.title(f"Confusion Matrix ({optimizer_name}): {name}") # Added optimizer name to title.
        plt.show()
    print(f"\n--- Classification Summary Table ({optimizer_name}) ---") # Added optimizer name to title.
    df = pd.DataFrame(summary)
    print(df.sort_values(by="F1-Score", ascending=False).to_string(index=False))

# 5. Clustering

# Function to run various clustering algorithms
def run_clustering(X, y, optimizer_name): # Added optimizer_name to track which model's features are used.
    print(f"\n--- Running Clustering on features trained with {optimizer_name} ---")
    clustering_algos = {
        'KMeans': KMeans(n_clusters=10, random_state=42),
        'DBSCAN': DBSCAN(eps=5),
        'Agglomerative': AgglomerativeClustering(n_clusters=10)
    }
    summary = []
    for name, algo in clustering_algos.items(): # Iterate through clustering algorithms.
        try:
            labels = algo.fit_predict(X) # Fit and predict clusters.
            if len(set(labels)) > 1 and -1 not in labels:
                score = silhouette_score(X, labels) # Calculate Silhouette score.
                print(f"{name} Silhouette Score ({optimizer_name}): {score:.4f}") # Added optimizer name to output.
                summary.append({"Clustering": name, "Silhouette Score": score})
            else:
                print(f"{name} ({optimizer_name}): Silhouette Score not applicable") # Added optimizer name to output.
                summary.append({"Clustering": name, "Silhouette Score": None})
        except Exception as e:
            print(f"{name} failed ({optimizer_name}): {e}") # Added optimizer name to output.
            summary.append({"Clustering": name, "Silhouette Score": None})
    print(f"\n--- Clustering Summary Table ({optimizer_name}) ---") # Added optimizer name to title.
    df = pd.DataFrame(summary)
    print(df.sort_values(by="Silhouette Score", ascending=False).to_string(index=False))

# 6. Main Execution
if __name__ == '__main__': # Ensures that the following code only runs when the script is executed directly.
    # Train the model using different optimizers
    trained_models_results = train_model_comparative()

    # Plot the training loss for each optimizer
    plt.figure(figsize=(10, 6))
    for optimizer_name, results in trained_models_results.items():
        plt.plot(results["losses"], label=optimizer_name)
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title("Training Loss Comparison Across Optimizers")
    plt.legend()
    plt.grid(True)
    plt.show()

    # Choose one of the trained models for downstream tasks (e.g., the one with the lowest final loss)
    # Or, you can iterate through all trained models and run the downstream tasks for each to compare
    # based on the features extracted from models trained with different optimizers.

    # Example: Use the model trained with Adam for downstream tasks
    # selected_optimizer = 'Adam'
    # selected_model = trained_models_results[selected_optimizer]["model"]

    # Or, loop through all trained models
    for optimizer_name, results in trained_models_results.items():
        print(f"\n--- Processing with model trained using {optimizer_name} optimizer ---")
        selected_model = results["model"]

        # Extract features and perform dimensionality reduction
        features, labels, tsne_feat = extract_features(selected_model)

        # Display a sample of features and labels
        print("\n--- Sample of Extracted Features (first 5 samples) ---")
        print(features[:5])
        print("\n--- Sample of Labels (first 50 samples) ---")
        print(labels[:50])


        # Run classification algorithms
        run_classifiers(features, labels, optimizer_name)

        # Run clustering algorithms
        run_clustering(features, labels, optimizer_name)
```
