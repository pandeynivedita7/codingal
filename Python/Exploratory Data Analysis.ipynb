```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# -----------------------------
# Load Titanic Dataset
# -----------------------------
data = pd.read_csv('Titanic Dataset.csv')

# Display first 5 rows
print(data.head())


# -----------------------------
# 1. Which gender survived most?
# -----------------------------
plt.figure(figsize=(7, 5))

sns.countplot(
    x='Gender',
    hue='Survived',
    data=data
)

plt.title('Survival by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.show()


# -----------------------------
# 2. Which passenger class
# survived most and least?
# -----------------------------
plt.figure(figsize=(7, 5))

sns.countplot(
    x='Pclass',
    hue='Survived',
    data=data
)

plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')
plt.show()


# -----------------------------
# 3. Highest number of passengers
# belong to which age?
# -----------------------------
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x='Age',
    bins=40,
    kde=False
)

plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()


# -----------------------------
# 4. Highest number of passengers
# belong to which gender?
# -----------------------------
plt.figure(figsize=(7, 5))

sns.countplot(
    x='Gender',
    data=data
)

plt.title('Passenger Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.show()


# -----------------------------
# 5. Association between
# SibSp and Survived
# -----------------------------
plt.figure(figsize=(8, 5))

sns.countplot(
    x='Survived',
    hue='SibSp',
    data=data,
    palette='mako'
)

plt.title('Survival Based on Siblings/Spouses')
plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.show()


# -----------------------------
# 6. Association between
# Parch and Survived
# -----------------------------
plt.figure(figsize=(8, 5))

sns.countplot(
    x='Survived',
    hue='Parch',
    data=data,
    palette='mako'
)

plt.title('Survival Based on Parents/Children')
plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.show()


# -----------------------------
# 7. Fare distribution
# -----------------------------
plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x='Fare',
    bins=40,
    kde=True
)

plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()


# -----------------------------
# 8. Age distribution according
# to passenger class
# -----------------------------
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Pclass',
    y='Age',
    data=data,
    palette='winter'
)

plt.title('Age Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.show()


# -----------------------------
# 9. Correlation of numerical
# features with Survived
# -----------------------------

# Select only numerical columns
numeric_data = data.select_dtypes(include=np.number)

# Calculate correlation
correlation = numeric_data.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix')
plt.show()


# -----------------------------
# Correlation specifically with
# Survived
# -----------------------------
if 'Survived' in correlation.columns:
    print("\nCorrelation with Survived:")
    print(
        correlation['Survived']
        .sort_values(ascending=False)
    )
```

```python
####**Import Libraries**

# Import Libraries
# pandas is used for data manipulation and analysis.
import pandas as pd
# numpy is used for numerical operations, especially with arrays.
import numpy as np
# matplotlib.pyplot is used for creating static, animated, and interactive visualizations.
import matplotlib.pyplot as plt
# seaborn is a data visualization library based on matplotlib, providing a high-level interface for drawing attractive statistical graphics.
import seaborn as sns



# Import dataset: pd.read_csv reads a CSV file into a pandas DataFrame.
data = pd.read_csv('Iris.csv')

# data.head(5) displays the first 5 rows of the DataFrame.
data.head(5)

"""#### **Check Null Values**"""

# data.isnull().sum() calculates the sum of null values for each column.
data.isnull().sum()

"""- No null values are present

#### **Statistical Information of features**"""

# data.describe() generates descriptive statistics of the DataFrame's numerical columns.
data.describe()

"""#### **CreateBoxplot of all the features and check if outliers are present**"""

# Define a list of labels for the columns to be analyzed.
labels = ['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
# Loop through each label to create a boxplot.
for label in labels:
  # Print the distribution title for the current label.
  print('Distribution of', label)
  # sns.boxplot creates a box plot for the current column.
  sns.boxplot(data[label])
  # plt.show() displays the plot.
  plt.show()

"""**Conclusion**

- SepalWidthCm has outliers

#### **Check correlation between all the features**"""

# data.corr() calculates the pairwise correlation of all numerical columns.
# sns.heatmap visualizes this correlation matrix.
sns.heatmap(data.corr())

"""#### **Check if any feature is skewed or not**"""

# Define a list of labels for the columns to be analyzed.
labels = ['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
# Loop through each label to create a distribution plot.
for label in labels:
  # Print the distribution title for the current label.
  print('Distribution of', label)
  # sns.distplot plots the distribution of the current column.
  sns.distplot(data[label])
  # plt.show() displays the plot.
  plt.show()

# Define a list of labels for the columns to calculate skewness.
labels = ['Id','SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']
# Loop through each label to calculate and print its skewness.
for label in labels:
  # Print the skewness title for the current label.
  print('skewness of ', label)
  # data[label].skew() calculates the skewness of the current column.
  print(data[label].skew())
```

```python
# ============================================
# DATA SCIENCE - BASIC DATA ANALYSIS IN PYTHON
# Topics:
# 1. Exploratory Data Analysis
# 2. Univariate Analysis
# 3. Bivariate Analysis
# 4. Multivariate Analysis
# ============================================


# -------------------------
# Import Libraries
# -------------------------

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# -------------------------
# Load Dataset
# -------------------------

data = pd.read_csv("Titanic Dataset.csv")

print("Dataset Loaded Successfully")


# ============================================
# 1. EXPLORATORY DATA ANALYSIS - EDA
# ============================================

print("\nFirst 5 Rows:")
print(data.head())

print("\nLast 5 Rows:")
print(data.tail())

print("\nShape of Dataset:")
print(data.shape)

print("\nColumn Names:")
print(data.columns)

print("\nDataset Information:")
print(data.info())

print("\nStatistical Summary:")
print(data.describe())

print("\nMissing Values:")
print(data.isnull().sum())

print("\nDuplicate Rows:")
print(data.duplicated().sum())


# ============================================
# 2. UNIVARIATE ANALYSIS
# Analysis of ONE variable
# ============================================


# --------------------------------------------
# A. Gender Distribution
# --------------------------------------------

plt.figure(figsize=(6, 4))

sns.countplot(
    x='Gender',
    data=data
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()


# --------------------------------------------
# B. Passenger Class Distribution
# --------------------------------------------

plt.figure(figsize=(6, 4))

sns.countplot(
    x='Pclass',
    data=data
)

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Passenger Count")
plt.show()


# --------------------------------------------
# C. Age Distribution
# --------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x='Age',
    bins=30,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()


# --------------------------------------------
# D. Fare Distribution
# --------------------------------------------

plt.figure(figsize=(8, 5))

sns.histplot(
    data=data,
    x='Fare',
    bins=30,
    kde=True
)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()


# ============================================
# 3. BIVARIATE ANALYSIS
# Analysis of TWO variables
# ============================================


# --------------------------------------------
# A. Gender vs Survival
# --------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    x='Gender',
    hue='Survived',
    data=data
)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Passenger Count")
plt.show()


# --------------------------------------------
# B. Passenger Class vs Survival
# --------------------------------------------

plt.figure(figsize=(7, 5))

sns.countplot(
    x='Pclass',
    hue='Survived',
    data=data
)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Passenger Count")
plt.show()


# --------------------------------------------
# C. Age vs Fare
# --------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x='Age',
    y='Fare',
    data=data
)

plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()


# --------------------------------------------
# D. Passenger Class vs Age
# --------------------------------------------

plt.figure(figsize=(7, 5))

sns.boxplot(
    x='Pclass',
    y='Age',
    data=data
)

plt.title("Age Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()


# ============================================
# 4. MULTIVARIATE ANALYSIS
# Analysis of THREE OR MORE variables
# ============================================


# --------------------------------------------
# A. Age vs Fare based on Survival
# --------------------------------------------

plt.figure(figsize=(8, 5))

sns.scatterplot(
    x='Age',
    y='Fare',
    hue='Survived',
    data=data
)

plt.title("Age vs Fare Based on Survival")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()


# --------------------------------------------
# B. Age vs Fare based on Gender and Survival
# --------------------------------------------

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x='Age',
    y='Fare',
    hue='Survived',
    style='Gender',
    data=data
)

plt.title("Age, Fare, Gender and Survival")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()


# --------------------------------------------
# C. Passenger Class, Age and Survival
# --------------------------------------------

plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Pclass',
    y='Age',
    hue='Survived',
    data=data
)

plt.title("Passenger Class, Age and Survival")
plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.show()


# --------------------------------------------
# D. Correlation Heatmap
# --------------------------------------------

numeric_data = data.select_dtypes(include=np.number)

correlation = numeric_data.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap")
plt.show()
```
