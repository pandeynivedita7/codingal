### List Methods in Python

Lists are ordered, mutable collections of items in Python. They come with several built-in methods that allow you to modify, inspect, and manipulate their contents efficiently. Here's a breakdown of the most commonly used list methods with examples:

#### 1. `append(element)`

Adds a single `element` to the end of the list.

```python
my_list = [1, 2, 3]
my_list.append(4)
print(my_list)
# Output: [1, 2, 3, 4]
```

#### 2. `extend(iterable)`

Appends all the elements from an `iterable` (like another list, tuple, or string) to the end of the list. This is similar to using the `+` operator for lists but modifies the list in-place.

```python
my_list = [1, 2, 3]
another_list = [4, 5]
my_list.extend(another_list)
print(my_list)
# Output: [1, 2, 3, 4, 5]
```

#### 3. `insert(index, element)`

Inserts an `element` at a specified `index` in the list. The existing elements from that index onwards are shifted to the right.

```python
my_list = [1, 2, 4, 5]
my_list.insert(2, 3) # Insert 3 at index 2
print(my_list)
# Output: [1, 2, 3, 4, 5]
```

#### 4. `remove(element)`

Removes the first occurrence of the specified `element` from the list. If the element is not found, it raises a `ValueError`.

```python
my_list = [1, 2, 3, 2, 4]
my_list.remove(2) # Removes the first '2'
print(my_list)
# Output: [1, 3, 2, 4]
```

#### 5. `pop(index=-1)`

Removes and returns the element at the specified `index`. If no index is provided, `pop()` removes and returns the last element of the list.

```python
my_list = [10, 20, 30, 40]
removed_item = my_list.pop(1) # Remove element at index 1 (20)
print(f"Removed: {removed_item}, List: {my_list}")

last_item = my_list.pop() # Remove the last element (40)
print(f"Removed: {last_item}, List: {my_list}")
# Output:
# Removed: 20, List: [10, 30, 40]
# Removed: 40, List: [10, 30]
```

#### 6. `clear()`

Removes all elements from the list, making it empty.

```python
my_list = [1, 2, 3]
my_list.clear()
print(my_list)
# Output: []
```

#### 7. `index(element, start=0, end=len(list))`

Returns the index of the first occurrence of the specified `element`. You can optionally specify `start` and `end` indices to search within a sub-section of the list. Raises a `ValueError` if the element is not found.

```python
my_list = ['apple', 'banana', 'cherry', 'apple']
index_banana = my_list.index('banana')
print(f"Index of 'banana': {index_banana}")

index_first_apple = my_list.index('apple')
print(f"Index of first 'apple': {index_first_apple}")

index_second_apple = my_list.index('apple', 1) # Search from index 1
print(f"Index of second 'apple' (searching from index 1): {index_second_apple}")
# Output:
# Index of 'banana': 1
# Index of first 'apple': 0
# Index of second 'apple' (searching from index 1): 3
```

#### 8. `count(element)`

Returns the number of times a specified `element` appears in the list.

```python
my_list = [1, 2, 2, 3, 2, 4]
count_two = my_list.count(2)
print(f"Count of 2: {count_two}")

count_five = my_list.count(5)
print(f"Count of 5: {count_five}")
# Output:
# Count of 2: 3
# Count of 5: 0
```

#### 9. `sort(key=None, reverse=False)`

Sorts the items of the list in-place (i.e., modifies the original list). By default, it sorts in ascending order. You can specify a `key` function for custom sorting or set `reverse=True` for descending order.

```python
numbers = [3, 1, 4, 1, 5, 9, 2]
numbers.sort() # Sorts in ascending order
print(f"Sorted ascending: {numbers}")

words = ['banana', 'apple', 'cherry']
words.sort(reverse=True) # Sorts in descending order
print(f"Sorted descending: {words}")

# Custom sort using a key (e.g., sort by length of string)
names = ['Alice', 'Bob', 'Charlie']
names.sort(key=len) # Sort by length
print(f"Sorted by length: {names}")
# Output:
# Sorted ascending: [1, 1, 2, 3, 4, 5, 9]
# Sorted descending: ['cherry', 'banana', 'apple']
# Sorted by length: ['Bob', 'Alice', 'Charlie']
```

```python
# function to check whether
# first and last character of words match
def match_words(words):
	ctr = 0
	lst = []
	for word in words:
		if len(word) > 1 and word[0] == word[-1]:
			ctr += 1
			lst.append(word)

	print("List of words with first and last character same\n", lst)
	return ctr

count = match_words(['abc', 'cfc','xyz', 'aba', '1221'])
print("Number of words having first and last character same:", count)
```

#### 10. `reverse()`

Reverses the order of the elements in the list in-place.

```python
my_list = [1, 2, 3, 4, 5]
my_list.reverse()
print(my_list)
# Output: [5, 4, 3, 2, 1]
```

#### 11. `copy()`

Returns a shallow copy of the list. This means a new list is created with references to the original elements. If the elements themselves are mutable (e.g., nested lists), changes to those mutable elements will be reflected in both the original and copied list.

```python
original_list = [1, 2, [3, 4]]
copied_list = original_list.copy()

print(f"Original: {original_list}, Copied: {copied_list}")

# Modify the copied list (primitive element)
copied_list[0] = 100
print(f"After modifying copied[0]: Original: {original_list}, Copied: {copied_list}")

# Modify a nested mutable element
copied_list[2][0] = 300
print(f"After modifying copied[2][0]: Original: {original_list}, Copied: {copied_list}")
# Output:
# Original: [1, 2, [3, 4]], Copied: [1, 2, [3, 4]]
# After modifying copied[0]: Original: [1, 2, [3, 4]], Copied: [100, 2, [3, 4]]
# After modifying copied[2][0]: Original: [1, 2, [300, 4]], Copied: [100, 2, [300, 4]]
```

## Square it Out!

This section will generate a list of square values for numbers within a user-defined range and then separate these squared values into lists of odd and even numbers.

```python
# Get user input for the range
try:
    start_range = int(input("Enter the starting number for the range: "))
    end_range = int(input("Enter the ending number for the range: "))
except ValueError:
    print("Invalid input. Please enter integers for the range.")
    start_range = 1
    end_range = 10
    print(f"Defaulting to range from {start_range} to {end_range}.")

# Ensure start_range is not greater than end_range
if start_range > end_range:
    start_range, end_range = end_range, start_range
    print(f"Swapped range: now from {start_range} to {end_range}.")

# Create a list of square values
squared_numbers = []
for num in range(start_range, end_range + 1):
    squared_numbers.append(num ** 2)

print(f"Original numbers range: {list(range(start_range, end_range + 1))}")
print(f"List of squared numbers: {squared_numbers}")
```

### Separating Odd and Even Squared Values

```python
# Separate odd and even values
odd_squares = []
even_squares = []

for square in squared_numbers:
    if square % 2 == 0:
        even_squares.append(square)
    else:
        odd_squares.append(square)

print(f"Odd squared numbers: {odd_squares}")
print(f"Even squared numbers: {even_squares}")
```
