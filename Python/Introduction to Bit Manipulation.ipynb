```python
a = 10   # binary: 1010
b = 6    # binary: 0110

def bits(n, width=4):
    # This function converts an integer 'n' to its binary representation
    # of a specified 'width'.
    # (1 << width) creates a number with 'width' zeros followed by a one (e.g., for width=4, 10000).
    # (1 << width) - 1 creates a mask of 'width' ones (e.g., for width=4, 1111).
    # n & ((1 << width) - 1) applies the mask to 'n', ensuring only the lowest 'width' bits are considered.
    # format(..., f'0{width}b') formats the result as a binary string, padding with leading zeros to meet the specified 'width'.
    return format(n & ((1 << width) - 1), f'0{width}b')

print('a =', a, '->', bits(a))   # Prints the value of 'a' and its 4-bit binary representation.
print('b =', b, '->', bits(b))   # Prints the value of 'b' and its 4-bit binary representation.

# How 10 becomes 1010:
# Position:  8  4  2  1
# Bits:      1  0  1  0
# Value:    8 + 0 + 2 + 0 = 10
```

```python
# BIT EXPLORER
# Topics: Bits & Binary | AND & OR | NOT & XOR | Shifts | Odd/Even | Count Bits

a = 10   # Decimal 10, binary representation: 1010
b = 6    # Decimal 6,  binary representation: 0110

def bits(n, width=4):
    # This function converts an integer 'n' to its binary representation
    # of a specified 'width'.
    # 1. `(1 << width)`: Shifts 1 left by `width` positions. This creates a number
    #    with a '1' followed by `width` zeros (e.g., for width=4, 1 << 4 = 10000₂ = 16₁₀).
    # 2. `((1 << width) - 1)`: Subtracting 1 from the above result creates a mask
    #    consisting of `width` ones (e.g., for width=4, 16 - 1 = 15₁₀ = 1111₂).
    #    This mask is used to isolate the desired number of least significant bits.
    # 3. `n & ((1 << width) - 1)`: Performs a bitwise AND operation between `n` and the mask.
    #    This effectively truncates `n` to its lowest `width` bits, ensuring that only
    #    the bits relevant to the specified width are considered.
    #    Example for a=10 (1010₂) and width=4, mask is 1111₂:
    #        1010₂ (n)
    #    &   1111₂ (mask)
    #    ----------
    #        1010₂ (result is still 10, but ensures it fits in 4 bits if 'n' was larger)
    # 4. `format(..., f'0{width}b')`: Formats the result as a binary string.
    #    - `0`: Pads the string with leading zeros if necessary.
    #    - `{width}`: Specifies the total width of the binary string (e.g., '4' for 4 bits).
    #    - `b`: Indicates that the number should be formatted as a binary string.
    #    Example: format(10, '04b') produces '1010'
    return format(n & ((1 << width) - 1), f'0{width}b')

# PART 1: Bits and Binary - Introduction to binary representations
print("=== Bit Explorer ===")
print("a =", a, "->", bits(a))   # Output: a = 10 -> 1010 (Decimal 10 to 4-bit binary)
print("b =", b, "->", bits(b))   # Output: b = 6 -> 0110 (Decimal 6 to 4-bit binary)
print()

# PART 2: AND and OR - Demonstrating fundamental bitwise operations
print("AND  a & b =", a & b,      "->", bits(a & b))    # Bitwise AND: result bit is 1 only if *both* corresponding bits are 1.
# a (1010)
# & b (0110)
# ---------
#   0010 (Decimal 2)
print("OR   a | b =", a | b,      "->", bits(a | b))    # Bitwise OR: result bit is 1 if *either* corresponding bit is 1.
# a (1010)
# | b (0110)
# ---------
#   1110 (Decimal 14)
print()

# PART 3: NOT and XOR - Demonstrating inversion and exclusive OR
# NOTE: ~a (bitwise NOT) results in a negative number in Python for positive 'a'
# To get the expected unsigned bit-flip for a fixed width, we use '& 0xFF' (mask for 8 bits)
print("NOT  ~a    =", ~a & 0xFF,  "->", bits(~a, 8))    # Bitwise NOT: Flips all bits. (e.g., for a=10, ~10 = -11. For 8-bit, 00001010 becomes 11110101 = 245)
# Original a (8-bit): 00001010
# ~a (8-bit):         11110101 (Decimal 245)
print("XOR  a ^ b =", a ^ b,      "->", bits(a ^ b))    # Bitwise XOR: result bit is 1 if corresponding bits are *different*.
# a (1010)
# ^ b (0110)
# ---------
#   1100 (Decimal 12)
print()

# PART 4: Left Shift and Right Shift - Multiplying and dividing by powers of 2
print("LEFT  a << 1 =", a << 1, " (a x 2)")   # Left shift by 1: Multiplies 'a' by 2^1. (1010₂ << 1 = 10100₂ = 20₁₀)
print("RIGHT a >> 1 =", a >> 1, " (a / 2)")   # Right shift by 1: Divides 'a' by 2^1 (integer division). (1010₂ >> 1 = 0101₂ = 5₁₀)
# Shift trace for a = 10:
# Binary a:      1 0 1 0
# a << 1:      1 0 1 0 0   = 20  (slide bits left, add 0 on the right)
# a >> 1:          0 1 0 1 = 5   (slide bits right, drop the rightmost bit)
# Pattern:
# a << n = a * (2^n)
# a >> n = a // (2^n) (integer division)
print()

# PART 5: Odd or Even with XOR - A clever trick using XOR for parity check
print("Odd or Even:")
for n in [7, 10, 15, 4]:
    # An integer 'n' is odd if its least significant bit (LSB) is 1.
    # An integer 'n' is even if its LSB is 0.
    # XORing with 1 (0001₂):
    # If LSB of 'n' is 0 (even), then n ^ 1 flips LSB to 1, effectively adding 1 (n + 1).
    #   e.g., n=10 (1010₂) ^ 1 (0001₂) = 1011₂ (11₁₀) --> 10 + 1. So n ^ 1 == n + 1 is True for even numbers.
    # If LSB of 'n' is 1 (odd), then n ^ 1 flips LSB to 0, effectively subtracting 1 (n - 1).
    #   e.g., n=7 (0111₂) ^ 1 (0001₂) = 0110₂ (6₁₀) --> 7 - 1. So n ^ 1 == n + 1 is False for odd numbers.
    result = "Even" if n ^ 1 == n + 1 else "Odd"
    print(n, "->", result)
# Trace for n=10 (even): 1010₂ ^ 0001₂ = 1011₂ = 11₁₀. Since 11 == 10 + 1, it's 'Even'.
# Trace for n=7  (odd):  0111₂ ^ 0001₂ = 0110₂ = 6₁₀. Since 6 != 7 + 1, it's 'Odd'.
print()

# PART 6: Count Bits - Counting the number of set bits (1s) in a binary representation
def count_bits(n):
    count = 0
    # Loop continues as long as 'n' is not zero.
    while n:
        # `count += 1` increments the counter each time a bit is found.
        count += 1
        # `n >>= 1` performs a right shift by 1, effectively discarding the rightmost bit
        # and moving the next bit to the LSB position for the next iteration.
    # The loop terminates when all bits have been shifted out and 'n' becomes 0.
        n >>= 1
    return count

print("Bit count:")
for n in [a, b, 255]:
    # For a=10, bits(10, count_bits(10)) will display '1010'
    # For b=6, bits(6, count_bits(6)) will display '110'
    # For 255, bits(255, count_bits(255)) will display '11111111'
    print(n, "->", count_bits(n), "bits |", bits(n, count_bits(n)))
```
