```python
!pip install requests

import requests

HF_API_KEY = input("Enter your Hugging Face API key: ")
DEFAULT_MODEL = "google/pegasus-xsum"

def build_api_url(model_name):
    return f"https://api-inference.huggingface.co/models/{model_name}"

def query(payload, model_name=DEFAULT_MODEL):
    api_url = build_api_url(model_name)
    headers = {"Authorization": f"Bearer {HF_API_KEY}"}
    response = requests.post(api_url, headers=headers, json=payload)
    return response.json()

def summarize_text(text, min_length, max_length, model_name=DEFAULT_MODEL):
    payload = {
        "inputs": text,
        "parameters": {"min_length": min_length, "max_length": max_length}
    }
    print(f"\nPerforming summarization using: {model_name}")
    result = query(payload, model_name=model_name)
    if isinstance(result, list) and result and "summary_text" in result[0]:
        return result[0]["summary_text"]
    else:
        print("Error in summarization response:", result)
        return None

# Get user input
user_name = input("Hi there! What's your name? ").strip() or "User"
print(f"Welcome, {user_name}!")

user_text = input("\nEnter the text you want to summarize:\n").strip()
if not user_text:
    print("No text provided. Exiting.")
else:
    model_choice = input("Enter model name (leave blank for default): ").strip() or DEFAULT_MODEL
    style_choice = input("Choose style (1. Standard, 2. Enhanced): ").strip()

    if style_choice == "2":
        min_length = 80
        max_length = 200
    else:
        min_length = 50
        max_length = 150

    summary = summarize_text(user_text, min_length, max_length, model_name=model_choice)
    if summary:
        print(f"\nSummary for {user_name}:\n", summary)
    else:
        print("Failed to generate summary.")
```
