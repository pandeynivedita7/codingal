```python
# import asarray from numpy for creating arrays
from numpy import asarray
# import arange from numpy for creating sequences of numbers
from numpy import arange
# import rand from numpy.random for generating random numbers
from numpy.random import rand
# import pyplot from matplotlib for plotting graphs
from matplotlib import pyplot

# objective function, which is x squared
def objective(x):
    return x**2.0

# derivative of the objective function, which is 2x
def derivative(x):
    return x * 2.0

# gradient descent algorithm implementation
def gradient_descent(objective, derivative, bounds, n_iter, step_size):
    # initialize lists to track all solutions and their scores
    solutions, scores = list(), list()
    # generate an initial random point within the defined bounds
    solution = bounds[:, 0] + rand(len(bounds)) * (bounds[:, 1] - bounds[:, 0])
    # iterate for a specified number of iterations
    for i in range(n_iter):
        # calculate the gradient of the objective function at the current solution
        gradient = derivative(solution)
        # take a step in the opposite direction of the gradient to find a new solution
        solution = solution - step_size * gradient
        # evaluate the objective function at the new candidate solution
        solution_eval = objective(solution)
        # store the new solution in the solutions list
        solutions.append(solution)
        # store the score (objective function value) of the new solution
        scores.append(solution_eval)
        # print the progress: iteration number, solution, and its evaluated score
        print('>%d f(%s) = %.5f' % (i, solution, solution_eval))
    # return the list of solutions and their corresponding scores
    return [solutions, scores]

# define the range (bounds) for the input variable x, from -1.0 to 1.0
bounds = asarray([[-1.0, 1.0]])
# define the total number of iterations for the gradient descent
n_iter = 30
# define the step size (learning rate) for the gradient descent algorithm
step_size = 0.1

# perform the gradient descent search using the defined objective, derivative, bounds, iterations, and step size
solutions, scores = gradient_descent(objective, derivative, bounds, n_iter, step_size)
# sample input range uniformly at 0.1 increments for plotting the objective function
inputs = arange(bounds[0,0], bounds[0,1]+0.1, 0.1)
# compute the target values (y-values) for the sampled inputs using the objective function
results = objective(inputs)
# create a line plot of the input values vs. their corresponding results (the objective function curve)
pyplot.plot(inputs, results)
# plot the solutions found by the gradient descent algorithm, marked with red dots and lines
pyplot.plot(solutions, scores, '.-', color='red')
# display the plot
pyplot.show()
```
