```python
# 01-sum-down.py
# Topic: What Is Recursion?, The Base Case, How Recursion Builds and Unwinds

def sum_down(n):
    if n == 0:
        return 0
    return n + sum_down(n - 1)

input("Recursion - a function calls itself until a base case stops it.  Press Enter ")
print("  sum_down(3) = 3 + 2 + 1 + 0 =", sum_down(3))
print("  sum_down(4) = 4 + 3 + 2 + 1 + 0 =", sum_down(4))

n = int(input("Enter a number (try 5 or 6): "))
guess = input("What is sum_down(" + str(n) + ")? ")
input("sum_down adds n to sum_down(n-1) until n reaches 0.  Press Enter ")
print("  sum_down(" + str(n) + ") =", sum_down(n), "  your guess:", guess)
```

```python
# 02-factorial.py
# Topic: Counting with Recursion, Factorial Using Recursion

def factorial(n):
    # Step 1: Define the recursive function `factorial` that takes an integer `n`.
    # Step 2: Check for the base case.
    # If `n` is 0, the recursion stops, and the function returns 1 (as 0! = 1).
    if n == 0:
        return 1
    # Step 3: If not the base case, perform the recursive step.
    # The function returns the current number `n` multiplied by the result of calling itself with `n-1`.
    # This process continues, decrementing `n` until the base case (n=0) is met.
    return n * factorial(n - 1)

# Step 4: Explain the concept of factorial and its base case to the user and wait for interaction.
input("Factorial - n! = n x (n-1) x ... x 1  base case: 0! = 1.  Press Enter ")
# Step 5: Demonstrate the function with `n=4`.
# This line calls `factorial(4)`, which will compute 4 * 3 * 2 * 1 * 1.
print("  4! =", factorial(4), "  = 4 x 3 x 2 x 1")
# Step 6: Demonstrate the function with `n=5`.
# This line calls `factorial(5)`, which will compute 5 * 4 * 3 * 2 * 1 * 1.
print("  5! =", factorial(5), "  = 5 x 4 x 3 x 2 x 1")

# Step 7: Prompt the user to enter a number to try the function themselves.
n = int(input("Enter a number (try 3 or 6): "))
# Step 8: Ask the user to guess the output for their chosen number.
guess = input("What is " + str(n) + "!? ")
# Step 9: Provide a brief explanation of how `factorial` works again and wait for user interaction.
input("n! = n x factorial(n-1)  base case: 0! = 1.  Press Enter ")
# Step 10: Display the actual result of `factorial` for the user's input and compare it with their guess.
print(" ", str(n) + "! =", factorial(n), "  your guess:", guess)
```

```python
# 03-stack-overflow.py
# Topic: Stack Overflow

def sum_down(n):
    # Step 1: Define the recursive function `sum_down` that takes an integer `n`.
    # Step 2: Check for the base case.
    # If `n` is 0, the recursion stops, and the function returns 0.
    if n == 0:
        return 0
    # Step 3: If not the base case, perform the recursive step.
    # The function returns the current number `n` plus the result of calling itself with `n-1`.
    # Each recursive call adds a new frame to the call stack.
    return n + sum_down(n - 1)

# Step 4: Explain the concept of stack overflow in the context of recursion and wait for interaction.
# Python has a default recursion limit (usually 1000), meaning it can only add up to 1000 frames to the stack.
input("Stack overflow - each call adds a frame to the stack  limit is 1000.  Press Enter ")
# Step 5: Demonstrate the function with `n=3` and show the number of frames used.
print("  sum_down(3) =", sum_down(3), "  uses 3 frames")
# Step 6: Demonstrate the function with `n=5` and show the number of frames used.
print("  sum_down(5) =", sum_down(5), "  uses 5 frames")

# Step 7: Prompt the user to enter a number to understand the frame usage.
n = int(input("Enter a number (try 6 or 9): "))
# Step 8: Provide an explanation of frame usage related to the input `n`.
input("sum_down(n) = n + (n-1) + ... + 1  uses n frames.  Press Enter ")
# Step 9: Display the result of `sum_down` for the user's input and indicate the number of frames used.
# If the user enters a number greater than the recursion limit, this will result in a RecursionError (stack overflow).
print("  sum_down(" + str(n) + ") =", sum_down(n), "  frames used:", n)
```
