```python
class fruit:
    # class variable
    taste = 'Sweet'

    # instance variable
    def __init__(self, name, color):
        self.name = name
        self.color = color

# Object Creation
apple = fruit('Apple','Red')
banana = fruit('Banana', 'Yellow')

print(apple.taste)
print(apple.name, apple.color)
print(banana.name, banana.color)
```

```python
class Parrot:

    # class attribute
    species = "bird"

    # instance attribute
    def __init__(self, name, age):
        self.name = name
        self.age = age

# instantiate the Parrot class
blu = Parrot("Blu", 10)
woo = Parrot("Woo", 15)

# access the class attributes
print("Blu is a {}".format(blu.species))
print("Woo is also a {}".format(woo.species))

# access the instance attributes
print("{} is {} years old".format( blu.name, blu.age))
print("{} is {} years old".format( woo.name, woo.age))
```

```python
class Parrot:

    # instance attributes
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # instance method
    def sing(self, song):
        return "{} sings {}".format(self.name, song)

    def dance(self):
        return "{} is now dancing".format(self.name)

# instantiate the object
blu = Parrot("Blu", 10)

# call our instance methods
print(blu.sing("'Happy'"))
print(blu.dance())
```

### Robot Introduction using OOP

Here's a program that uses Object-Oriented Programming to introduce a robot. The `Robot` class encapsulates the robot's properties (like its name and purpose) and behavior (like introducing itself).

```python
class Robot:
    # Class attribute for all robots
    robot_type = "AI Assistant"

    def __init__(self, name, purpose):
        # Instance attributes for specific robot
        self.name = name
        self.purpose = purpose

    def introduce(self):
        """A method for the robot to introduce itself."""
        print(f"Hello! My name is {self.name}.")
        print(f"I am a {self.robot_type} robot.")
        print(f"My purpose is to {self.purpose}.")

    def perform_action(self, action):
        """A method to demonstrate a robot's action."""
        print(f"{self.name} is now {action}.")

# Create an instance of the Robot class
my_robot = Robot("Colab Assistant", "help users with their programming tasks in Google Colab")

# Call the introduction method
my_robot.introduce()

print("\n") # Add a newline for better readability

# Call another method to show an action
my_robot.perform_action("analyzing your code")
```
