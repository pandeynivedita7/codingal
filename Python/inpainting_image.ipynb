```python
# ✅ Install required libraries
!pip install -q diffusers transformers accelerate scipy safetensors

# ✅ Import necessary modules
import torch
from PIL import Image
from diffusers import StableDiffusionInpaintPipeline
import requests
from io import BytesIO

# ✅ Hugging Face Login (Use your token here)
from huggingface_hub import login
login("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")  # Replace with your actual token

# ✅ Load Inpainting Model
pipe = StableDiffusionInpaintPipeline.from_pretrained(
    "stabilityai/stable-diffusion-2-inpainting",
    torch_dtype=torch.float16,
    revision="fp16",
    use_auth_token=True).to("cuda")

# ✅ Upload image and mask (Use Colab's file upload dialog)
from google.colab import files

print("Upload the BASE image:")
base_image = files.upload()
base_image_path = list(base_image.keys())[0]

print("Upload the MASK image:")
mask_image = files.upload()
mask_image_path = list(mask_image.keys())[0]

# ✅ Open both images
init_image = Image.open(base_image_path).convert("RGB").resize((512, 512))
mask_image = Image.open(mask_image_path).convert("L").resize((512, 512))

# ✅ Prompt for modification
prompt = input("Enter your prompt for inpainting (e.g., 'add a blue sky'): ")

# ✅ Run the pipeline
result = pipe(prompt=prompt, image=init_image, mask_image=mask_image).images[0]

# ✅ Show result
result.show()

# ✅ Save output
result.save("inpainted_result.png")
print("✅ Image saved as inpainted_result.png")
```
