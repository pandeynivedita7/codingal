## Art Supplies Billing Tool

```python
# PART 1: Define a function with no arguments to greet the customer
def greet_customer_art_supply():
    print("Welcome to the Art Supplies Billing Tool!")
    print("Let's get your art supplies tallied.")

# Call the greet_customer_art_supply function
greet_customer_art_supply()
```

### Get Item Details

```python
# PART 2: Ask for item details (name, price, quantity)
item_name = input("Enter the name of the art supply item: ")
price_per_item = float(input(f"Enter the price per {item_name} in dollars: "))
quantity_sold = int(input(f"Enter the quantity of {item_name} sold: "))
```

### Calculate Total Cost per Item

```python
# PART 3: Define a function that takes arguments and returns the total cost for an item
def calculate_item_total(price, quantity):
    total = price * quantity
    return total

# Call calculate_item_total and store the value it returns
item_total_cost = calculate_item_total(price_per_item, quantity_sold)
rounded_item_total = round(item_total_cost, 2)
print(f"Total for {item_name}: ${rounded_item_total}")
```

### Handle Payment and Change

```python
# PART 4: Ask how much money the customer paid
amount_paid_art_supply = float(input("Enter the amount paid by the customer: "))

# Define a function that takes arguments and returns the change due
def calculate_change_art_supply(paid, total):
    change = paid - total
    return change

# Call calculate_change_art_supply and store the value it returns
change_due_art_supply = calculate_change_art_supply(amount_paid_art_supply, rounded_item_total)
rounded_change_art_supply = round(change_due_art_supply, 2)
```

### Generate a Thank You Message

```python
# PART 5: Define a function that returns a thank you message based on quantity
def thank_you_message_art_supply(quantity):
    if quantity > 5:
        return "That's a lot of art supplies! Thanks for your bulk purchase!"
    elif quantity > 1:
        return "Thanks for your purchase, hope you create something great!"
    else:
        return "Thanks for stopping by!"

# Call thank_you_message_art_supply and store the value it returns
closing_message_art_supply = thank_you_message_art_supply(quantity_sold)
```

### Print the Final Receipt

```python
# PART 6: Print the final art supplies receipt
print("\n===== ART SUPPLIES RECEIPT ====")
print(f"Item: {item_name}")
print(f"Price Per Item: ${price_per_item}")
print(f"Quantity Sold: {quantity_sold}")
print(f"Total Cost: ${rounded_item_total}")
print(f"Amount Paid: ${amount_paid_art_supply}")
print(f"Change Due: ${rounded_change_art_supply}")
print(closing_message_art_supply)
print("===============================")
```

```python
# PART 1: Define a function with no arguments to greet the customer
def greet_customer():
    print("Welcome to the Lemonade Stand!")
    print("Fresh lemonade, made just for you.")

# PART 2: Call the greet_customer function
greet_customer()

# PART 3: Ask for the price per cup and the number of cups sold
price_per_cup = float(input("Enter the price per cup in dollars: "))
cups_sold = int(input("Enter the number of cups sold: "))

# PART 4: Define a function that takes arguments and returns the total cost
def calculate_total(price, cups):
    total = price * cups
    return total

# PART 5: Call calculate_total and store the value it returns
total_cost = calculate_total(price_per_cup, cups_sold)

# PART 6: Use a built-in function to round the total, then print it
rounded_total = round(total_cost, 2)
print("Total Cost:", rounded_total)

# PART 7: Ask how much money the customer paid
amount_paid = float(input("Enter the amount paid by the customer: "))

# PART 8: Define a function that takes arguments and returns the change due
def calculate_change(paid, total):
    change = paid - total
    return change

# PART 9: Call calculate_change and store the value it returns
change_due = calculate_change(amount_paid, rounded_total)
rounded_change = round(change_due, 2)

# PART 10: Define a function that returns a thank you message based on cups sold
def thank_you_message(cups):
    if cups >= 5:
        return "Wow, big order! Thanks so much for your support!"
    else:
        return "Thanks for stopping by the stand!"

# PART 11: Call thank_you_message and store the value it returns
closing_message = thank_you_message(cups_sold)

# PART 12: Print the final lemonade stand receipt
print("")
print("===== LEMONADE STAND RECEIPT =====")
print("Price Per Cup:", price_per_cup)
print("Cups Sold:", cups_sold)
print("Total Cost:", rounded_total)
print("Amount Paid:", amount_paid)
print("Change Due:", rounded_change)
print(closing_message)
print("===================================")
```
