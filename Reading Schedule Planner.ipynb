```python
import tkinter as tk
from tkinter import messagebox

def open_planner():
    # Create the separate Toplevel window
    planner_window = tk.Toplevel(root)
    planner_window.title("Plan Your Reading")
    planner_window.geometry("300x250")

    # Input field for Total Pages
    tk.Label(planner_window, text="Total Pages in Book:").pack(pady=5)
    entry_total = tk.Entry(planner_window)
    entry_total.pack()

    # Input field for Pages Per Day
    tk.Label(planner_window, text="Pages to Read Daily:").pack(pady=5)
    entry_daily = tk.Entry(planner_window)
    entry_daily.pack()

    # Label to display the calculated result
    result_label = tk.Label(planner_window, text="", fg="blue", font=("Arial", 10, "bold"))
    result_label.pack(pady=15)

    # Calculation function with try/except error handling
    def calculate_schedule():
        try:
            total_pages = int(entry_total.get())
            daily_pages = int(entry_daily.get())

            if daily_pages <= 0 or total_pages <= 0:
                messagebox.showerror("Error", "Numbers must be greater than zero!")
                return

            # Floor division for full days and Modulo for leftover pages
            days = total_pages // daily_pages
            leftover = total_pages % daily_pages

            # Formatting the result output
            output = f"Complete Days: {days} days\n"
            if leftover > 0:
                output += f"Final Day Pages: {leftover} pages"
            else:
                output += "No remaining pages!"

            result_label.config(text=output)

        except ValueError:
            messagebox.showerror("Invalid Input", "Please enter valid whole numbers.")

    # Calculate Button
    tk.Button(planner_window, text="Calculate", command=calculate_schedule).pack(pady=5)

# --- Main Window Setup ---
root = tk.Tk()
root.title("Reading Schedule Planner")
root.geometry("350x150")

tk.Label(root, text="Welcome to Reading Schedule Planner", font=("Arial", 12, "bold")).pack(pady=20)
tk.Button(root, text="Open Planner Window", command=open_planner, bg="lightgreen").pack(pady=10)

root.mainloop()
```
