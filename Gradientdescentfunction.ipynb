```python
# Importing required libraries
from numpy import asarray, arange #asarray: Converts lists into NumPy arrays. arange: Generates an array with regularly spaced values.
from numpy.random import rand # Generates random values between 0 and 1.
from matplotlib import pyplot #Used for plotting graphs (from matplotlib).

# Define the objective function: f(x) = x^2
def objective(x):# def funname(parmeters):
    return x**2.0

# Derivative of the objective function: f'(x) = 2x
def derivative(x):
    return x * 2.0

# Gradient descent implementation algorithm
def gradient_descent(objective, derivative, bounds, n_iter, step_size):# min function,f`(x)=2x bounds range between -1 to+1 no of steps
    solutions, scores = list(), list()  # to store solutions and scores

    # Generate initial point within the given bounds
    solution = bounds[:, 0] + rand(len(bounds)) * (bounds[:, 1] - bounds[:, 0])

    # Run the gradient descent loop for n_iter times
    for i in range(n_iter):# for var in range(start,stop,steps)
        gradient = derivative(solution)                   # Compute gradient
        solution = solution - step_size * gradient        # Take step in negative gradient direction
        solution_eval = objective(solution)               # Evaluate the solution
        solutions.append(solution)                        # Store the solution
        scores.append(solution_eval)                      # Store the objective value
        print('>%d f(%s) = %.5f' % (i, solution, solution_eval))  # Print progress

    return [solutions, scores]

# ---- MAIN EXECUTION ----

# Define search bounds for x (from -1.0 to 1.0)
bounds = asarray([[-1.0, 1.0]])

# Total number of iterations
n_iter = 30

# Step size (learning rate)
step_size = 0.1

# Run gradient descent
solutions, scores = gradient_descent(objective, derivative, bounds, n_iter, step_size)

# Plotting the objective function
inputs = arange(bounds[0, 0], bounds[0, 1]+0.1, 0.1)  # Inputs from -1.0 to 1.1
results = objective(inputs)  # f(x) = x^2

# Plot the objective function curve
pyplot.plot(inputs, results)

# Overlay the solution points found by gradient descent
pyplot.plot(solutions, scores, '.-', color='red')

# Show plot
pyplot.show()
```
