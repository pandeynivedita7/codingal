```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sns as sns
from scipy import stats  # Library for statistical methods

data = pd.read_csv('Titanic Dataset.csv')
data.head()

"""####**Descriptive Statistics using scipy.stats**"""

# 1. Clean data by removing missing values (NaN) to prevent calculation issues
age_clean = data['Age'].dropna()
fare_clean = data['Fare'].dropna()

# 2. Mean Calculation (Arithmetic Mean)
# Note: scipy.stats does not have a basic arithmetic mean function because it relies on numpy.mean,
# but it provides special means like tmean (trimmed mean) or gmean (geometric mean).
mean_age = np.mean(age_clean)
mean_fare = np.mean(fare_clean)
print("Mean Age is - ", mean_age)
print("Mean Fare is - ", mean_fare)

# 3. Median Calculation
# Note: scipy.stats relies on numpy.median for the base median calculation.
median_age = np.median(age_clean)
median_fare = np.median(fare_clean)
print("Median Age is - ", median_age)
print("Median Fare is - ", median_fare)

# 4. Mode Calculation using scipy.stats
# stats.mode returns an object containing an array of the most frequent values and their counts.
mode_age_result = stats.mode(age_clean, keepdims=True)
mode_fare_result = stats.mode(fare_clean, keepdims=True)

print("Mode Age is - ", mode_age_result.mode[0], " (Appears ", mode_age_result.count[0], " times)")
print("Mode Fare is - ", mode_fare_result.mode[0], " (Appears ", mode_fare_result.count[0], " times)")
```
