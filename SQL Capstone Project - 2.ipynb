```python
-- Create Restaurant table
CREATE TABLE IF NOT EXISTS Restaurant (
  name TEXT,
  neighborhood TEXT,
  cuisine TEXT,
  review REAL,
  price TEXT,
  health TEXT
);

-- Insert data into the Restaurant table
INSERT INTO Restaurant (name, neighborhood, cuisine, review, price, health)
VALUES
  ('Peter', 'Brooklyn', 'Steak', 4.4, '$$$$', 'A'),
  ('Jongro', 'Midtown', 'Korean', 3.5, '$$', 'A'),
  ('Pocha', 'Midtown', 'Pizza', 4.0, '$$$', 'B'),
  ('Lighthouse', 'Queens', 'Chinese', 3.9, '$', 'A'),
  ('Minca', 'Downtown', 'American', 4.6, '$$$', ''),
  ('Marea', 'Chinatown', 'Chinese', 3.0, '$$', ''),
  ('Dirty Candy', 'Uptown', 'Italian', 4.9, '$$$$', 'B'),
  ('Di Fara Pizza', 'Brooklyn', 'Pizza', 3.8, '$$', 'A'),
  ('Golden Unicorn', 'Uptown', 'Italian', 3.8, '$$', 'A');

-- 1) Selects all distinct neighborhood values from the Restaurant table
SELECT DISTINCT neighborhood
FROM Restaurant;

-- 2) Selects all distinct cuisine types from the Restaurant table
SELECT DISTINCT cuisine
FROM Restaurant;

-- 3) Selects all columns for restaurants where the cuisine is 'Chinese'
SELECT *
FROM Restaurant
WHERE cuisine = 'Chinese';

-- 4) Selects all columns for restaurants with a review score of 4.0 or higher
SELECT *
FROM Restaurant
WHERE review >= 4.0 ;

-- 5) Selects all columns for Italian restaurants with a price range of '$$' or '$$$'
SELECT *
FROM Restaurant
WHERE cuisine = 'Italian'
  AND price IN ('$$', '$$$');

-- 6) Selects all columns for restaurants with an exact price of '$$$'
SELECT *
FROM Restaurant
WHERE price = '$$$';

-- 7) Selects all columns for restaurants where the name contains the substring 'Candy'
SELECT *
FROM Restaurant
WHERE name LIKE '%Candy%';

-- 8) Selects all columns for restaurants located in 'Midtown', 'Downtown', or 'Chinatown'
SELECT *
FROM Restaurant
WHERE neighborhood IN ('Midtown', 'Downtown', 'Chinatown');

-- 9) Selects all columns for restaurants where the health grade is empty or NULL (indicating pending)
SELECT *
FROM Restaurant
WHERE health = '' OR health IS NULL;

-- 10) Selects the top 4 restaurants ordered by their review score in descending order
SELECT *
FROM Restaurant
ORDER BY review DESC
LIMIT 4;
```
