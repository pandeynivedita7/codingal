```python
# ATM Cash Dispenser
# uppper(), strip()
# This script simulates a simple ATM cash dispenser.
print("=== ATM Cash Dispenser ===\n") # Prints a header for the ATM application.

# Initialize variables to keep track of dispensed notes and ATM session statistics.
total_100 = total_50 = total_20 = total_10 = total_5 = total_1 = 0 # Counters for each denomination.
customers_served = 0 # Counter for the number of customers served.
total_dispensed = 0 # Sum of all money dispensed.

serving = True # Flag to control the main ATM loop.
while serving: # Main loop: continues as long as 'serving' is True (for multiple customers).
    name = input("Enter customer name: ") # Prompts for customer's name.
    amount = int(input(f"Hello {name}! Enter withdrawal amount: ")) # Prompts for withdrawal amount and converts to integer.

    if amount <= 0: # Checks if the entered amount is valid (positive).
        print("Invalid amount. Please enter a positive number.\n") # Error message for invalid amount.
        continue # Skips to the next iteration of the 'while serving' loop.

    print(f"\nDispensing {amount} units for {name}:") # Informs about the current transaction.
    remaining = amount # Amount left to dispense.
    idx = 1 # Index to iterate through denominations.

    while idx <= 6: # Inner loop: iterates through each denomination (100, 50, 20, 10, 5, 1).
        # Determine the current denomination value based on the index.
        if idx == 1: value = 100
        elif idx == 2: value = 50
        elif idx == 3: value = 20
        elif idx == 4: value = 10
        elif idx == 5: value = 5
        else: value = 1 # For idx = 6, value is 1.

        count = remaining // value # Calculates how many notes of the current 'value' can be dispensed.
        if count > 0: # If at least one note of this denomination can be dispensed.
            print(f"  {count} x {value}-unit note(s) = {count * value}") # Prints details of notes dispensed.
            remaining -= count * value # Subtracts the value of dispensed notes from 'remaining'.

            # Updates the total count for the respective denomination.
            if value == 100: total_100 += count
            elif value == 50: total_50 += count
            elif value == 20: total_20 += count
            elif value == 10: total_10 += count
            elif value == 5: total_5 += count
            else: total_1 += count
        idx += 1 # Moves to the next denomination.

    customers_served += 1 # Increments the count of customers served.
    total_dispensed += amount # Adds the current withdrawal amount to the total dispensed.
    print(f"Transaction complete, {name}!\n") # Confirms the transaction is complete.

    again = input("Next customer? (yes/no): ").strip().lower() # Asks if there's another customer.
    if again != "yes": # If the input is anything other than "yes".
        serving = False # Sets the flag to False to exit the main loop.

print("\n=== Daily Denomination Report ===") # Prints a header for the daily report.
for slot in range(1, 7): # Loop to iterate through each denomination for the report.
    # Assigns the correct value and total count for reporting.
    if slot == 1: value, total = 100, total_100
    elif slot == 2: value, total = 50, total_50
    elif slot == 3: value, total = 20, total_20
    elif slot == 4: value, total = 10, total_10
    elif slot == 5: value, total = 5, total_5
    else: value, total = 1, total_1

    if total > 0: # If any notes of this denomination were dispensed.
        print(f"  {value}-unit notes dispensed : {total} ", end="") # Prints the denomination and count.
        for note in range(total): # Inner loop to print '=' symbols for visual representation.
            print("=", end="") # Prints '=' without a newline.
        print() # Prints a newline after all '=' symbols for a denomination.

print(f"\nCustomers served : {customers_served}") # Prints the total number of customers served.
print(f"Total dispensed  : {total_dispensed} units") # Prints the total amount of money dispensed.
print("ATM session closed. Goodbye!") # Farewell message.
```
