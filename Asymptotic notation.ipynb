```python
# My Quiz Result Searcher

## Outline
In this assignment, you will build a Quiz Result Searcher using Python. We will compare three different ways of working with quiz scores:
1. **Direct Access**
2. **Linear Search**
3. **Pair Comparison**

Through these examples, we will understand asymptotic analysis, Big-O, Omega, Theta, best case, average case, worst case, and the differences between $O(1)$, $O(n)$, and $O(n^2)$.
```

```python
import time
import random

# 1. Direct Access - O(1) Complexity
# Accessing a student's score directly using their ID if indexable (or dict lookup)
def direct_access(scores_dict, student_id):
    try:
        return scores_dict[student_id]
    except KeyError:
        return None

# 2. Linear Search - O(n) Complexity
# Searching for a specific score in a list of scores
def linear_search(scores_list, target_score):
    for idx, score in enumerate(scores_list):
        if score == target_score:
            return idx  # Found at index idx
    return -1  # Not found

# 3. Pair Comparison - O(n^2) Complexity
# Finding if any two students have the exact same score
def find_duplicate_scores(scores_list):
    duplicates = []
    n = len(scores_list)
    for i in range(n):
        for j in range(i + 1, n):
            if scores_list[i] == scores_list[j]:
                duplicates.append((i, j))
    return duplicates
```

```python
# Generate mock data
num_students = 1000
random.seed(42)
quiz_scores = [random.randint(50, 100) for _ in range(num_students)]
scores_dictionary = {f"student_{i}": score for i, score in enumerate(quiz_scores)}

# --- Test 1: Direct Access ---
start = time.perf_counter()
score = direct_access(scores_dictionary, "student_500")
end = time.perf_counter()
print(f"Direct Access (O(1)): Found score {score} in {end - start:.8f} seconds.")

# --- Test 2: Linear Search ---
start = time.perf_counter()
index = linear_search(quiz_scores, 100)
end = time.perf_counter()
print(f"Linear Search (O(n)): Found score 100 at index {index} in {end - start:.8f} seconds.")

# --- Test 3: Pair Comparison ---
# We use a smaller subset to avoid long execution times due to O(n^2) scaling
subset_scores = quiz_scores[:300]
start = time.perf_counter()
duplicates = find_duplicate_scores(subset_scores)
end = time.perf_counter()
print(f"Pair Comparison (O(n^2)) on {len(subset_scores)} elements: Found {len(duplicates)} duplicate pairs in {end - start:.8f} seconds.")
```

```python
scores = [85, 90, 75, 95, 80]
target = int(input("Enter quiz score to search: "))

# 1. Direct Access - O(1)
print("Direct access:", scores[2])

# 2. Linear Search - O(n)
found = False
for score in scores:
    if score == target:
        found = True
        break
print("Linear search:", found)

# 3. Pair Comparison - O(n^2)
for i in range(len(scores)):
    for j in range(i + 1, len(scores)):
        if scores[i] == scores[j]:
            print("Duplicate scores:", scores[i])
```
