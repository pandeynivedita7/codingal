```python
# PART 1: Create the parent class with shared family traits
# This defines a base class called `FamilyMember`.
class FamilyMember:
    # The __init__ method is the constructor. It initializes new objects of this class.
    # It takes `eye_color` and `height_cm` as arguments and stores them as attributes.
    def __init__(self, eye_color, height_cm):
        self.eye_color = eye_color
        self.height_cm = height_cm

    # The show_traits method prints the eye color and height of the family member.
    def show_traits(self):
        print("Eye Color:", self.eye_color)
        print("Height (cm):", self.height_cm)

# PART 2: Create the child class that inherits from FamilyMember
# This defines a class `Kid` that inherits from `FamilyMember`.
# Inheritance means `Kid` automatically gets all the methods and attributes of `FamilyMember`.
class Kid(FamilyMember):

    # PART 3: Give Kid its own details, plus the inherited traits
    # The Kid's constructor takes `name` and `age` (specific to Kid)
    # as well as `eye_color` and `height_cm` (inherited from FamilyMember).
    def __init__(self, name, age, eye_color, height_cm):
        self.name = name # Store the kid's name
        self.age = age   # Store the kid's age
        # `super().__init__` calls the constructor of the parent class (FamilyMember)
        # to initialize the inherited traits (`eye_color`, `height_cm`).
        super().__init__(eye_color, height_cm)

    # PART 4: Override show_traits to add the kid's own details too
    # This method overrides the `show_traits` method from the parent class.
    # It first prints the kid's name and age.
    def show_traits(self):
        print("Name:", self.name)
        print("Age:", self.age)
        # Then, it calls the `show_traits` method of the parent class
        # using `super().show_traits()` to print the inherited traits.
        super().show_traits()

    # PART 5: Add a brand new method that only Kid has
    # This is a new method unique to the `Kid` class. It takes a hobby and prints it.
    def favorite_hobby(self, hobby):
        print(self.name, "loves", hobby)

# PART 6: Create a Kid object with real family trait values
# An instance of the `Kid` class is created, named 'Maya', aged 10, with brown eyes and 140 cm height.
child = Kid("Maya", 10, "brown", 140)

# PART 7: Call the overridden method and the new method
# Call the `show_traits` method on the `child` object. This will use the overridden method in `Kid`.
child.show_traits()
# Call the `favorite_hobby` method on the `child` object, which is unique to `Kid`.
child.favorite_hobby("painting")

# PART 8: Check whether Kid is really a subclass of FamilyMember
# The `issubclass()` function checks if `Kid` is a subclass of `FamilyMember` and prints the boolean result.
print("Is Kid a subclass of FamilyMember?", issubclass(Kid, FamilyMember))
```

```python
# PART 1: Create the parent class `Vehicle` with shared traits
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"Make: {self.make}")
        print(f"Model: {self.model}")
        print(f"Year: {self.year}")

# PART 2: Create the child class `Car` that inherits from `Vehicle`
class Car(Vehicle):
    # PART 3: Give Car its own details, plus the inherited traits
    def __init__(self, make, model, year, num_doors, fuel_type):
        self.num_doors = num_doors
        self.fuel_type = fuel_type
        # Call the parent class's constructor to initialize inherited traits
        super().__init__(make, model, year)

    # PART 4: Override `display_info` to add the car's own details
    def display_info(self):
        # Call the parent class's `display_info` method first
        super().display_info()
        print(f"Number of Doors: {self.num_doors}")
        print(f"Fuel Type: {self.fuel_type}")

    # PART 5: Add a brand new method that only `Car` has
    def start_engine(self):
        print(f"The {self.make} {self.model}'s engine is starting...")

# PART 6: Create a `Car` object
my_car = Car("Toyota", "Camry", 2022, 4, "Gasoline")

# PART 7: Call the overridden method and the new method
print("\n--- Car Information ---")
my_car.display_info()

print("\n--- Car Action ---")
my_car.start_engine()

# PART 8: Check whether `Car` is really a subclass of `Vehicle`
print("\nIs Car a subclass of Vehicle?", issubclass(Car, Vehicle))
```
