A company uses an automated system to detect fraudulent transactions.
- 10% of all transactions are fraudulent.
- If a transaction is fraudulent, the system flags it 90% of the time.
- If a transaction is not fraudulent, the system still flags it 5% of the time.
Question: If a transaction is flagged by the system, what is the probability that it is actually fraudulent?
Use:
\[
P(\text{Fraud} \mid \text{Flagged})
=
\frac{P(\text{Flagged} \mid \text{Fraud}) \times P(\text{Fraud})}
{P(\text{Flagged})}
\]
where:
\[
P(\text{Flagged})
=
P(\text{Flagged} \mid \text{Fraud})P(\text{Fraud})
+
P(\text{Flagged} \mid \text{Not Fraud})P(\text{Not Fraud})
\]
Try solving it in Python using:

```python
P_fraud = 0.10
P_not_fraud = 0.90

P_flagged_given_fraud = 0.90
P_flagged_given_not_fraud = 0.05
# Total probability that a transaction is flagged
P_flagged = (
    P_flagged_given_fraud * P_fraud
    + P_flagged_given_not_fraud * P_not_fraud
)
# Bayes' Theorem
P_fraud_given_flagged = (
    P_flagged_given_fraud * P_fraud
) / P_flagged
# Display the result
print("Probability of Flagged Transaction:", P_flagged)
print("Probability of Fraud given Flagged:", P_fraud_given_flagged)
print("Percentage:", round(P_fraud_given_flagged * 100, 2), "%")
```
