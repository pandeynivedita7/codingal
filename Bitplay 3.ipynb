```python
# Sign detection — used inside the divide function
# (a < 0) ^ (b < 0) is True when exactly one of a, b is negative

print((50 < 0) ^ (2 < 0))    # False ^ False = False (positive result)
print((-50 < 0) ^ (2 < 0))   # True  ^ False = True  (negative result)
print((50 < 0) ^ (-2 < 0))   # False ^ True  = True  (negative result)
print((-50 < 0) ^ (-2 < 0))  # True  ^ True  = False (positive result)
# Output:
# False
# True
# True
# False
```

```python
# PART 4: Divide without /
def divide(a, b):
    negative = (a < 0) ^ (b < 0)   # True if exactly one is negative
    a = abs(a)                       # work with positive numbers
    b = abs(b)
    count = 0
    while a >= b:                   # subtract b from a, count the steps
        a = a - b
        count = count + 1
    if negative:                    # restore the correct sign
        count = -count
    return count

print('Divide without /:')
print('  50 / 2  =', divide(50, 2))
print('  72 / 3  =', divide(72, 3))
print(' -50 / 2  =', divide(-50, 2))
print('  50 / -2 =', divide(50, -2))
# Output:
#   50 / 2  = 25
#   72 / 3  = 24
#  -50 / 2  = -25
#   50 / -2 = -25
```

```python
# Program to swap two numbers without using 3rd variable

def swap1(a,b):

    # Code to swap 'a' and 'b'
    a = a ^ b #xor
    b = a ^ b
    a = a ^ b
    print ("After Swapping: a = ", a, " b =", b)

def swap2(a, b):

    a = (a & b) + (a | b)
    b = a + (~b) + 1
    a = a + (~b) + 1
    print ("After Swapping: a = ", a, " b =", b)

swap1(1,2)
swap2(1,2)
```

```python
# Program to divide two numbers without using the division operator

def divide(ourDividend, ourDivisor):

    # Check if divisor is +ve or -ve
    sign = (-1 if((ourDividend < 0) ^
                (ourDivisor < 0)) else 1);

    # Make both positive
    ourDividend = abs(ourDividend);
    ourDivisor = abs(ourDivisor);

    quotientNumber = 0
    tempNumber = 0

    # Go from 31 to 0 and accumulate all valid bits

    for i in range(31, -1, -1):# for(start,end,step):

        if (tempNumber + (ourDivisor << i) <= ourDividend):
            tempNumber += ourDivisor << i
            quotientNumber |= 1 << i

    # Assuming the sign value computed earlier is -1, negate the quotient value
    if sign ==-1 :
        quotientNumber=-quotientNumber
    return quotientNumber


a = int(input("Enter a for a/b : "))
b = int(input("Enter b for a/b : "))
print("Result of ",a,"/",b,"is",divide(a, b))
```

```python
# Program to find the longest consecutive 1's in the binary representation of a number

def longest_consecutive_ones(n):
    binary = bin(n)[2:]          # convert to binary (remove '0b')
    max_ones = 0
    count = 0

    for bit in binary:
        if bit == '1':
            count += 1
            max_ones = max(max_ones, count)
        else:
            count = 0

    return binary, max_ones


# Input
n = int(input("Enter a number: "))

# Output
binary, result = longest_consecutive_ones(n)
print("Binary representation:", binary)
print("Longest consecutive 1's:", result)
```
