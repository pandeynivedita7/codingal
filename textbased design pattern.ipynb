```python
# Alphabet Design - Letter A

rows = 7# 0 to6

for i in range(rows):# 7
    for j in range(7): # row 0 to 7 and colunms 0 to
        if ((j == 0 or j == 6) and i != 0) or \
           (i == 0 and j != 0 and j != 6) or \
           (i == 3):
            print("*", end="")
        else:
            print(" ", end="")
    print()
```
