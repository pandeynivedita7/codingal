````python
# Project: Calculate Number of Vowels in a Given String

## Project Outline

In this project, you will create a simple **Flask web application** that accepts a string from the user and calculates the total number of vowels present in the entered text. The application will provide a user-friendly interface where users can enter a string, submit it, and immediately view the number of vowels.

The project combines basic **Python programming concepts** with introductory **Flask web development**, helping students understand how user input can be collected from a web page, processed on the server, and returned as a result.

---

## Project Objective

The objective of this project is to develop a Flask-based application that can:

* Accept a string from the user.
* Process the entered string using Python.
* Identify vowels: **a, e, i, o, u**.
* Handle both uppercase and lowercase vowels.
* Calculate the total number of vowels.
* Display the original string and vowel count.
* Handle empty or invalid input appropriately.

---

## Learning Outcomes

After completing this project, students will be able to:

1. **Explain** the basic structure and working of a Flask web application.
2. **Apply** Python string-processing techniques to analyze user-provided text.
3. **Develop** a Flask application that accepts, processes, and displays user input.
4. **Use** conditional statements and loops to identify vowels in a string.
5. **Demonstrate** the use of HTML forms and Flask templates for user interaction.

---

## Software and Tools Required

* Python 3.x
* Flask
* HTML
* CSS
* Visual Studio Code or another Python IDE
* Web browser
* Command Prompt/Terminal

---

## Functional Requirements

The application should provide the following functionality:

* A text input field for entering a string.
* A **Calculate Vowels** button.
* Processing of the submitted string.
* Counting of uppercase and lowercase vowels.
* Display of the total vowel count.
* Display of the entered string.
* Validation for empty input.
* A simple and readable user interface.

---

## Suggested Project Structure

```text
vowel_counter/
│
├── app.py
│
├── templates/
│   └── index.html
│
└── static/
    └── style.css
```

---

## Step 1 – Install Flask

Open the terminal or command prompt and install Flask:

```bash
pip install flask
```

---

## Step 2 – Create the Flask Application

Create a file named **app.py**.

```python
from flask import Flask, render_template, request

app = Flask(__name__)

def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0

    for character in text:
        if character in vowels:
            count += 1

    return count


@app.route("/", methods=["GET", "POST"])
def index():
    vowel_count = None
    entered_text = ""
    error = None

    if request.method == "POST":
        entered_text = request.form.get("text", "")

        if entered_text.strip() == "":
            error = "Please enter a valid string."
        else:
            vowel_count = count_vowels(entered_text)

    return render_template(
        "index.html",
        vowel_count=vowel_count,
        entered_text=entered_text,
        error=error
    )


if __name__ == "__main__":
    app.run(debug=True)
```

---

## Step 3 – Create the HTML Template

Create a folder named **templates** and add a file named **index.html**.

```html
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Vowel Counter</title>
    <link rel="stylesheet"
          href="{{ url_for('static', filename='style.css') }}">
</head>

<body>

    <div class="container">

        <h1>Vowel Counter</h1>

        <p>Enter a string below to calculate the number of vowels.</p>

        <form method="POST">

            <input
                type="text"
                name="text"
                placeholder="Enter a string"
                value="{{ entered_text }}"
                required
            >

            <button type="submit">
                Calculate Vowels
            </button>

        </form>

        {% if error %}
            <p class="error">{{ error }}</p>
        {% endif %}

        {% if vowel_count is not none %}

            <div class="result">

                <h2>Result</h2>

                <p>
                    <strong>Entered String:</strong>
                    {{ entered_text }}
                </p>

                <p>
                    <strong>Number of Vowels:</strong>
                    {{ vowel_count }}
                </p>

            </div>

        {% endif %}

    </div>

</body>
</html>
```

---

## Step 4 – Add CSS Styling

Create a folder named **static** and add a file named **style.css**.

```css
body {
    font-family: Arial, sans-serif;
    background: #f4f4f4;
    margin: 0;
    padding: 0;
}

.container {
    width: 500px;
    margin: 100px auto;
    background: white;
    padding: 30px;
    border-radius: 10px;
    text-align: center;
    box-shadow: 0 4px 10px rgba(0, 0, 0, 0.1);
}

h1 {
    margin-bottom: 10px;
}

input {
    width: 90%;
    padding: 12px;
    margin: 15px 0;
    font-size: 16px;
}

button {
    padding: 12px 20px;
    border: none;
    cursor: pointer;
    font-size: 16px;
    border-radius: 5px;
}

.result {
    margin-top: 25px;
    padding: 15px;
    background: #f0f0f0;
    border-radius: 5px;
}

.error {
    margin-top: 15px;
    font-weight: bold;
}
```

---

## Step 5 – Run the Application

Open the terminal in the project directory and execute:

```bash
python app.py
```

Flask will start the development server.

Open the local address displayed in the terminal in your web browser.

---

## Application Workflow

The application follows this process:

```text
User enters a string
        ↓
User clicks "Calculate Vowels"
        ↓
HTML form sends data to Flask
        ↓
Flask receives the input
        ↓
Python checks each character
        ↓
Vowels are identified
        ↓
Vowel count is calculated
        ↓
Result is sent to the HTML template
        ↓
Browser displays the result
```

---

## Algorithm

**Step 1:** Start the application.

**Step 2:** Accept a string from the user.

**Step 3:** Define the vowels as:

```text
a, e, i, o, u
```

**Step 4:** Initialize the vowel count to zero.

**Step 5:** Examine each character in the entered string.

**Step 6:** Check whether the character is a vowel.

**Step 7:** If it is a vowel, increase the count by one.

**Step 8:** Continue until all characters have been checked.

**Step 9:** Display the total number of vowels.

**Step 10:** End.

---

## Example

### Input

```text
Hello World
```

### Processing

The vowels are:

```text
e, o, o
```

### Output

```text
Number of Vowels: 3
```

---

## Additional Test Cases

| Input            | Expected Vowel Count |
| ---------------- | -------------------: |
| Hello            |                    2 |
| Education        |                    5 |
| Python           |                    1 |
| FLASK            |                    1 |
| Computer Science |                    6 |
| AEIOU            |                    5 |
| Rhythm           |                    0 |
| Networking       |                    3 |

---

## Testing Requirements

Students should test the application using:

* Strings containing only vowels.
* Strings containing no vowels.
* Uppercase characters.
* Lowercase characters.
* Mixed uppercase and lowercase characters.
* Strings containing spaces.
* Strings containing numbers and special characters.
* Empty input.

---

## Expected Output

The completed application should display:

**Vowel Counter**

```text
Enter a string:
[________________________]

[ Calculate Vowels ]
```

After submission:

```text
Entered String: Hello World

Number of Vowels: 3
```

---

## Possible Enhancements

Students can extend the project by:

* Displaying each vowel found in the string.
* Counting each vowel separately.
* Showing the number of consonants.
* Counting digits and special characters.
* Calculating the total number of words.
* Adding a Reset/Clear button.
* Improving the user interface using Bootstrap.
* Displaying results using charts or visual indicators.

---

## Project Deliverables

Students should submit:

1. `app.py`
2. `templates/index.html`
3. `static/style.css`
4. Screenshot of the running application
5. Screenshot showing sample input and output
6. Short explanation of the algorithm
7. Test cases and results

---

## Conclusion

This project demonstrates how **Python and Flask** can be combined to create a simple interactive web application. Students learn how to collect user input through an HTML form, process the input using Python programming logic, and dynamically display the result using Flask templates.

The project also reinforces fundamental programming concepts such as **strings, loops, conditional statements, functions, input validation, and basic web application development**.
````
