### 1. Student Details using Tuples

First, we'll store fixed student details in a tuple, which is an immutable sequence. Then, we'll demonstrate how to access these details.

```python
# My School Subject Planner

# Step 1: Store fixed student details in a tuple
student_details = ("Nivedita Pandey", 101, "Grade 10")

# Display student details
print("Student Details")
print("Name :", student_details[0])
print("Roll Number :", student_details[1])
print("Class :", student_details[2])

# Step 2: Create subject sets for different days
monday_subjects = {"Math", "Science", "English", "Computer"}
tuesday_subjects = {"Science", "English", "History", "Art"}

print("\nMonday Subjects:", monday_subjects)
print("Tuesday Subjects:", tuesday_subjects)

# Step 3: Modify sets
monday_subjects.add("Physical Education")
print("\nAfter Adding a Subject to Monday:")
print(monday_subjects)

monday_subjects.remove("Computer")
print("\nAfter Removing a Subject from Monday:")
print(monday_subjects)

# Step 4: Set Operations

# Common subjects
common_subjects = monday_subjects.intersection(tuesday_subjects)
print("\nCommon Subjects:", common_subjects)

# All subjects
all_subjects = monday_subjects.union(tuesday_subjects)
print("All Subjects:", all_subjects)

# Subjects only on Monday
only_monday = monday_subjects.difference(tuesday_subjects)
print("Subjects Only on Monday:", only_monday)

# Subjects only on Tuesday
only_tuesday = tuesday_subjects.difference(monday_subjects)
print("Subjects Only on Tuesday:", only_tuesday)
```
