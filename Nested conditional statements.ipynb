```python
# Custom Ride Builder
# File  : ride_builder.py
# Lesson: SPCM2L1 — Nested Conditional Statements (Jr)

print("====================================")
print("      Welcome to Ride Builder!      ")
print("====================================")
print()

print("Step 1: Pick your vehicle")
print("  1 - Bike")
print("  2 - Car")
print()

choice = int(input("Enter 1 or 2: "))
print()

if choice == 1:# condition
    # Nested if-else — runs only when choice is 1
    print("Step 2: Pick your bike type")
    print("  1 - Scooty")
    print("  2 - Mountain Bike")
    print()

    bike_type = int(input("Enter 1 or 2: "))
    print()

    if bike_type == 1:
        print("You picked  : Scooty")
        print("Top speed   : 80 km/h")
        print("Best for    : City roads")
    else:
        print("You picked  : Mountain Bike")
        print("Top speed   : 40 km/h")
        print("Best for    : Off-road trails")

elif choice == 2:
    # Nested if-else — runs only when choice is 2
    print("Step 2: Pick your car type")
    print("  1 - Sedan")
    print("  2 - SUV")
    print()

    car_type = int(input("Enter 1 or 2: "))
    print()

    if car_type == 1:# TRue
        print("You picked  : Sedan")
        print("Seats       : 5 passengers")
        print("Best for    : Family trips")
    else:
        print("You picked  : SUV")# false
        print("Seats       : 7 passengers")
        print("Best for    : Off-road adventures")

else:
    print("That was not a valid choice.")
    print("Please enter 1 for Bike or 2 for Car.")

print()
print("====================================")
print("   Your custom ride is ready!       ")
print("   Enjoy the journey!               ")
print("====================================")
```

```python
# Holiday Activity Planner

print("===========================================")
print("      Welcome to Holiday Activity Planner!      ")
print("===========================================")
print()

print("Step 1: Choose your preferred holiday type:")
print("  1 - Beach Holiday")
print("  2 - Mountain Holiday")
print("  3 - City Break")
print()

holiday_choice = int(input("Enter 1, 2, or 3: "))
print()

if holiday_choice == 1:
    print("Step 2: What kind of beach activity are you looking for?")
    print("  1 - Relaxing by the sea")
    print("  2 - Water sports and adventure")
    print()

    beach_activity = int(input("Enter 1 or 2: "))
    print()

    if beach_activity == 1:
        print("Your plan: Find a quiet spot, read a book, and enjoy the sun. Perhaps a gentle swim!")
    else:
        print("Your plan: Go snorkeling, try surfing, or explore hidden coves with a kayak!")

elif holiday_choice == 2:
    print("Step 2: What season are you planning your mountain holiday for?")
    print("  1 - Winter (snow activities)")
    print("  2 - Summer (hiking and nature)")
    print()

    mountain_season = int(input("Enter 1 or 2: "))
    print()

    if mountain_season == 1:
        print("Your plan: Skiing, snowboarding, or a cozy evening by the fireplace in a cabin.")
    else:
        print("Your plan: Explore scenic hiking trails, enjoy mountain biking, or picnic with stunning views.")

elif holiday_choice == 3:
    print("Step 2: What is your main interest for a city break?")
    print("  1 - Culture and history")
    print("  2 - Food and nightlife")
    print()

    city_interest = int(input("Enter 1 or 2: "))
    print()

    if city_interest == 1:
        print("Your plan: Visit museums, historical landmarks, and explore art galleries. Don't forget a walking tour!")
    else:
        print("Your plan: Discover local eateries, experience vibrant nightlife, and find hidden bars.")

else:
    print("That was not a valid choice. Please enter 1, 2, or 3 for your holiday type.")

print()
print("===========================================")
print("   Your personalized holiday plan is ready!   ")
print("   Happy travels!                           ")
print("===========================================")
```
