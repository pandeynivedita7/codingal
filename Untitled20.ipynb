```python
import numpy as np
import ctypes

# Create a standard numpy array (e.g., 1D array of integers)
arr = np.array([10, 20, 30, 40, 50], dtype=np.int32)
print("Original NumPy Array:", arr)

# Get the memory address of the first element (pointer)
pointer_address = arr.__array_interface__['data'][0]
print(f"Memory Address (Pointer): {hex(pointer_address)}")
```

```python
# Access and modify the array's data using ctypes pointers
# Cast the memory address to an integer pointer
ptr = ctypes.cast(pointer_address, ctypes.POINTER(ctypes.c_int32))

# Read values via pointer
print("Reading values through the pointer:")
for i in range(len(arr)):
    print(f"Element {i}: {ptr[i]}")

# Modify an element using the pointer
ptr[2] = 99

print("\nModified Array (changes made via pointer):", arr)
```
