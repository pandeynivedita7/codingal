```python
True
False
```

* `or`
* `not`

# 4. IF-ELIF-ELSE

## Program 7: Grade Calculator

```python
marks = int(input("Enter your marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
elif marks >= 60:
    print("Grade D")
else:
    print("Grade F")
```

### Explain Execution

Suppose:

```text
marks = 85
```

Python checks:

```python
marks >= 90
```

False.

Then:

```python
marks >= 80
```

True.

Therefore:

```text
Grade B
```

is displayed and Python does not check the remaining conditions.

### Teaching Point

Order is extremely important.

Incorrect:

```python
if marks >= 60:
    print("D")
elif marks >= 90:
    print("A")
```

A student with `95` would incorrectly receive `D`.

# 5. Nested Conditional Statements

A **nested condition** means putting an `if` statement inside another `if`.

## Program 8: Login System

```python
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":

    if password == "python123":
        print("Login successful")
    else:
        print("Incorrect password")

else:
    print("User does not exist")
```

### Explain the Logic

Python first checks:

```python
username == "admin"
```

Only if this is true does Python check:

```python
password == "python123"
```

Students can visualize it as:

```text
Is username correct?
       |
       YES
       |
Is password correct?
   /          \
 YES          NO
 |             |
Login      Wrong password
```

# 10. WHILE Loops

A `while` loop repeats **while a condition remains true**.

## Program 13: Counting with WHILE

```python
count = 1

while count <= 5:
    print(count)
    count = count + 1
```

Output:

```text
1
2
3
4
5
```

### Explain the Three Important Parts

Initialization:

```python
count = 1
```

Condition:

```python
count <= 5
```

Update:

```python
count = count + 1
```

Without the update, students may accidentally create an infinite loop.

# 12. FOR vs WHILE

Teach students this simple distinction:

**FOR**

Use when the number of repetitions is known.

```python
for i in range(10):
    print(i)
```

**WHILE**

Use when repetition depends on a condition.

```python
while password != "secret":
    password = input("Password: ")
```

# 19. Bug Program 4: Infinite Loop

Incorrect:

```python
count = 1

while count <= 5:
    print(count)
```

The value of `count` never changes.

Therefore the condition:

```python
count <= 5
```

remains `True` forever.

Correct:

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

# 21. Complete Program Development

The final objective should combine several concepts into one working application.

A good classroom project is a **Student Grade Management Program**.

## Program 18: Complete Student Result Program

```python
name = input("Enter student name: ")

total = 0

for i in range(1, 4):

    marks = float(input(f"Enter marks for subject {i}: "))

    while marks < 0 or marks > 100:
        print("Invalid marks.")
        marks = float(input("Enter marks between 0 and 100: "))

    total += marks

average = total / 3

print("\nStudent:", name)
print("Total:", total)
print("Average:", average)

if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)

if average >= 60:
    print("Result: Pass")
else:
    print("Result: Fail")
```

## Concepts Covered

This one program includes:

```text
input()
Data type conversion
Variables
Boolean logic
OR operator
for loop
while loop
validation
calculation
if
elif
else
comparison operators
formatted output
```

# 22. Another Final Program: ATM Simulation

This is particularly useful for a timed programming test.

```python
balance = 5000
correct_pin = "1234"

pin = input("Enter PIN: ")

if pin == correct_pin:

    while True:

        print("\n1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            print("Balance:", balance)

        elif choice == "2":
            amount = float(input("Enter deposit amount: "))

            if amount > 0:
                balance += amount
                print("Deposit successful")
            else:
                print("Invalid amount")

        elif choice == "3":
            amount = float(input("Enter withdrawal amount: "))

            if amount <= 0:
                print("Invalid amount")

            elif amount > balance:
                print("Insufficient balance")

            else:
                balance -= amount
                print("Withdrawal successful")

        elif choice == "4":
            print("Thank you.")
            break

        else:
            print("Invalid choice")

else:
    print("Incorrect PIN")
```

This is an excellent final teaching example because students see how different Python concepts work **together rather than independently**.

# Recommended Teaching Sequence

I would teach these concepts in this order:

1. **Boolean values and comparisons** — `True`, `False`, `>`, `<`, `==`, `!=`
2. **Data conversion** — `int()`, `float()`, `str()`, `bool()`
3. **Boolean operators** — `and`, `or`, `not`
4. **Simple `if`**
5. **`if-else`**
6. **`if-elif-else`**
7. **Nested conditions**
8. **Basic `for` loops**
9. **`range()`**
10. **Loop calculations and accumulators**
11. **`while` loops**
12. **Input validation using `while`**
13. **Nested loops**
14. **Star and number patterns**
15. **Debugging syntax and runtime errors**
16. **Debugging logical errors**
17. **Integrated mini-programs**
18. **Timed independent programming task**

For every program, have students follow the same cycle: **predict the output → trace variables → run the code → modify one requirement → write a similar program without looking at the original**. This builds actual programming ability rather than only code recognition.
