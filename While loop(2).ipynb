```python
# My Chore Checklist Countdown

# PART 1: Set today's total number of chores (no list needed)
# Initialize the total number of chores for the day.
total_chores = 4
# Store the initial total chore count to display at the end.
original_count = total_chores
# Print a welcoming message indicating the number of chores.
print(f"You have {original_count} chores to finish today!\n")


# PART 2: Keep a counter for completed chores and the current chore number
# Initialize a counter for completed chores to zero.
completed_count = 0
# Initialize the current chore number to 1, as chores are 1-indexed.
chore_num = 1

# PART 3: Repeat while there are still chores left to check off
# Start a loop that continues as long as there are chores remaining to be processed.
while chore_num <= total_chores:

    # PART 4: Work out the current chore's name from its number
    # Determine the name of the current chore based on its number.
    if chore_num == 1: next_chore = "Make your bed"
    elif chore_num == 2: next_chore = "Feed the pet"
    elif chore_num == 3: next_chore = "Take out the trash"
    else: next_chore = "Wash the dishes"

    # Prompt the user if they have finished the current chore.
    answer = input(f"Have you finished: {next_chore}? (yes/no): ")

    # PART 5: Only move on to the next chore once it is marked done
    # Check if the user confirmed completion of the chore.
    if answer == "yes":
        # Increment the count of completed chores.
        completed_count += 1
        # Move to the next chore number.
        chore_num += 1
        # Inform the user that the chore is completed.
        print("Great job! Chore completed.")
    else:
        # If not completed, ask the user to finish it and check again.
        print("Okay, finish it and check again!")

    # PART 6: Print how many chores remain after each check
    # Display the number of chores still remaining.
    print("Chores remaining:", total_chores - completed_count)
    # Print an empty line for better readability.
    print()

# PART 7: This only prints once every chore is marked done
# Print a message indicating all chores are complete after the loop finishes.
print("===== ALL CHORES COMPLETE! ====")
# Print a congratulatory message.
print("Great work finishing your entire checklist today!\n")

# PART 8: A safe look at what an infinite loop would look like
# Introduce a demonstration of an infinite loop (which will be safely broken).
print("Now let's safely peek at an infinite loop...")
# Initialize a test value that will keep the loop condition true.
test_value = 0
# Initialize a safety counter to prevent a true infinite loop.
safety_counter = 0
# Start a 'while' loop whose condition (`test_value <= 0`) remains true.
while test_value <= 0:
    # Print a message indicating the nature of an infinite loop.
    print("This condition never changes, so this would run forever!")
    # Increment the safety counter with each iteration.
    safety_counter += 1
    # Check if the safety counter has reached its limit.
    if safety_counter == 3:
        # Explain that the loop is being stopped intentionally.
        print("(Stopping here on purpose - a real infinite loop never stops on its own!)")
        # Use 'break' to exit the loop prematurely.
        break

# PART 9: Print the final chore checklist summary
# Print a header for the summary.
print("\n===== CHORE CHECKLIST SUMMARY ====")
# Display the total number of chores initially assigned.
print("Chores Assigned Today:", original_count)
# Display the total number of chores completed.
print("Chores Completed:", completed_count)
# Display the number of chores remaining (should be 0 if all completed).
print("Chores Remaining:", total_chores - completed_count)
# Print a separator for the summary.
print("======================================")
```
