```python
import numpy as np# multidim array

# Create first array: 0–8 reshaped to 3×3
a = np.arange(9, dtype=np.float_).reshape(3, 3)#0-8
#[[0. 1. 2.3 4 5 6 7 8]]     [0 1 2][3 4 5][6 7 8] a
print('First array:')
print(a)
print('\n')

# Create second array
b = np.array([10, 10, 10])
print('Second array:')
print(b)
print('\n')

print('Add the two arrays:')
print(np.add(a, b))#[0+10 1+10 2+10]=[10 11 12 ]
#[3+10 4+10 5+10]=[13 14 15]\ [6+10 7+10 8+10]=[16 17 18]

print('\n')

print('Subtract the two arrays:')
print(np.subtract(a, b))# a+b a-b a*b a/b nupmy they have inbuild methods
print('\n')

print('Multiply the two arrays:')
print(np.multiply(a, b))
print('\n')

print('Divide the two arrays:')
print(np.divide(a, b))
print('\n')
```

```python
import numpy as np # Import the NumPy library, commonly aliased as np

arr = np.array([40, 10, 30, 20]) # Create a NumPy array named 'arr' with the specified integer values

arr.sort() # Sort the 'arr' array in ascending order in-place
desc = arr[::-1] # Create a new array 'desc' by reversing the sorted 'arr' array, effectively creating a descending order

print(desc) # Print the 'desc' array, which now contains the elements in descending order
```
