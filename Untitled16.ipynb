```python
# game-stats-analyser.py
# Activity: Gaming Leaderboard Analyser
# Lesson: Introduction to Pandas

import pandas as pd# rows and column
import numpy as np # Import numpy for additional numerical operations

# PART 1 - Create a Pandas Series of top player scores
print('--- PART 1: Pandas Series ---')
# A Pandas Series is a one-dimensional labeled array capable of holding any data type.
# Here, we create a Series from a list of scores and assign custom string labels (player names) as its index.
scores = [98500, 87200, 76400, 65100, 54800]
players = pd.Series(scores, index=['NightWolf', 'StarBlaze', 'PixelKing', 'CyberFox', 'IronStorm'])
print('Player Scores Series:')
display(players)

# PART 2 - Create a DataFrame of gaming stats
print('
--- PART 2: Pandas DataFrame ---')
# A Pandas DataFrame is a two-dimensional labeled data structure with columns of potentially different types.
# It's similar to a spreadsheet or SQL table.
# We create a DataFrame from a dictionary where keys become column names and values are lists of data.
data = {
    'Player':  ['NightWolf', 'StarBlaze', 'PixelKing', 'CyberFox', 'IronStorm'],
    'Level':   [42, 38, 35, 30, 27],
    'Score':   [98500, 87200, 76400, 65100, 54800],
    'Wins':    [210, 185, 162, 140, 118]
}
df = pd.DataFrame(data)
print('Gaming Stats DataFrame:')
display(df)

# PART 3 - Access rows using .loc
print('
--- PART 3: Accessing Rows ---')
# .loc is primarily label-based indexing. It allows selection by labels of rows and columns.
# Here, since the DataFrame uses default integer indexing, .loc[0] accesses the first row.
print('Row 0 (top player) using .loc[0]:')
display(df.loc[0])
print('Rows 2 and 3 using .loc[2:3] (inclusive of end label):')
display(df.loc[2:3])

# PART 4 - Load leaderboard.csv and view the data
print('--- PART 4: Reading a CSV File ---')
# pd.read_csv() is used to load data from a CSV file into a DataFrame.
# Make sure 'leaderboard.csv' is in the same directory or provide the full path.
full_df = pd.read_csv('/content/sample_data')
print('First 5 rows (head) of full_df:')
display(full_df.head())
print('Last 3 rows (tail) of full_df:')
display(full_df.tail(3))
print('Dataset info (summary of DataFrame including data types and non-null values):')
full_df.info()

# PART 5 - Clean the data
print('
--- PART 5: Cleaning Data ---
')
# .dropna() removes rows (or columns) with any missing values (NaN).
print('Rows with missing values removed (dropna):')
clean_df = full_df.dropna()
display(clean_df)
print('Missing values filled with 0 (fillna):')
# .fillna() replaces missing values with a specified value.
filled_df = full_df.fillna(0)
display(filled_df)

# PART 6 - Basic Pandas Operations (on df DataFrame)
print('
--- PART 6: Basic Pandas Operations ---
')
# .describe() generates descriptive statistics of the DataFrame's numerical columns.
print('Descriptive statistics of df:')
display(df.describe())

# Selecting a single column as a Series
print('
Selecting the Score column:')
display(df['Score'])

# Selecting multiple columns as a new DataFrame
print('
Selecting Player and Wins columns:')
display(df[['Player', 'Wins']])

# Filtering rows based on a condition (Players with Score > 70000)
print('
Players with Score > 70000:')
display(df[df['Score'] > 70000])

# Sorting the DataFrame by Score in descending order
print('
DataFrame sorted by Score (descending):')
display(df.sort_values(by='Score', ascending=False))

# Adding a new column: Win_Ratio (Wins per Level)
# This demonstrates a simple element-wise operation to create a new feature.
df['Win_Ratio'] = df['Wins'] / df['Level']
print('
DataFrame with new Win_Ratio column:')
display(df)

# Grouping data: Calculate the average score per level
# .groupby() is used to split the data into groups based on some criteria, and then apply a function to each group.
print('
Average Score per Level:')
display(df.groupby('Level')['Score'].mean())

# PART 7 - Basic NumPy Operation
print('
--- PART 7: Basic NumPy Operation ---
')
# NumPy is fundamental for numerical computing in Python. It provides array objects and tools for working with them.
# Creating a NumPy array from a list
numpy_array_scores = np.array(scores)
print('NumPy array created from scores:')
print(numpy_array_scores)

# Performing a simple mathematical operation (e.g., adding 1000 to all scores)
increased_scores = numpy_array_scores + 1000
print('
Scores after adding 1000 using NumPy:')
print(increased_scores)
```
