```python
#Take input
print("Half Pyramid Pattern of Stars (*):")
n = int(input("enter the number of rows: "))#typecasting in int
#outer loop to handle number of rows 2 for loop
for i in range(n): #6
  #inner loop to handle number of columns
    for j in range(i+1):
      #display result
        print("* ", end="")
    print()
```

```python
#Take input
print("Left Triangle Pattern of Stars (*):")
n = int(input("enter the number of rows: "))#4

#outer loop to handle number of rows
for i in range(1, n + 1):#1
    #inner loop to print spaces
    for j in range(n - i):#3
        print("  ", end="")
    #inner loop to print stars
    for k in range(i):
        print("* ", end="")
    print()
```

```python
#Take input from user
rows = int(input("Please Enter the total Number of Rows  : "))
number = 1 #initialise by 1

print("Floyd's Triangle")
#outer loop for number of rows
for i in range(1, rows + 1):
  #inner loop for number of columns
    for j in range(1, i + 1):
      #display result
        print(number, end = '  ')
        number = number + 1
    print()
```

```python
#take input from user
rowSize = int(input("enter the number of rows: "))
if rowSize%2==0: #conditions
  halfDiamRow = int(rowSize/2)# even condition 2 4 6 8
else:
  halfDiamRow = int(rowSize/2)+1# odd condition 5
space = halfDiamRow-1
#loop for upper part
for i in range(1, halfDiamRow+1): #loop for rows
  for j in range(1, space+1): #loop for columns
    print(end=" ")
  space = space-1
  num = 1
  for j in range(2*i-1):
    print(end=str(num))
  #incerementing number at each column
    num = num+1
  print()
space = 1
#loop for lower part
for i in range(1, halfDiamRow): #loop for rows
  for j in range(1, space+1):  #loop for columns
    print(end=" ")
  space = space+1
  num = 1
  for j in range(1, 2*(halfDiamRow-i)):
    print(end=str(num)) #display result
  #incerementing number at each column
    num = num+1
  print()
```
