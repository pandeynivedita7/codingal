```python
"""
AI Image Stylist: Your Signature Look
Uses Hugging Face InferenceClient + Stable Diffusion XL + Pillow

Install:
    pip install huggingface_hub pillow

Files:
    config.py  -> contains HF_API_KEY = "your_token"

What it does:
1. Takes a text prompt from the user
2. Generates an image using Stable Diffusion XL
3. Applies a selected visual style using Pillow
4. Shows and optionally saves the final image
"""

from huggingface_hub import InferenceClient
from PIL import Image, ImageEnhance, ImageFilter
from config import HF_API_KEY
from datetime import datetime
import os


# Official SDXL base model
MODEL_ID = "stabilityai/stable-diffusion-xl-base-1.0"


def build_client():
    """Create and return Hugging Face inference client."""
    return InferenceClient(api_key=HF_API_KEY)


def generate_image_from_text(
    client: InferenceClient,
    prompt: str,
    negative_prompt: str = "blurry, low quality, distorted, extra fingers, watermark, text",
    width: int = 1024,
    height: int = 1024,
    steps: int = 30,
    guidance: float = 7.5,
    seed: int | None = None,
) -> Image.Image:
    """
    Generate an image from text using Stable Diffusion XL.
    Returns a PIL Image.
    """
    image = client.text_to_image(
        prompt=prompt,
        negative_prompt=negative_prompt,
        width=width,
        height=height,
        num_inference_steps=steps,
        guidance_scale=guidance,
        model=MODEL_ID,
        seed=seed,
    )
    return image.convert("RGB")


def apply_style_filter(image: Image.Image, style_name: str) -> Image.Image:
    """
    Apply a visual style to the generated image.
    """
    img = image.copy()

    if style_name == "cinematic":
        img = ImageEnhance.Contrast(img).enhance(1.35)
        img = ImageEnhance.Color(img).enhance(1.15)
        img = ImageEnhance.Sharpness(img).enhance(1.2)
        img = img.filter(ImageFilter.GaussianBlur(radius=0.5))

    elif style_name == "dreamy":
        img = ImageEnhance.Brightness(img).enhance(1.08)
        img = ImageEnhance.Color(img).enhance(1.25)
        soft = img.filter(ImageFilter.GaussianBlur(radius=2))
        img = Image.blend(img, soft, alpha=0.35)

    elif style_name == "vintage":
        img = ImageEnhance.Color(img).enhance(0.75)
        img = ImageEnhance.Contrast(img).enhance(1.1)

        # Simple warm vintage tint
        r, g, b = img.split()
        r = r.point(lambda p: min(255, int(p * 1.08)))
        g = g.point(lambda p: min(255, int(p * 1.02)))
        b = b.point(lambda p: min(255, int(p * 0.88)))
        img = Image.merge("RGB", (r, g, b))

    elif style_name == "neon":
        img = ImageEnhance.Color(img).enhance(1.6)
        img = ImageEnhance.Contrast(img).enhance(1.4)
        img = ImageEnhance.Sharpness(img).enhance(1.5)

    elif style_name == "soft-glow":
        glow = img.filter(ImageFilter.GaussianBlur(radius=4))
        img = Image.blend(img, glow, alpha=0.25)
        img = ImageEnhance.Brightness(img).enhance(1.05)

    else:
        # default style
        img = ImageEnhance.Contrast(img).enhance(1.15)
        img = ImageEnhance.Sharpness(img).enhance(1.1)

    return img


def post_process_image(image: Image.Image) -> Image.Image:
    """
    General cleanup / enhancement after style filter.
    """
    img = image.copy()
    img = ImageEnhance.Brightness(img).enhance(1.05)
    img = ImageEnhance.Contrast(img).enhance(1.12)
    img = img.filter(ImageFilter.MedianFilter(size=3))
    return img


def save_image(image: Image.Image, folder: str = "output", file_prefix: str = "styled_image") -> str:
    """
    Save the image and return the file path.
    """
    os.makedirs(folder, exist_ok=True)
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    file_path = os.path.join(folder, f"{file_prefix}_{timestamp}.png")
    image.save(file_path)
    return file_path


def get_style_choice() -> str:
    """
    Ask the user to choose a style.
    """
    print("\nChoose a style:")
    print("1. cinematic")
    print("2. dreamy")
    print("3. vintage")
    print("4. neon")
    print("5. soft-glow")
    print("6. default")

    style_map = {
        "1": "cinematic",
        "2": "dreamy",
        "3": "vintage",
        "4": "neon",
        "5": "soft-glow",
        "6": "default",
    }

    choice = input("Enter your choice (1-6): ").strip()
    return style_map.get(choice, "default")


def main():
    print("=" * 80)
    print("Welcome to AI Image Stylist: Your Signature Look")
    print("Generate an image with Stable Diffusion XL and apply your own visual style.")
    print("Type 'exit' anytime to quit.")
    print("=" * 80)

    client = build_client()

    while True:
        prompt = input("\nEnter image description:\n> ").strip()
        if prompt.lower() == "exit":
            print("Goodbye!")
            break

        if not prompt:
            print("Please enter a valid prompt.")
            continue

        style = get_style_choice()

        try:
            print("\nGenerating image from prompt...")
            image = generate_image_from_text(
                client=client,
                prompt=prompt,
                negative_prompt="blurry, low quality, bad anatomy, extra hands, watermark, text",
                width=1024,
                height=1024,
                steps=30,
                guidance=7.5,
                seed=None,
            )

            print("Applying signature style...")
            styled = apply_style_filter(image, style)

            print("Applying final enhancements...")
            final_image = post_process_image(styled)

            final_image.show()

            save_option = input("\nDo you want to save the image? (yes/no): ").strip().lower()
            if save_option == "yes":
                custom_name = input("Enter file name (without extension): ").strip()
                if not custom_name:
                    custom_name = "styled_image"

                saved_path = save_image(final_image, file_prefix=custom_name)
                print(f"Saved successfully: {saved_path}")

            print("\nDone.")
            print("-" * 80)

        except Exception as e:
            print(f"\nAn error occurred: {e}")
            print("Check your Hugging Face API key, internet connection, and model access.")
            print("-" * 80)


if __name__ == "__main__":
    main()
```
