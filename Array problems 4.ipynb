```python
# 01-eq-mid.py
# Topic: Slices and their sums, Left-right balance, Equilibrium point

def eq_mid(n):
    # Create a balanced list containing n ones, a single zero in the middle, and n ones.
    # For example, if n=2, arr = [1, 1, 0, 1, 1]
    arr = [1]*n + [0] + [1]*n
    size = len(arr)

    # Iterate through every index to find if it acts as an equilibrium point
    for i in range(size):
        left_sum = 0
        right_sum = 0

        # Calculate the sum of elements strictly to the left of index i
        for j in range(i):
            left_sum += arr[j]

        # Calculate the sum of elements strictly to the right of index i
        for k in range(i + 1, size):
            right_sum += arr[k]

        # If the left sum equals the right sum, this index is our equilibrium point
        if left_sum == right_sum:
            return i

    # Return -1 if no equilibrium point is found (though with this structure, there always is one)
    return -1

# Prompt the user to enter a value for n
n = int(input("Enter n (try 5 or 6): "))
# Prompt the user to guess the equilibrium index
guess = input("What is eq_mid(" + str(n) + ")? ")
# Compute the equilibrium index and print both the actual result and the user's guess
print("  eq_mid(" + str(n) + ") =", eq_mid(n), "  your guess:", guess)
```

```python
# 02-eq-val.py
# Topic: Slices and their sums, Equilibrium point

def eq_val(n):
    # Create a balanced list containing n ones, the value n in the middle, and n ones.
    # For example, if n=2, arr = [1, 1, 2, 1, 1]
    arr = [1]*n + [n] + [1]*n
    size = len(arr)

    # Iterate through every index to find if it acts as an equilibrium point
    for i in range(size):
        left_sum = 0
        right_sum = 0

        # Calculate the sum of elements strictly to the left of index i
        for j in range(i):
            left_sum += arr[j]

        # Calculate the sum of elements strictly to the right of index i
        for k in range(i + 1, size):
            right_sum += arr[k]

        # If the left sum equals the right sum, return the value at this equilibrium index
        if left_sum == right_sum:
            return arr[i]

    # Return -1 if no equilibrium point is found
    return -1

# Prompt the user to enter a value for n
n = int(input("Enter n (try 5 or 6): "))
# Prompt the user to guess the equilibrium value
guess = input("What is eq_val(" + str(n) + ")? ")
# Compute the equilibrium value and print both the actual result and the user's guess
print("  eq_val(" + str(n) + ") =", eq_val(n), "  your guess:", guess)
```

```python
# 03-win-sum.py
# Topic: Subarray window, Target sum search

def win_sum(n):
    # Create a list containing n zeros followed by n ones.
    # For example, if n=2, arr = [0, 0, 1, 1]
    arr = [0]*n + [1]*n
    size = len(arr)

    # Iterate through all possible starting indices for a window of size n
    for i in range(size - n + 1):
        curr_sum = 0

        # Sum up the elements inside the current window of size n starting at index i
        for j in range(i, i + n):
            curr_sum += arr[j]

        # If the sum of the current window equals our target n, return the starting index
        if curr_sum == n:
            return i

    # Return -1 if no such window is found
    return -1

# Prompt the user to enter a value for n
n = int(input("Enter n (try 5 or 6): "))
# Prompt the user to guess the starting index of the window that sums to n
guess = input("What is win_sum(" + str(n) + ")? ")
# Compute the window sum starting index and print both the actual result and the user's guess
print("  win_sum(" + str(n) + ") =", win_sum(n), "  your guess:", guess)
```

```python
# Target Tracker Lab

## Outline
In this assignment, you will build a Python program that tracks array sums, checks left-right balance, finds an equilibrium point, grows a subarray window, and searches for a target sum. The activity helps you understand how arrays can be sliced, scanned, and checked using loops and conditions.
```

```python
# 04-target-tracker.py
# Target Tracker Lab: Code Implementation

def find_equilibrium_index(arr):
    """Finds the index where the sum of left elements equals the sum of right elements."""
    size = len(arr)
    for i in range(size):
        left_sum = sum(arr[:i])
        right_sum = sum(arr[i+1:])
        if left_sum == right_sum:
            return i
    return -1

def find_equilibrium_value(arr):
    """Finds the value at the equilibrium point where left sum equals right sum."""
    idx = find_equilibrium_index(arr)
    if idx != -1:
        return arr[idx]
    return -1

def find_window_target_sum(arr, window_size, target_sum):
    """Searches for a subarray window of a given size that matches the target sum."""
    size = len(arr)
    for i in range(size - window_size + 1):
        curr_sum = sum(arr[i:i + window_size])
        if curr_sum == target_sum:
            return i
    return -1

# Example Usage & Demonstrations
if __name__ == '__main__':
    demo_arr1 = [1, 2, 3, 0, 6]
    print(f"Array: {demo_arr1}")
    print(f"Equilibrium Index: {find_equilibrium_index(demo_arr1)} (Left sum and right sum match around index)")
    print(f"Equilibrium Value: {find_equilibrium_value(demo_arr1)}")

    demo_arr2 = [1, 2, 3, 4, 5]
    win_sz = 3
    target = 9
    print(f"\nArray: {demo_arr2}")
    print(f"Searching for window of size {win_sz} that sums to {target}...")
    start_idx = find_window_target_sum(demo_arr2, win_sz, target)
    print(f"Found starting at index: {start_idx} (Subarray is {demo_arr2[start_idx:start_idx+win_sz]})")
```
