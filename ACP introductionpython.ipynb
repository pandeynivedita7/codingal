## Daily Routine Program

```python
print('--- My Daily Routine ---')
print('1. Wake up at 7:00 AM')
print('2. Have breakfast at 7:30 AM')
print('3. Go to school at 8:30 AM')
print('4. Attend classes from 9:00 AM to 3:00 PM')
print('5. Play sports from 3:30 PM to 5:00 PM')
print('6. Study from 5:30 PM to 7:00 PM')
print('7. Have dinner at 7:30 PM')
print('8. Read a book from 8:30 PM to 9:30 PM')
print('9. Go to bed at 10:00 PM')
```

### Weekly Study Time Calculation

```python
# Calculate daily study time in minutes
daily_study_hours = 1.5 # From 5:30 PM to 7:00 PM is 1.5 hours
daily_study_minutes = daily_study_hours * 60

# Calculate weekly study time
study_days_per_week = 5 # Assuming weekdays
weekly_study_minutes = daily_study_minutes * study_days_per_week

print(f'Daily study time: {daily_study_hours} hours ({int(daily_study_minutes)} minutes)')
print(f'Total study time in a week (assuming {study_days_per_week} days of study): {weekly_study_minutes} minutes')
```
