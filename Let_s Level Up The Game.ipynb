```python
#add sprite collision
import pygame # Import the Pygame library for game development
import random # Import the random module for generating random numbers

# Constants for easier adjustments
SCREEN_WIDTH, SCREEN_HEIGHT = 500, 400 # Define screen dimensions (width, height)
MOVEMENT_SPEED = 5 # Define the speed at which the player sprite moves
FONT_SIZE = 72 # Define the font size for text display

# Initialize Pygame
pygame.init() # Initialize all the Pygame modules

# Load and transform the background image
background_image = pygame.transform.scale(pygame.image.load("road.jpg"), # Load 'road.jpg' and scale it to screen dimensions
                                          (SCREEN_WIDTH, SCREEN_HEIGHT))

# Load font once at the beginning
font = pygame.font.SysFont("Times New Roman", FONT_SIZE) # Create a font object for rendering text

class Sprite(pygame.sprite.Sprite): # Define a custom Sprite class inheriting from Pygame's Sprite class

  def __init__(self, color, height, width): # Constructor for the Sprite class
    super().__init__() # Call the constructor of the parent class (pygame.sprite.Sprite)
    self.image = pygame.Surface([width, height]) # Create a blank surface for the sprite's image
    self.image.fill(pygame.Color('dodgerblue'))  # Fill the background of the sprite's image with 'dodgerblue'
    pygame.draw.rect(self.image, color, pygame.Rect(0, 0, width, height)) # Draw a rectangle on the surface with the given color and dimensions
    self.rect = self.image.get_rect() # Get the bounding rectangle of the sprite's image

  def move(self, x_change, y_change): # Method to move the sprite
    self.rect.x = max( # Update the x-coordinate of the sprite's rectangle
        min(self.rect.x + x_change, SCREEN_WIDTH - self.rect.width), 0) # Ensure sprite stays within screen bounds horizontally
    self.rect.y = max( # Update the y-coordinate of the sprite's rectangle
        min(self.rect.y + y_change, SCREEN_HEIGHT - self.rect.height), 0) # Ensure sprite stays within screen bounds vertically


# Setup
screen = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT)) # Set up the display screen
pygame.display.set_caption("Sprite Collision") # Set the window title
all_sprites = pygame.sprite.Group() # Create a sprite group to hold all sprites

# Create sprites
sprite1 = Sprite(pygame.Color('black'), 20, 30) # Create the first sprite (player) with black color
sprite1.rect.x, sprite1.rect.y = random.randint( # Set random initial position for sprite1
    0, SCREEN_WIDTH - sprite1.rect.width), random.randint( # Random x-coordinate within screen bounds
        0, SCREEN_HEIGHT - sprite1.rect.height) # Random y-coordinate within screen bounds
all_sprites.add(sprite1) # Add sprite1 to the sprite group

sprite2 = Sprite(pygame.Color('red'), 20, 30) # Create the second sprite (target) with red color
sprite2.rect.x, sprite2.rect.y = random.randint( # Set random initial position for sprite2
    0, SCREEN_WIDTH - sprite2.rect.width), random.randint( # Random x-coordinate within screen bounds
        0, SCREEN_HEIGHT - sprite2.rect.height) # Random y-coordinate within screen bounds
all_sprites.add(sprite2) # Add sprite2 to the sprite group

# Game loop control variables
running, won = True, False # 'running' controls the game loop, 'won' tracks if the player has won
clock = pygame.time.Clock() # Create a Clock object to control frame rate

# Main game loop
while running: # Loop continues as long as 'running' is True
  for event in pygame.event.get(): # Iterate through all pending Pygame events
    if event.type == pygame.QUIT or (event.type == pygame.KEYDOWN # Check if the user closed the window or pressed 'X'
                                     and event.key == pygame.K_x): # Check for 'x' key press
      running = False # Set 'running' to False to exit the game loop

  if not won: # Only allow player movement if the game is not won yet
    keys = pygame.key.get_pressed() # Get the state of all keyboard keys
    x_change = (keys[pygame.K_RIGHT] - keys[pygame.K_LEFT]) * MOVEMENT_SPEED # Calculate horizontal movement based on arrow keys
    y_change = (keys[pygame.K_DOWN] - keys[pygame.K_UP]) * MOVEMENT_SPEED # Calculate vertical movement based on arrow keys
    sprite1.move(x_change, y_change) # Move sprite1 based on calculated changes

    if sprite1.rect.colliderect(sprite2.rect): # Check for collision between sprite1 and sprite2
      all_sprites.remove(sprite2) # If collision occurs, remove sprite2 from the group
      won = True # Set 'won' to True, indicating the player has won

  # Drawing
  screen.blit(background_image, (0, 0)) # Draw the background image on the screen
  all_sprites.draw(screen) # Draw all sprites from the group onto the screen

  # Display win message
  if won: # If the player has won
    win_text = font.render("You win!", True, pygame.Color('black')) # Render the 'You win!' text in black
    screen.blit(win_text, ((SCREEN_WIDTH - win_text.get_width()) // 2, # Blit the win text to the center of the screen horizontally
                           (SCREEN_HEIGHT - win_text.get_height()) // 2)) # Blit the win text to the center of the screen vertically

  pygame.display.flip() # Update the full display Surface to the screen
  clock.tick(90) # Control the frame rate to a maximum of 90 frames per second

pygame.quit() # Uninitialize Pygame modules when the game loop ends
```

```python
#Add Custom Event
import pygame
import random

# Initialize Pygame
pygame.init()

# Screen settings
WIDTH, HEIGHT = 600, 400
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Custom & Collision Events")

# Colors
WHITE = (255, 255, 255)

# Define two sprite rectangles
sprite1 = pygame.Rect(150, 150, 80, 80)
sprite2 = pygame.Rect(350, 150, 80, 80)

# Initial colors
sprite1_color = (200, 50, 50)
sprite2_color = (50, 50, 200)

# Define custom events
COLOR_CHANGE_EVENT = pygame.USEREVENT + 1
COLLISION_EVENT = pygame.USEREVENT + 2

# Trigger COLOR_CHANGE_EVENT every 2 seconds
pygame.time.set_timer(COLOR_CHANGE_EVENT, 2000)

# Function to generate random colors
def random_color():
    return (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))

# Game loop
running = True
clock = pygame.time.Clock()

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

        # Custom color change event
        if event.type == COLOR_CHANGE_EVENT:
            sprite1_color = random_color()
            sprite2_color = random_color()

        # Collision event handling
        if event.type == COLLISION_EVENT:
            print("Collision Detected! Sprites overlapped!")
            # For effect, turn both black
            sprite1_color = (0, 0, 0)
            sprite2_color = (0, 0, 0)

    # Key controls for sprite1
    keys = pygame.key.get_pressed()
    if keys[pygame.K_LEFT]:  sprite1.x -= 5
    if keys[pygame.K_RIGHT]: sprite1.x += 5
    if keys[pygame.K_UP]:    sprite1.y -= 5
    if keys[pygame.K_DOWN]:  sprite1.y += 5

    # Collision check
    if sprite1.colliderect(sprite2):
        pygame.event.post(pygame.event.Event(COLLISION_EVENT))

    # Draw background
    screen.fill(WHITE)

    # Draw sprites
    pygame.draw.rect(screen, sprite1_color, sprite1)
    pygame.draw.rect(screen, sprite2_color, sprite2)

    # Update display
    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```
