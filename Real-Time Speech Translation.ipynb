```python
import speech_recognition as sr  # Imports the SpeechRecognition library for converting speech to text
import pyttsx3  # Imports the pyttsx3 library for text-to-speech conversion
from googletrans import Translator  # Imports the Translator class from googletrans for language translation

# Initialize text-to-speech engine
def speak(text, language="en"):
    engine = pyttsx3.init()  # Initializes the text-to-speech engine
    engine.setProperty('rate', 150)  # Sets the speech rate to 150 words per minute
    voices = engine.getProperty('voices')  # Gets a list of available voices

    # Set voice for English or other language if supported by pyttsx3
    if language == "en":
        engine.setProperty('voice', voices[0].id)  # Sets the voice to the first available English voice
    else:
        engine.setProperty('voice', voices[1].id)  # Fallback to another voice if available for other languages

    engine.say(text)  # Queues the text to be spoken
    engine.runAndWait()  # Waits for the speech to complete before continuing

# Speech-to-Text: Recognize spoken language (English)
def speech_to_text():
    recognizer = sr.Recognizer()  # Creates a Recognizer instance
    with sr.Microphone() as source:  # Uses the default microphone as the audio source
        print("???? Please speak now in English...")  # Prompts the user to speak
        audio = recognizer.listen(source)  # Listens for speech from the microphone

    try:
        print("???? Recognizing speech...")  # Informs the user that speech is being recognized
        # Uses Google's Web Speech API to recognize speech in US English
        text = recognizer.recognize_google(audio, language="en-US")
        print(f"✅ You said: {text}")  # Prints the recognized text
        return text  # Returns the recognized text
    except sr.UnknownValueError:
        print("❌ Could not understand the audio.")  # Handles cases where speech is unintelligible
    except sr.RequestError as e:
        print(f"❌ API Error: {e}")  # Handles API errors (e.g., network issues)
    return ""  # Returns an empty string if recognition fails

# Translate text using Google Translate API
def translate_text(text, target_language="es"):  # Defines a function to translate text, defaulting to Spanish
    translator = Translator()  # Creates a Translator instance
    translation = translator.translate(text, dest=target_language)  # Translates the text to the target language
    print(f"???? Translated text: {translation.text}")  # Prints the translated text
    return translation.text  # Returns the translated text

# Display language options to the user
def display_language_options():
    print("???? Available translation languages: ")  # Informs the user about available languages
    print("1. Hindi (hi)")  # Lists Hindi option
    print("2. Tamil (ta)")  # Lists Tamil option
    print("3. Telugu (te)")  # Lists Telugu option
    print("4. Bengali (bn)")  # Lists Bengali option
    print("5. Marathi (mr)")  # Lists Marathi option
    print("6. Gujarati (gu)")  # Lists Gujarati option
    print("7. Malayalam (ml)")  # Lists Malayalam option
    print("8. Punjabi (pa)")  # Lists Punjabi option

    # User selects language
    choice = input("Please select the target language number (1-8): ")  # Prompts user for language choice
    language_dict = {  # Dictionary mapping choice numbers to language codes
        "1": "hi",
        "2": "ta",
        "3": "te",
        "4": "bn",
        "5": "mr",
        "6": "gu",
        "7": "ml",
        "8": "pa"
    }

    return language_dict.get(choice, "es")  # Returns the chosen language code, or 'es' (Spanish) as default if invalid input

# Main function to combine all steps
def main():
    # Step 1: Display language options and get user's choice
    target_language = display_language_options()  # Calls function to get target language

    # Step 2: Speech-to-Text (recognizing English speech)
    original_text = speech_to_text()  # Calls function to recognize speech

    if original_text:
        # Step 3: Translate to selected target language
        # Calls function to translate the recognized text to the target language
        translated_text = translate_text(original_text, target_language=target_language)

        # Step 4: Text-to-Speech (Translate output and speak it)
        # Calls function to speak the translated text in English (note: 'en' for speak function is currently hardcoded)
        speak(translated_text, language="en")
        print("✅ Translation spoken out!")  # Confirms that the translation has been spoken

if __name__ == "__main__":
    main()  # Calls the main function when the script is executed
```

```python
import speech_recognition as sr
from deep_translator import GoogleTranslator


# Supported languages
LANGUAGES = {
    "1": ("English", "en"),
    "2": ("Hindi", "hi"),
    "3": ("French", "fr"),
    "4": ("Spanish", "es"),
    "5": ("German", "de"),
    "6": ("Italian", "it"),
    "7": ("Japanese", "ja"),
    "8": ("Korean", "ko"),
    "9": ("Chinese", "zh-CN"),
    "10": ("Arabic", "ar")
}


def display_languages():
    """Display the available languages."""
    print("\nAvailable Languages")
    print("-" * 30)

    for number, (name, code) in LANGUAGES.items():
        print(f"{number}. {name} ({code})")


def select_language(message):
    """Allow the user to select a language."""

    while True:
        display_languages()

        choice = input(f"\n{message}: ").strip()

        if choice in LANGUAGES:
            return LANGUAGES[choice]

        print("\nInvalid selection. Please choose a valid number.")


def recognize_speech(language_code):
    """
    Capture and recognize speech from the microphone.
    Returns recognized text or None if recognition fails.
    """

    recognizer = sr.Recognizer()

    try:
        with sr.Microphone() as source:

            print("\nAdjusting for background noise...")
            recognizer.adjust_for_ambient_noise(source, duration=1)

            print("Speak now...")

            try:
                audio = recognizer.listen(
                    source,
                    timeout=5,
                    phrase_time_limit=10
                )

            except sr.WaitTimeoutError:
                print("Error: No speech detected within the time limit.")
                return None

        try:
            text = recognizer.recognize_google(
                audio,
                language=language_code
            )

            return text

        except sr.UnknownValueError:
            print(
                "Error: Speech could not be understood. "
                "Please speak clearly and try again."
            )

        except sr.RequestError as error:
            print(
                "Error: Speech recognition service is unavailable."
            )
            print("Details:", error)

    except OSError as error:
        print("Error: Microphone could not be accessed.")
        print("Check that a microphone is connected and enabled.")
        print("Details:", error)

    except Exception as error:
        print("Unexpected speech-recognition error:", error)

    return None


def translate_text(text, source_code, target_code):
    """Translate recognized text into the target language."""

    try:

        translated_text = GoogleTranslator(
            source=source_code,
            target=target_code
        ).translate(text)

        return translated_text

    except Exception as error:

        print("\nTranslation failed.")
        print("Details:", error)

        return None


def main():

    print("=" * 50)
    print("      REAL-TIME SPEECH TRANSLATOR")
    print("=" * 50)

    # Select source language
    source_name, source_code = select_language(
        "Select the SOURCE language"
    )

    # Select target language
    target_name, target_code = select_language(
        "Select the TARGET language"
    )

    if source_code == target_code:
        print(
            "\nWarning: Source and target languages "
            "are the same."
        )

    print("\nTranslation Settings")
    print("-" * 30)
    print(f"Source Language : {source_name}")
    print(f"Target Language : {target_name}")

    print("\nStart speaking.")
    print("Press Ctrl+C to stop the application.")

    try:

        while True:

            print("\n" + "=" * 50)

            recognized_text = recognize_speech(source_code)

            if not recognized_text:
                print("Trying again...")
                continue

            print(f"\nRecognized ({source_name}):")
            print(recognized_text)

            translated_text = translate_text(
                recognized_text,
                source_code,
                target_code
            )

            if translated_text:

                print(f"\nTranslation ({target_name}):")
                print(translated_text)

    except KeyboardInterrupt:

        print("\n\nSpeech translator stopped.")


if __name__ == "__main__":
    main()
```
