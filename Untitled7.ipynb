```python
from PIL import Image
import pytesseract

# 1. Tell Python which image you want to read
image_path = "sample_text.png"

try:
    # 2. Open the image using the Pillow library
    opened_image = Image.open(image_path)

    # 3. Ask Tesseract to read the words in the image
    extracted_text = pytesseract.image_to_string(opened_image)

    # 4. Print the secret text on the screen!
    print("--- Here is the text from your image: ---")
    print(extracted_text)
    print("-----------------------------------------")

except FileNotFoundError:
    print(f"Oops! Could not find the image file named '{image_path}'.")
    print("Make sure the image is in the same folder as this Python script!")
```
