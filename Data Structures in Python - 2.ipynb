```python
print('--- Python Set Commands Demo ---')

# 1. Creating Sets
my_set = {1, 2, 3, 'apple', 'banana'}
print(f"Original set: {my_set}")

set_from_list = set([4, 5, 6, 'orange'])
print(f"Set created from a list: {set_from_list}")

# 2. Adding Elements
my_set.add('grape')
print(f"Set after adding 'grape': {my_set}")

my_set.update(['mango', 'kiwi', 2]) # 2 is already in set_from_list, won't be added again
print(f"Set after updating with ['mango', 'kiwi', 2]: {my_set}")

# 3. Removing Elements
my_set.remove('apple')
print(f"Set after removing 'apple': {my_set}")

# Using discard() which doesn't raise an error if item not found
my_set.discard('non_existent_item')
print(f"Set after discarding 'non_existent_item': {my_set}")

# Pop an arbitrary element
popped_item = my_set.pop()
print(f"Set after popping {popped_item}: {my_set}")

# 4. Set Operations (Mathematical)
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# Union
union_set = set1.union(set2)
# Alternatively: union_set = set1 | set2
print(f"\nUnion of set1 and set2: {union_set}")

# Intersection
intersection_set = set1.intersection(set2)
# Alternatively: intersection_set = set1 & set2
print(f"Intersection of set1 and set2: {intersection_set}")

# Difference
difference_set = set1.difference(set2)
# Alternatively: difference_set = set1 - set2
print(f"Difference (set1 - set2): {difference_set}")

# Symmetric Difference
symmetric_difference_set = set1.symmetric_difference(set2)
# Alternatively: symmetric_difference_set = set1 ^ set2
print(f"Symmetric Difference (elements in either but not both): {symmetric_difference_set}")

# 5. Membership and Subsets/Supersets
print(f"\nIs 3 in set1? {3 in set1}")
print(f"Is 9 in set1? {9 in set1}")

subset_test = {1, 2}
print(f"Is {{1, 2}} a subset of set1? {subset_test.issubset(set1)}") # Alternatively: subset_test <= set1

superset_test = {1, 2, 3, 4, 5, 6}
print(f"Is set1 a superset of {{1, 2, 3}}? {set1.issuperset({1, 2, 3})}") # Alternatively: set1 >= {1, 2, 3}

# 6. Disjoint sets
disjoint_set1 = {1, 2}
disjoint_set2 = {3, 4}
disjoint_set3 = {2, 5}

print(f"\nAre {{1, 2}} and {{3, 4}} disjoint? {disjoint_set1.isdisjoint(disjoint_set2)}")
print(f"Are {{1, 2}} and {{2, 5}} disjoint? {disjoint_set1.isdisjoint(disjoint_set3)}")

# 7. Other methods
print(f"\nNumber of elements in my_set: {len(my_set)}")

# Clear a set
my_set.clear()
print(f"my_set after clearing: {my_set}")
```

```python
# ---- Recipe Explorer ----

# STEP 1 - Create tuples for recipe details (fixed — cannot be changed)
pasta = ("Pasta Arrabiata", "Italian", 20, "Medium")
biryani = ("Chicken Biryani", "Indian", 45, "Hard")
print("Recipe 1:", pasta)
print("Name:", pasta[0])
print("Cuisine:", pasta[1])
print("Difficulty:", pasta[-1])

# STEP 2 - Nested tuples and slicing
all_recipes = (pasta, biryani)
print("\nFirst recipe name:", all_recipes[0][0])
print("Second recipe time:", all_recipes[1][2], "mins")
print("Pasta details (sliced):", pasta[1:3])

# STEP 3 - Iterate through a tuple
print("\nPasta Recipe details:")
for detail in pasta:
    print(" -", detail)

# STEP 4 - Create sets for ingredients (no duplicates allowed)
pasta_ingredients = {"tomato", "garlic", "olive oil", "chilli", "pasta", "garlic"}
biryani_ingredients = {"rice", "chicken", "garlic", "onion", "tomato", "spices"}
print("\nPasta ingredients:", pasta_ingredients)
print("Biryani ingredients:", biryani_ingredients)
print("Total pasta ingredients:", len(pasta_ingredients))

# STEP 5 - Modify the set
pasta_ingredients.add("parmesan")
pasta_ingredients.discard("chilli")
print("\nUpdated pasta ingredients:", pasta_ingredients)

# STEP 6 - Set operations
all_ingredients = pasta_ingredients.union(biryani_ingredients)
common = pasta_ingredients.intersection(biryani_ingredients)
only_pasta = pasta_ingredients.difference(biryani_ingredients)
unique_to_each = pasta_ingredients.symmetric_difference(biryani_ingredients)

print("\nAll ingredients (union):", all_ingredients)
print("Common ingredients (intersection):", common)
print("Only in Pasta (difference):", only_pasta)
print("Not shared (sym. difference):", unique_to_each)
```

```python
# --- My School Subject Planner ---

# 1. Fixed Student Details (Tuple)
student_info = ("Alice Smith", 10, "Grade 5", "alice.smith@example.com")
print(f"\nStudent Name: {student_info[0]}")
print(f"Student ID: {student_info[1]}")
print(f"Grade Level: {student_info[2]}")

# 2. Subject Sets for Different Days
monday_subjects = {"Math", "Science", "English"}
tuesday_subjects = {"History", "Math", "Art"}
wednesday_subjects = {"Science", "Music", "English"}

print(f"\nMonday's Subjects: {monday_subjects}")
print(f"Tuesday's Subjects: {tuesday_subjects}")

# 3. Modify Sets
# Add a new subject to Monday
monday_subjects.add("PE")
print(f"Monday's Subjects after adding PE: {monday_subjects}")

# Remove a subject from Tuesday
tuesday_subjects.discard("Art") # Use discard to avoid error if not present
print(f"Tuesday's Subjects after removing Art: {tuesday_subjects}")

# 4. Compare Subjects Using Set Operations

# Subjects common to Monday and Tuesday
common_subjects_mon_tue = monday_subjects.intersection(tuesday_subjects)
print(f"\nCommon subjects on Monday and Tuesday: {common_subjects_mon_tue}")

# All unique subjects taught on Monday or Tuesday
all_subjects_mon_tue = monday_subjects.union(tuesday_subjects)
print(f"All unique subjects on Monday and Tuesday: {all_subjects_mon_tue}")

# Subjects only on Monday, not Tuesday
only_monday_subjects = monday_subjects.difference(tuesday_subjects)
print(f"Subjects only on Monday (not Tuesday): {only_monday_subjects}")

# Subjects unique to either Monday or Tuesday (but not both)
unique_mon_tue = monday_subjects.symmetric_difference(tuesday_subjects)
print(f"Subjects unique to either Monday or Tuesday: {unique_mon_tue}")

```
