```python
from IPython.display import display, Markdown
import ipywidgets as widgets

# Create input widgets
name = widgets.Text(description="Name:")
age = widgets.IntText(description="Age:")
grade = widgets.Text(description="Grade:")
student_id = widgets.Text(description="Student ID:")

# Output area
output = widgets.Output()

# Function to display profile
def show_profile(btn):
    with output:
        output.clear_output()
        if not (name.value and age.value and grade.value and student_id.value):
            print("Please fill out all fields.")
            return
        print("-" * 30)
        print("       STUDENT PROFILE")
        print("-" * 30)
        print(f"Name       : {name.value}")
        print(f"Age        : {age.value}")
        print(f"Grade      : {grade.value}")
        print(f"Student ID : {student_id.value}")
        print("-" * 30)

# Button to trigger profile display
button = widgets.Button(description="Show Profile", button_style='success')
button.on_click(show_profile)

# Display all widgets
display(name, age, grade, student_id, button, output)
```
