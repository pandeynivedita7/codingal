```python
# Import Libraries
# pandas for data manipulation and analysis
import pandas as pd
# numpy for numerical operations
import numpy as np
# matplotlib.pyplot for plotting
import matplotlib.pyplot as plt
# seaborn for enhanced data visualizations
import seaborn as sns

# Read the uploaded CSV file into a pandas DataFrame
data = pd.read_csv('Titanic Dataset.csv')

# Display the first 5 rows of the DataFrame to get a quick overview of the data
data.head(5)

# Minimum and Maximum Values of Age
# Calculate the minimum age in the 'Age' column
minimum_age = data['Age'].min()
print('Minimum Age :', minimum_age)

# Calculate the maximum age in the 'Age' column
maximum_age = data['Age'].max()
print('Maximum Age :', maximum_age)

# Creating binned age and giving it a label
# Define the age bins for categorization
bins = [0, 15, 30, 45, 60, 75]

# Create a new column 'binned_age' by binning the 'Age' column using the defined bins
data['binned_age'] = pd.cut(data['Age'], bins)

# Print the head of 'binned_age' and 'Age' columns to check the initial binning
print(data[['binned_age', 'Age']].head())

# Define custom labels for the age bins
age_labels = ['Young', 'Young - Adult', 'Middle Aged', 'Middle-Older Age', 'Senior']

# Re-bin the 'Age' column with the custom labels
data['binned_age'] = pd.cut(data['Age'], bins, labels = age_labels)

# Barplot for binned age
# Count the occurrences of each age bin and plot them as a bar graph
data['binned_age'].value_counts().plot(kind='bar')

# Set the title of the bar graph
plt.title('Dance Class Age Distribution')
# Set the label for the x-axis
plt.xlabel('Ages')
# Set the label for the y-axis
plt.ylabel('Count')

# Conclusion regarding features Gender and Embarked
# Features Gender and Embarked have only 2 and 3 categories respectively
# Other categorical features have too many categories to even collapse

# Check distribution and skewness of all the features
# Define a list of numerical labels (features) to analyze
labels = ['PassengerId','Survived','Pclass','Age','SibSp','Parch','Fare']
# Loop through each label to plot its distribution and calculate skewness
for label in labels:
  print('Distribution of', label)
  # Plot the distribution of the current feature using seaborn's distplot
  sns.distplot(data[label])
  # Display the plot
  plt.show()
  # Calculate and print the skewness of the current feature
  print('Skewness -', data[label].skew())

# Conclusion regarding skewed features
# Features - SibSp, Parch and Fare are skewed

# Log Transform Skewed Features
# Apply a natural logarithm transformation to 'SibSp' to reduce skewness
data['log_SibSp'] = np.log(data['SibSp'])
# Apply a natural logarithm transformation to 'Parch' to reduce skewness
data['log_Parch'] = np.log(data['Parch'])
# Apply a natural logarithm transformation to 'Fare' to reduce skewness
data['log_Fare'] = np.log(data['Fare'])
```

```python
# Import Libraries
# pandas for data manipulation and analysis
import pandas as pd
# numpy for numerical operations
import numpy as np
# matplotlib.pyplot for plotting
import matplotlib.pyplot as plt
# seaborn for enhanced data visualizations
import seaborn as sns

# Import Dataset
# Used to upload files to Google Colab environment
from google.colab import files
# Prompt user to upload a file
uploaded = files.upload()

# Read the uploaded CSV file into a pandas DataFrame
data = pd.read_csv('Titanic Dataset.csv')

# Display the first 5 rows of the DataFrame to get a quick overview of the data
data.head(5)

# Check association between Embarked and Age
# Create a box plot to visualize the distribution of 'Age' across different 'Embarked' categories
sns.boxplot(data = data, x = 'Embarked', y = 'Age')
# Display the plot
plt.show()

# Conclusion: There is too much overlapping in the boxplots, which means there is not much association between Embarked and Age

# Check association between Survived and Fare, SibSp, Parch
# Create a scatter plot to visualize the relationship between 'Fare' and 'Survived'
plt.scatter(x = data['Fare'], y = data['Survived'])
# Set the label for the y-axis
plt.ylabel('Survived')
# Set the label for the x-axis
plt.xlabel('Fare')
# Display the plot
plt.show()

# Create a scatter plot to visualize the relationship between 'Parch' and 'Survived'
plt.scatter(x = data['Parch'], y = data['Survived'])
# Set the label for the y-axis
plt.ylabel('Survived')
# Set the label for the x-axis
plt.xlabel('Parch')
# Display the plot
plt.show()

# Create a scatter plot to visualize the relationship between 'SibSp' and 'Survived'
plt.scatter(x = data['SibSp'], y = data['Survived'])
# Set the label for the y-axis
plt.ylabel('Survived')
# Set the label for the x-axis
plt.xlabel('SibSp')
# Display the plot
plt.show()

# Conclusion: There is an association of features Parch, SibSp with Survived

# Check association between Gender and Embarked
# Create a contingency table (crosstab) to show the frequency distribution of 'Gender' and 'Embarked'
association_categorical = pd.crosstab(data['Gender'], data['Embarked'])
# Print the contingency table
print(association_categorical)
```
