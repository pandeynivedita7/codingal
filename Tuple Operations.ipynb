```python
print("----- TUPLE OPERATIONS IN PYTHON -----\n")

# Creating a tuple
tup = (10, 20, 30, 40, 50)
print("Original Tuple:", tup)

# Indexing
print("Element at index 2:", tup[2])

# Slicing
print("Tuple from index 1 to 3:", tup[1:4])

# Length
print("Length of tuple:", len(tup))

# Membership test
print("Is 20 in tuple?", 20 in tup)
print("Is 100 in tuple?", 100 in tup)

# Concatenation
tup2 = (60, 70)
print("Concatenated Tuple:", tup + tup2)

# Repetition
print("Tuple Repeated Twice:", tup * 2)

# Iterating through tuple
print("Iterating through tuple:")
for item in tup:
    print(item, end=" ")

# Tuple is immutable, so below will raise error if uncommented
# tup[0] = 99  # TypeError: 'tuple' object does not support item assignment

print("\nTuple is immutable. You cannot change its elements.\n")
```

```python
# Program to calculate product of all numbers in a tuple

# Given tuple
numbers = (2, 3, 4, 5)

product = 1  # start with 1 (multiplicative identity)

for num in numbers:
    product *= num   # multiply each element product=product*num

print("Tuple:", numbers)
print("Product of all numbers:", product)
```

```python
# Function to check whether a tuple is a palindrome or not
def palind(r):# def function part code that can reused def funname (par):
    e = len(r) - 1#string which you read from start or end guve same vale 121
    s = 0# 0 empty
    while s < e:# true
        if r[s] != r[e]:
            return False
        s += 1
        e -= 1
    return True

# Tuple to check
r = (1, 2, 3, 3, 2, 1)

if palind(r):# funname
    print("The Tuple is Flip-Flop")
else:
    print("The Tuple is not Flip-Flop")


```
