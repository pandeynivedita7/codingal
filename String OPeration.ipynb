```python
# ==========================================
# PYTHON STRING OPERATIONS - ONE PROGRAM
# ==========================================

text = "   Python Programming   "

print("Original String:", text)

# 1. Length
print("\n1. Length:")
print(len(text))

# 2. Indexing
print("\n2. Indexing:")
print(text[3])
print(text[-1])

# 3. Slicing
print("\n3. Slicing:")
print(text[3:9])
print(text[:6])
print(text[10:])
print(text[::-1])# reverse string

# 4. Concatenation
print("\n4. Concatenation:")
first = "Hello"
second = "Python"
print(first + " " + second)

# 5. Repetition
print("\n5. Repetition:")
print("Hi " * 3)

# 6. Membership
print("\n6. Membership:")
print("Python" in text)
print("Java" not in text)

# 7. Uppercase
print("\n7. Uppercase:")
print(text.upper())

# 8. Lowercase
print("\n8. Lowercase:")
print(text.lower())

# 9. Capitalize
print("\n9. Capitalize:")
print(text.capitalize())

# 10. Title
print("\n10. Title:")
print(text.title())

# 11. Swapcase
print("\n11. Swapcase:")
print(text.swapcase())

# 12. Casefold
print("\n12. Casefold:")
print(text.casefold())

# 13. Strip
print("\n13. Strip:")
print(text.strip())

# 14. Lstrip
print("\n14. Lstrip:")
print(text.lstrip())

# 15. Rstrip
print("\n15. Rstrip:")
print(text.rstrip())

# 16. Find
print("\n16. Find:")
print(text.find("Python"))

# 17. Rfind
print("\n17. Rfind:")
print(text.rfind("Python"))

# 18. Count
print("\n18. Count:")
print(text.count("m"))

# 19. Replace
print("\n19. Replace:")
print(text.replace("Python", "Java"))

# 20. Startswith
print("\n20. Startswith:")
print(text.strip().startswith("Python"))

# 21. Endswith
print("\n21. Endswith:")
print(text.strip().endswith("Programming"))

# 22. Split
print("\n22. Split:")
words = "Python Java C++".split()
print(words)

# 23. Join
print("\n23. Join:")
print("-".join(words))

# 24. Isalpha
print("\n24. Isalpha:")
print("Python".isalpha())

# 25. Isdigit
print("\n25. Isdigit:")
print("12345".isdigit())

# 26. Isalnum
print("\n26. Isalnum:")
print("Python123".isalnum())

# 27. Isspace
print("\n27. Isspace:")
print("   ".isspace())

# 28. Islower
print("\n28. Islower:")
print("python".islower())

# 29. Isupper
print("\n29. Isupper:")
print("PYTHON".isupper())

# 30. Istitle
print("\n30. Istitle:")
print("Python Programming".istitle())

# 31. Center
print("\n31. Center:")
print("Python".center(20, "-"))

# 32. Ljust
print("\n32. Ljust:")
print("Python".ljust(20, "-"))

# 33. Rjust
print("\n33. Rjust:")
print("Python".rjust(20, "-"))

# 34. Zfill
print("\n34. Zfill:")
print("25".zfill(5))

# 35. Partition
print("\n35. Partition:")
print("Python-is-easy".partition("-"))

# 36. Remove Prefix
print("\n36. Remove Prefix:")
print("PythonProgramming".removeprefix("Python"))

# 37. Remove Suffix
print("\n37. Remove Suffix:")
print("program.py".removesuffix(".py"))

# 38. Ord
print("\n38. Ord:")
print(ord("A"))

# 39. Chr
print("\n39. Chr:")
print(chr(65))

# 40. String Formatting
print("\n40. String Formatting:")
name = "Nivedita"
age = 38
print(f"My name is {name} and I am {age} years old.")

# 41. String Comparison
print("\n41. String Comparison:")
a = "Python"
b = "Python"
print(a == b)
print(a != b)

# 42. Loop through String
print("\n42. Loop through String:")
for character in "Python":
    print(character)

# 43. Reverse String
print("\n43. Reverse:")
print("Python"[::-1])

print("\n==========================================")
print("        STRING OPERATIONS COMPLETED")
print("==========================================")
```
