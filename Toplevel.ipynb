```python
import tkinter as tk

def open_new_window():
    # 1. Create the top-level window
    new_window = tk.Toplevel(root)
    new_window.title("Secondary Window")
    new_window.geometry("300x200")

    # 2. Add widgets to the top-level window
    label = tk.Label(new_window, text="This is a Toplevel Window!")
    label.pack(pady=20)

    close_btn = tk.Button(new_window, text="Close Window", command=new_window.destroy)
    close_btn.pack()

# Create the main window
root = tk.Tk()
root.title("Main Window")
root.geometry("400x300")

# Button to trigger the top-level window
open_btn = tk.Button(root, text="Open Toplevel Window", command=open_new_window)
open_btn.pack(pady=50)

root.mainloop()
```

```python
# my-photo-album.py
# Activity: My Photo Album
# Lesson: Tkinter Widgets | Grade 4-6

# PART 1 - Import libraries and set up the main window
from tkinter import *
from tkinter import messagebox
from PIL import Image, ImageTk
window = Tk()
window.title('My Photo Album')
window.geometry('400x420')

# PART 2 - Add a title label and display an image
title = Label(window, text='My Photo Album', fg='white', bg='purple', width=40)
title.pack(pady=10)
img_file = Image.open('img.jfif')
img_file = img_file.resize((300, 180))
photo = ImageTk.PhotoImage(img_file)
pic = Label(window, image=photo)
pic.pack(pady=5)

# PART 3 - Create a function to show a message box
def show_message():
    messagebox.showinfo('Great!', 'You clicked the photo!')
msg_btn = Button(window, text='Click to React', bg='blue', fg='white', command=show_message)
msg_btn.pack(pady=5)

# PART 4 - Create a function to open a new window with photo details
def show_details():
    top = Toplevel()
    top.title('Photo Details')
    top.geometry('200x120')
    info = Label(top, text='Taken on: 1 June 2025')
    info.pack(pady=10)
    place = Label(top, text='Location: My Garden')
    place.pack()
    top.mainloop()
details_btn = Button(window, text='See Details', bg='green', fg='white', command=show_details)
details_btn.pack(pady=5)

# PART 5 - Run the main window loop
window.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox

def open_warning_window():
    # 1. Create the top-level window
    warn_window = tk.Toplevel(root)
    warn_window.title("Alert Screen")
    warn_window.geometry("300x150")

    # 2. Force focus so user must interact with this window
    warn_window.grab_set()

    # 3. Trigger the built-in warning message box
    messagebox.showwarning(
        title="System Warning",
        message="An unauthorized change was detected!",
        parent=warn_window  # Centers the alert over the top-level window
    )

    # 4. Optional label inside the top-level window itself
    label = tk.Label(warn_window, text="Please resolve the issue above.", fg="red")
    label.pack(pady=20)

    close_btn = tk.Button(warn_window, text="Dismiss", command=warn_window.destroy)
    close_btn.pack()

# Main Application Window
root = tk.Tk()
root.title("Main Application")
root.geometry("400x300")

launch_btn = tk.Button(root, text="Trigger Warning Window", command=open_warning_window)
launch_btn.pack(pady=100)

root.mainloop()
```
