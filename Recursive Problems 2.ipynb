```python
# 01-stair-climb.py
# Topic: The Stair Climb Problem, Solving Stair Climb with Code

# Define a function named 'ways' that takes the number of stairs as input.
def ways(stairs):
    # Base case: If the number of stairs is negative, it's an invalid path, so return 0 ways.
    if stairs < 0:
        return 0
    # Base case: If there are 0 stairs, there is one way to climb (do nothing).
    if stairs == 1:
        return 1
    # Recursive step: The number of ways to climb 'stairs' is the sum of:
    # 1. Ways to climb 'stairs - 1' (taking a 1-step first)
    # 2. Ways to climb 'stairs - 2' (taking a 2-step first)
    return ways(stairs - 1) + ways(stairs - 2)

# Prompt the user to press Enter to proceed.
input("ways counts every distinct path up n stairs — 1 step or 2 steps at a time.  Press Enter ")
# Calculate and print the number of ways to climb 3 stairs.
print("  ways(3) =", ways(3))
# Calculate and print the number of ways to climb 4 stairs.
print("  ways(4) =", ways(4))

# Get the number of steps from the user and convert it to an integer.
n = int(input("Enter number of steps (try 5 or 6): "))
# Ask the user to guess the number of ways for 'n' stairs.
guess = input("What is ways(" + str(n) + ")? ")
# Prompt the user to press Enter to reveal the formula.
input("ways(stairs) = ways(stairs-1) + ways(stairs-2)  both branches always combine.  Press Enter ")
# Calculate and print the actual number of ways for 'n' stairs and compare it with the user's guess.
print("  ways(" + str(n) + ") =", ways(n), "  your guess:", guess)
```

```python
# 02-balanced-parens.py
# Topic: The Balanced Parentheses Problem, Solving Balanced Parentheses with Code

# Define a recursive function `count_paren` to count valid balanced parentheses sequences.
# n: total number of parenthesis pairs
# l: count of left parentheses used so far
# r: count of right parentheses used so far
def count_paren(n, l=0, r=0):
    # Base case: If both left and right parentheses counts equal n, a valid sequence is formed.
    # Return 1 to count this valid sequence.
    if l == n and r == n:
        return 1
    # Initialize total ways for the current state.
    total = 0
    # Rule for adding a right parenthesis: Can only add a right parenthesis if it doesn't precede a left parenthesis.
    # (i.e., the number of left parentheses used 'l' must be greater than the number of right parentheses used 'r').
    if l > r:
        # Recursively call with an incremented right parenthesis count.
        total += count_paren(n, l, r + 1)
    # Rule for adding a left parenthesis: Can always add a left parenthesis as long as we haven't used all 'n' of them.
    if l < n:
        # Recursively call with an incremented left parenthesis count.
        total += count_paren(n, l + 1, r)
    # Return the total number of valid sequences from this state.
    return total

# Prompt the user to press Enter to proceed with the explanation.
input("count_paren counts every valid {} sequence — returns 1 at each valid end.  Press Enter ")
# Calculate and print the number of valid sequences for 1 pair of parentheses.
print("  count_paren(1) =", count_paren(1))
# Calculate and print the number of valid sequences for 2 pairs of parentheses.
print("  count_paren(2) =", count_paren(2))

# Get the number of parenthesis pairs from the user and convert to an integer.
n = int(input("Enter number of pairs (try 3 or 4): "))
# Ask the user to guess the number of valid sequences for 'n' pairs.
guess = input("What is count_paren(" + str(n) + ")? ")
# Prompt the user to press Enter to reveal the underlying logic.
input("l>r closes  l<n opens — adds 1 at every valid ending.  Press Enter ")
# Calculate and print the actual number of valid sequences for 'n' pairs and compare with the user's guess.
print("  count_paren(" + str(n) + ") =", count_paren(n), "  your guess:", guess)
```

```python
# ============================================
# Stair Climb and Brace Builder Using Recursion
# ============================================


# ------------------------------------------------
# Part 1: Stair Climb Problem
# ------------------------------------------------
# Problem:
# A person can climb 1 or 2 steps at a time.
# Find the number of possible ways to reach the top.
#
# Example:
# n = 4
# Ways:
# 1+1+1+1
# 1+1+2
# 1+2+1
# 2+1+1
# 2+2
# Answer = 5
# ------------------------------------------------


def stair_climb(n):
    """
    Recursive function to calculate
    number of ways to climb stairs.
    """

    # Base cases
    if n == 0:
        return 1

    if n == 1:
        return 1

    # Recursive choices
    return stair_climb(n - 1) + stair_climb(n - 2)


# ------------------------------------------------
# Part 2: Trace Recursive Call Tree
# ------------------------------------------------

def trace_stair_calls(n, level=0):
    """
    Displays how recursive calls branch.
    """

    print("  " * level + f"stair_climb({n})")

    if n == 0 or n == 1:
        return

    trace_stair_calls(n - 1, level + 1)
    trace_stair_calls(n - 2, level + 1)



# ------------------------------------------------
# Part 3: Generate Stair Paths
# ------------------------------------------------
# Instead of only counting ways,
# generate the actual climbing sequences.
# ------------------------------------------------


def generate_stair_paths(n, path=[]):

    # Base case
    if n == 0:
        print(path)
        return

    # Choice 1: Take one step
    if n >= 1:
        generate_stair_paths(
            n - 1,
            path + [1]
        )

    # Choice 2: Take two steps
    if n >= 2:
        generate_stair_paths(
            n - 2,
            path + [2]
        )



# ------------------------------------------------
# Part 4: Balanced Brace Builder
# ------------------------------------------------
# Generate all valid combinations of braces.
#
# Example:
# n = 3
#
# Output:
# ((()))
# (()())
# (())()
# ()(())
# ()()()
# ------------------------------------------------


def generate_braces(n):

    result = []

    def backtrack(current, open_count, close_count):

        # Base case:
        # All braces are used
        if len(current) == n * 2:
            result.append(current)
            return


        # Add opening brace
        if open_count < n:
            backtrack(
                current + "{",
                open_count + 1,
                close_count
            )


        # Add closing brace only when valid
        if close_count < open_count:
            backtrack(
                current + "}",
                open_count,
                close_count + 1
            )


    backtrack("", 0, 0)

    return result



# ------------------------------------------------
# Main Program
# ------------------------------------------------


print("================================")
print("STAIR CLIMB RECURSION")
print("================================")

stairs = 5

print(
    "Number of ways to climb",
    stairs,
    "steps:",
    stair_climb(stairs)
)


print("\nRecursive Call Tree:")
trace_stair_calls(4)



print("\n================================")
print("GENERATED STAIR PATHS")
print("================================")

generate_stair_paths(4)



print("\n================================")
print("BALANCED BRACE BUILDER")
print("================================")

brace_count = 3

brace_patterns = generate_braces(brace_count)


print(
    "Valid brace combinations:",
    len(brace_patterns)
)


for pattern in brace_patterns:
    print(pattern)
```
