```python
import random

secret = random.randint(1, 50)
lives = 5

while lives > 0:
    print("Lives:", "❤️" * lives)
    try:
        guess = int(input("Guess (1-50): "))
    except ValueError:
        print("Enter a number!\n")
        continue

    if guess == secret:
        print("You won! 🎉")
        break

    lives -= 1
    if lives > 0:
        hint = "Too low" if guess < secret else "Too high"
        dist = "close" if abs(secret - guess) <= 5 else "far"
        print(f"{hint} and {dist}!\n")

if lives == 0:
    print("Game Over! Number was:", secret)
```

```python
import random

secret, lives = random.randint(1, 50), 5

while lives > 0:
  print(f"Lives: {'❤️' * lives}")
  guess = int(input("Guess (1-50): "))

  if guess == secret:
    print("You won! 🎉")
    break

  lives -= 1
  if lives > 0:
    print("Too low!\n" if guess < secret else "Too high!\n")
else:
  print("Game Over! Number was:", secret)
```
