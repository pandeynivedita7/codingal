```python
# 01-max-mid.py
# Topic: Subarrays, Running Sum with Reset, Max-So-Far Tracker, Kadane's Algorithm

def max_mid(n):
    # Step 1: Create the target array containing [-1, n, -1]
    arr = [-1, n, -1]# n=5 [-1,5,-1]
    #find the contiguous subarray whose elements have the largest possible sum.
    # Step 2: Initialize trackers.
    # 'current' tracks the running sum of the current subarray.
    # 'best' tracks the maximum subarray sum found so far.
    current = best = arr[0] # Current=-1 best -1

    # Step 3: Iterate through the array starting from the second element (index 1)
    for x in arr[1:]:
        # Step 4: Kadane's Reset Decision
        # If the running sum 'current' is negative, it's better to discard it
        # and start a new subarray sum from the current element 'x'.
        if current < 0: #-1 start from 5
            current = x
        # Otherwise, add the current element to our running sum.
        else:
            current = current + x# -1+5=4 5>4  5>-1 best 5
            # x=-1 current 5 Current=5+(-1)=4 Current =4 best=5 return 5 max_mis(5)=5

        # Step 5: Update the 'best' tracker if the new running sum is larger
        if current > best:
            best = current

    # Step 6: Return the maximum subarray sum found
    return best

# Interactive execution flow
input("max_mid(n) finds the max subarray sum in [-1, n, -1].  Press Enter ")
print("  max_mid(3) =", max_mid(3))
print("  max_mid(4) =", max_mid(4))
n = int(input("Enter n (try 5 or 6): "))
guess = input("What is max_mid(" + str(n) + ")? ")
input("negative at start resets — best locks in the peak value n.  Press Enter ")
print("  max_mid(" + str(n) + ") =", max_mid(n), "  your guess:", guess)
```

```python
# 02-max-front.py
# Topic: The Drag of Negatives, Max-So-Far Tracker

def max_front(n):
    # Step 1: Create the target array. For n=5, this is [1, 1, 1, 1, 1, -1] n=5 best sunarray [1,1,1,1,1]=5 max_front(5)=5
    arr = [1]*n + [-1]

    # Step 2: Initialize trackers with the first element of the array (1)
    current = best = arr[0] # current=1 best=1

    # Step 3: Iterate through the array starting from the second element
    for x in arr[1:]:
        # Step 4: Kadane's Reset Decision
        # Since the first n elements are 1, 'current' stays positive and we continually add to it.
        if current < 0: # current=2
            current = x
        else:
            current = current + x

        # Step 5: Update the 'best' tracker if the new running sum is larger.
        # 'best' will reach a maximum of n (e.g., 5) before hitting the final -1.
        # When it hits -1, 'current' drops to n-1, but 'best' retains the peak value of n.
        if current > best: # 2>best best=2 4>5 best 5 return 5
            best = current

    # Step 6: Return the maximum subarray sum found (which locks in the peak value 'n')
    return best

# Interactive execution flow
input("max_front(n) finds the max subarray sum in [1, 1, ..., 1, -1].  Press Enter ")
print("  max_front(3) =", max_front(3))
print("  max_front(4) =", max_front(4))
n = int(input("Enter n (try 5 or 6): "))
guess = input("What is max_front(" + str(n) + ")? ")
input("best locks in n ones — the last -1 drags sum down but best holds.  Press Enter ")
print("  max_front(" + str(n) + ") =", max_front(n), "  your guess:", guess)
```

```python
# 03-max-after.py
# Topic: The Drag of Negatives, Running Sum with Reset, Kadane's Algorithm
#A large negative value appears first, so the algorithm resets when it reaches the first positive 1. Then the positive values keep building the sum.
def max_after(n):
    # Step 1: Create the target array. For n=5, this is [-5, 1, 1, 1, 1, 1] [-n, 1, 1, 1, ..., 1] sum=5 max_after(5)=5
    arr = [-n] + [1]*n

    # Step 2: Initialize trackers.
    # 'current' and 'best' both start with the large negative value (-n)
    current = best = arr[0]#-5

    # Step 3: Iterate through the array starting from the second element (the first of the 1s)
    for x in arr[1:]:
        # Step 4: Kadane's Reset Decision
        # Since current initially starts as -n (which is negative), it resets to the first 1 (x=1).
        # In subsequent iterations, current is positive so it keeps adding up.
        if current < 0:#-5<0 current=1 -5+1=-4
            current = x
        else:
            current = current + x

        # Step 5: Update 'best' tracker
        # The reset instantly replaces the large negative value with 1, which becomes the new best.
        # 'best' then scales up to n as the remaining 1s are added.
        if current > best:#1>-4
            best = current# best -5 current 1 best 1

    # Step 6: Return the maximum subarray sum found (which will be n)
    return best

# Interactive execution flow
input("max_after(n) finds the max subarray sum in [-n, 1, 1, ..., 1].  Press Enter ")
print("  max_after(3) =", max_after(3))
print("  max_after(4) =", max_after(4))
n = int(input("Enter n (try 5 or 6): "))
guess = input("What is max_after(" + str(n) + ")? ")
input("big negative resets — then n ones add up to n.  Press Enter ")
print("  max_after(" + str(n) + ") =", max_after(n), "  your guess:", guess)
```
