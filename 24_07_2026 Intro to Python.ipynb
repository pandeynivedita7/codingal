```python
# Initialize two numbers
number1 = 10
number2 = 5

# Addition: Add number1 and number2
sum_result = number1 + number2
print(f"Addition: {number1} + {number2} = {sum_result}")

# Subtraction: Subtract number2 from number1
difference_result = number1 - number2
print(f"Subtraction: {number1} - {number2} = {difference_result}")

# Multiplication: Multiply number1 by number2
product_result = number1 * number2
print(f"Multiplication: {number1} * {number2} = {product_result}")

# Division: Divide number1 by number2
# Note: Division in Python 3 returns a float
division_result = number1 / number2
print(f"Division: {number1} / {number2} = {division_result}")

# Modulus: Get the remainder when number1 is divided by number2
modulus_result = number1 % number2
print(f"Modulus: {number1} % {number2} = {modulus_result}")

# Exponentiation: Raise number1 to the power of number2
power_result = number1 ** number2
print(f"Exponentiation: {number1} ** {number2} = {power_result}")
```
