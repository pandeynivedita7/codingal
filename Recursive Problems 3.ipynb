```python
# 01-tower-of-hanoi.py
# Topic: Tower of Hanoi, Solving Tower of Hanoi with Code

# Define the recursive function to calculate the minimum moves for Tower of Hanoi
# n: the number of disks
def hanoi(n):
    # Base case: If there are no disks, no moves are needed.
    if n == 0:
        return 0
    # Recursive step: The formula for Tower of Hanoi is 2 * hanoi(n-1) + 1.
    # This means:
    # 1. Move n-1 disks from source to auxiliary peg (hanoi(n-1) moves).
    # 2. Move the largest disk from source to destination peg (1 move).
    # 3. Move n-1 disks from auxiliary to destination peg (hanoi(n-1) moves).
    return 2 * hanoi(n - 1) + 1

# Prompt the user to understand the purpose of the hanoi function.
input("hanoi(n) counts the minimum moves to shift n disks from peg A to peg C.  Press Enter ")

# Demonstrate the function with 1 and 2 disks
print("  hanoi(1) =", hanoi(1))
print("  hanoi(2) =", hanoi(2))

# Ask the user for the number of disks to try
n = int(input("Enter number of disks (try 3 or 4): "))

# Ask the user for their guess for the number of moves
guess = input("What is hanoi(" + str(n) + ")? ")

# Explain the recursive formula for Tower of Hanoi moves
input("hanoi(n) = 2 * hanoi(n-1) + 1  move the stack twice plus the big disk once.  Press Enter ")

# Display the actual result from the hanoi function and compare it with the user's guess
print("  hanoi(" + str(n) + ") =", hanoi(n), "  your guess:", guess)
```

```python
# 02-phone-keypad.py
# Topic: Phone Keypad, Recursion Trees and How Problems Grow

# Define a dictionary mapping digits to their corresponding letters on a phone keypad.
# Note: '1' and '0' are typically not included as they don't have letters.
keys = {'2':'abc','3':'def','4':'ghi','5':'jkl','6':'mno','7':'pqrs','8':'tuv','9':'wxyz'}

# Define a recursive function to count the number of possible letter combinations
# for a given string of digits.
def count_combos(digits):
    # Base case: If there are no digits left, there is only one way to form an empty combination.
    if len(digits) == 0:
        return 1
    # Recursive step: Multiply the number of choices for the current digit
    # by the number of combinations for the remaining digits.
    # len(keys[digits[0]]) gets the number of letters for the first digit.
    # count_combos(digits[1:]) recursively calculates combinations for the rest of the digits.
    return len(keys[digits[0]]) * count_combos(digits[1:])

# Prompt the user to understand the logic of the count_combos function.
input("count_combos multiplies letter choices at each digit — 3 letters per step.  Press Enter ")

# Demonstrate the function with single and double digit inputs
print("  count_combos('2') =", count_combos('2'))
print("  count_combos('23') =", count_combos('23'))

# Ask the user to input a string of digits
d = input("Enter digits 2-9 (try '234' or '2345'): ")

# Ask the user for their guess for the number of combinations
guess = input("What is count_combos('" + d + "')? ")

# Explain the recursive multiplication principle behind counting combinations
input("count_combos = choices at digit 0  x  count_combos of the rest.  Press Enter ")

# Display the actual result from the count_combos function and compare it with the user's guess
print("  count_combos('" + d + "') =", count_combos(d), "  your guess:", guess)
```

```python
# 03-recursion-tree.py
# Topic: Recursion Trees and How Problems Grow

# Define a recursive function to count the number of unique paths from the top-left
# to the bottom-right of an m x n grid, moving only right or down.
def paths(m, n):
    # Base case: If we are in the first row (m=1) or first column (n=1),
    # there is only one possible path to reach any cell in that row/column.
    if m == 1 or n == 1:
        return 1
    # Recursive step: To reach cell (m, n), you must come either from (m-1, n) (moving down)
    # or from (m, n-1) (moving right). The total paths are the sum of paths to these two cells.
    return paths(m - 1, n) + paths(m, n - 1)

# Prompt the user to understand the purpose of the paths function.
input("paths(m, n) counts routes through an m x n grid moving only right or down.  Press Enter ")

# Demonstrate the function with 3x3 and 4x4 grids
print("  paths(3, 3) =", paths(3, 3))
print("  paths(4, 4) =", paths(4, 4))

# Ask the user to input a grid size (for both rows and columns, assuming a square grid)
n = int(input("Enter grid size for both rows and cols (try 5 or 6): "))

# Ask the user for their guess for the number of paths
guess = input("What is paths(" + str(n) + ", " + str(n) + ")? ")

# Explain the branching nature of the recursion, leading to rapid growth in computation.
input("each call branches into two — move down or move right — the tree grows fast.  Press Enter ")

# Display the actual result from the paths function and compare it with the user's guess
print("  paths(" + str(n) + ", " + str(n) + ") =", paths(n, n), "  your guess:", guess)
```

```python
keypad = {
    '2': 'ABC',
    '3': 'DEF',
    '4': 'GHI',
    '5': 'JKL',
    '6': 'MNO',
    '7': 'PQRS',
    '8': 'TUV',
    '9': 'WXYZ'
}

def generate_words(number, word=""):
    # Base case
    if number == "":
        print(word)
        return

    # Get letters for the first digit
    letters = keypad[number[0]]

    # Try every letter
    for letter in letters:
        generate_words(number[1:], word + letter)


number = input("Enter a number: ")

print("Possible words:")
generate_words(number)
```
