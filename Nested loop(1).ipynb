```python
# ATM Cash Dispenser Simulation
# This script simulates a basic ATM cash dispenser, handling customer withdrawals
# and providing a daily report of dispensed denominations.

print("=== ATM Cash Dispenser ===\n")

# Initialize variables to keep track of dispensed notes and session statistics
total_100 = total_50 = total_20 = total_10 = total_5 = total_1 = 0 # Counters for each denomination
customers_served = 0  # Total number of customers served in the session
total_dispensed = 0   # Total amount of money dispensed

serving = True # Flag to control the main loop for serving multiple customers
while serving: # outer while loop -- processes one customer per iteration
    name = input("Enter customer name: ") # Get customer's name
    amount = int(input(f"Hello {name}! Enter withdrawal amount: ")) # Get withdrawal amount

    # Validate the withdrawal amount
    if amount <= 0: # Check if the amount is non-positive
        print("Invalid amount. Please enter a positive number.\n")
        continue # Skip to the next customer iteration if invalid

    print(f"\nDispensing {amount} units for {name}:")
    remaining = amount # Amount left to dispense
    idx = 1 # Index to iterate through denominations

    while idx <= 6: # inner while loop -- breaks the total amount into individual notes
        # Determine the current denomination value based on the index
        if idx == 1: value = 100
        elif idx == 2: value = 50
        elif idx == 3: value = 20
        elif idx == 4: value = 10
        elif idx == 5: value = 5
        else: value = 1

        count = remaining // value # Calculate how many notes of the current denomination are needed

        # If notes of this denomination are required, dispense them
        if count > 0:
            print(f"  {count} x {value}-unit note(s) = {count * value}") # Print dispensed notes
            remaining -= count * value # Subtract the dispensed amount from the remaining total

            # Update the total count for the specific denomination
            if value == 100: total_100 += count
            elif value == 50: total_50 += count
            elif value == 20: total_20 += count
            elif value == 10: total_10 += count
            elif value == 5: total_5 += count
            else: total_1 += count
        idx += 1 # Move to the next denomination

    customers_served += 1 # Increment the count of served customers
    total_dispensed += amount # Add the current withdrawal amount to the total dispensed
    print(f"Transaction complete, {name}!\n")

    # Ask if there's another customer
    again = input("Next customer? (yes/no): ").strip().lower() # Get user input
    if again != "yes": # If input is not 'yes', end the serving loop
        serving = False

print("\n=== Daily Denomination Report ===")
# Generate and print a report of all dispensed denominations
for slot in range(1, 7): # outer for loop -- iterates through each possible denomination
    # Assign the correct value and total count for the current denomination slot
    if slot == 1: value, total = 100, total_100
    elif slot == 2: value, total = 50, total_50
    elif slot == 3: value, total = 20, total_20
    elif slot == 4: value, total = 10, total_10
    elif slot == 5: value, total = 5, total_5
    else: value, total = 1, total_1

    # Only print denominations that were actually dispensed
    if total > 0:
        print(f"  {value}-unit notes dispensed : {total} ", end="") # Print count of notes
        for note in range(total): # inner for loop -- prints a visual representation for each note
            print("=", end="") # Print '=' symbol for each note
        print() # Newline after each denomination line

# Print final summary statistics for the session
print(f"\nCustomers served : {customers_served}")
print(f"Total dispensed  : {total_dispensed} units")
print("ATM session closed. Goodbye!")
```

## Grocery Billing Queue Simulation

This script simulates a grocery billing queue, allowing you to process multiple customers, add items to their carts, and generate a final report based on item categories.

```python
# Initialize variables
total_customers_billed = 0
total_revenue = 0.0

print("=== Grocery Billing Queue ===\n")

serving_customers = True # Flag to control the main customer loop
while serving_customers: # Outer while loop for serving multiple customers
    customer_name = input("Enter customer name: ").strip()
    if not customer_name: # Validate customer name
        print("Customer name cannot be empty. Please try again.\n")
        continue

    print(f"\nProcessing order for {customer_name}:")
    customer_total_amount = 0.0

    adding_items = True # Flag to control the item adding loop for the current customer
    while adding_items: # Inner while loop for adding items to the current customer's cart
        item_name = input("  Enter item name (or 'done' to finish): ").strip()
        if item_name.lower() == 'done':
            adding_items = False
            continue
        elif not item_name:
            print("  Item name cannot be empty. Please try again.")
            continue

        # Removed try-except block and validation for positive price
        # Assumes valid numeric input for price, will raise ValueError if not.
        item_price = float(input(f"  Enter price for {item_name}: "))

        # Update customer's total directly
        customer_total_amount += item_price

        print(f"  Added '{item_name}' (Price: {item_price:.2f})\n")

    # Update overall totals if customer actually bought something
    if customer_total_amount > 0:
        total_customers_billed += 1
        total_revenue += customer_total_amount
        print(f"Bill for {customer_name}: {customer_total_amount:.2f} units\n")
    else:
        print(f"No items added for {customer_name}. Skipping this customer.\n")

    # Ask if there's another customer
    another_customer = input("Next customer? (yes/no): ").strip().lower()
    if another_customer != "yes":
        serving_customers = False

print("\n=== End of Billing Session ===\n")

# Final Report Section
```
